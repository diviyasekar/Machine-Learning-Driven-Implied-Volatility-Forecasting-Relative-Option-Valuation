```python
# check quote quality relative spread %

call_mid = (df_19['CE_Bid'] + df_19['CE_Ask']) / 2
put_mid = (df_19['PE_Bid'] + df_19['PE_Ask']) / 2

df_19['CE_Relative_spread'] = ((df_19['CE_Ask'] - df_19['CE_Bid']) / call_mid * 100).where((
    df_19['CE_Bid'] > 0) & (df_19['CE_Ask'] >= df_19['CE_Bid']) & (call_mid > 0))

df_19['PE_Relative_Spread'] = ((df_19['PE_Ask'] - df_19['PE_Bid']) / put_mid * 100).where(
    (df_19['PE_Bid'] > 0) & (df_19['PE_Ask'] >= df_19['PE_Bid']) & (put_mid > 0)
)

print('Call relative spread (%)')
print(df_19['CE_Relative_spread'].describe().round(2))

print('\nPut relative spread(%):')
print(df_19['PE_Relative_Spread'].describe().round(2))
```

```python
# compare candidate spread threshold for 19 th october

thresholds = [10, 20, 30]

for threshold in thresholds:
    call_valid = (df_19['CE_IV'].notna() & df_19['CE_Relative_spread'].notna() & (df_19['CE_Relative_spread'] <= threshold) & (df_19['CE_Volume'].fillna(0)>0) & (df_19['CE_OI'].fillna(0)>0))

    put_valid = (
        df_19['PE_IV'].notna() &
        df_19['PE_Relative_Spread'].notna() &
        (df_19['PE_Relative_Spread'] <= threshold) &
        (df_19['PE_Volume'].fillna(0) > 0) &
        (df_19['PE_OI'].fillna(0) > 0)
    )

    print(f"\nSpread threshold: {threshold}%")
    print("Usable call observations:", call_valid.sum())
    print("Usable put observations:", put_valid.sum())
```

```python
# Store all three datasets together thresholds
datasets = {
    "06 October": df,
    "13 October": df_13,
    "19 October": df_19
}

# Calculate relative spreads and compare candidate filters
thresholds = [10, 20, 30]

for expiry, data in datasets.items():

    # Work on a copy so the original dataframe is preserved
    data = data.copy()

    # Calculate relative spreads
    call_mid = (data['CE_Bid'] + data['CE_Ask']) / 2
    put_mid = (data['PE_Bid'] + data['PE_Ask']) / 2

    data['CE_Relative_Spread'] = (
        (data['CE_Ask'] - data['CE_Bid']) / call_mid * 100
    ).where(
        (data['CE_Bid'] > 0) &
        (data['CE_Ask'] >= data['CE_Bid']) &
        (call_mid > 0)
    )

    data['PE_Relative_Spread'] = (
        (data['PE_Ask'] - data['PE_Bid']) / put_mid * 100
    ).where(
        (data['PE_Bid'] > 0) &
        (data['PE_Ask'] >= data['PE_Bid']) &
        (put_mid > 0)
    )

    print(f"\n{'=' * 15} {expiry} {'=' * 15}")
    print("Total strike rows:", len(data))

    for threshold in thresholds:
        call_valid = (
            data['CE_IV'].notna() &
            data['CE_Relative_Spread'].notna() &
            (data['CE_Relative_Spread'] <= threshold) &
            (data['CE_Volume'] > 0) &
            (data['CE_OI'] > 0)
        )

        put_valid = (
            data['PE_IV'].notna() &
            data['PE_Relative_Spread'].notna() &
            (data['PE_Relative_Spread'] <= threshold) &
            (data['PE_Volume'] > 0) &
            (data['PE_OI'] > 0)
        )

        print(f"\nSpread threshold: {threshold}%")
        print("Usable call observations:", call_valid.sum())
        print("Usable put observations:", put_valid.sum())
```

```python
datasets = {
    "06 October": df,
    "13 October": df_13,
    "19 October": df_19
}

results = []

for expiry, original_data in datasets.items():
    data = original_data.copy()

    call_mid = (data['CE_Bid'] + data['CE_Ask']) / 2
    put_mid = (data['PE_Bid'] + data['PE_Ask']) / 2

    call_spread = (
        (data['CE_Ask'] - data['CE_Bid']) / call_mid * 100
    ).where(
        (data['CE_Bid'] > 0) &
        (data['CE_Ask'] >= data['CE_Bid']) &
        (call_mid > 0)
    )

    put_spread = (
        (data['PE_Ask'] - data['PE_Bid']) / put_mid * 100
    ).where(
        (data['PE_Bid'] > 0) &
        (data['PE_Ask'] >= data['PE_Bid']) &
        (put_mid > 0)
    )

    for threshold in [10, 20, 30]:
        calls = (
            data['CE_IV'].notna() &
            call_spread.notna() &
            (call_spread <= threshold) &
            (data['CE_Volume'] > 0) &
            (data['CE_OI'] > 0)
        )

        puts = (
            data['PE_IV'].notna() &
            put_spread.notna() &
            (put_spread <= threshold) &
            (data['PE_Volume'] > 0) &
            (data['PE_OI'] > 0)
        )

        results.append({
            'Expiry': expiry,
            'Threshold (%)': threshold,
            'Usable Calls': int(calls.sum()),
            'Usable Puts': int(puts.sum())
        })

results_df = pd.DataFrame(results)
print(results_df.to_string(index=False))
```

```python
#

datasets = {
    "06 October": (df, None),
    "13 October": (df_13, 22321.80),
    "19 October": (df_19, 22520.45)
}

results = []

for expiry, (original_data, spot) in datasets.items():

    data = original_data.copy()

    # Use the spot value available for each snapshot.
    # The 06 October spot is left blank until confirmed.
    if spot is None:
        print(f"{expiry}: Please confirm the matching NIFTY spot price.")
        continue

    data["Moneyness"] = data["Strike"] / spot

    # ATM region: strikes within 500 points of spot
    atm = data[
        (data["Strike"] - spot).abs() <= 500
    ].copy()

    call_mid = (atm["CE_Bid"] + atm["CE_Ask"]) / 2
    put_mid = (atm["PE_Bid"] + atm["PE_Ask"]) / 2

    call_spread = (
        (atm["CE_Ask"] - atm["CE_Bid"]) / call_mid * 100
    ).where(
        (atm["CE_Bid"] > 0) &
        (atm["CE_Ask"] >= atm["CE_Bid"]) &
        (call_mid > 0)
    )

    put_spread = (
        (atm["PE_Ask"] - atm["PE_Bid"]) / put_mid * 100
    ).where(
        (atm["PE_Bid"] > 0) &
        (atm["PE_Ask"] >= atm["PE_Bid"]) &
        (put_mid > 0)
    )

    usable_calls = (
        atm["CE_IV"].notna() &
        call_spread.notna() &
        (call_spread <= 20) &
        (atm["CE_Volume"] > 0) &
        (atm["CE_OI"] > 0)
    )

    usable_puts = (
        atm["PE_IV"].notna() &
        put_spread.notna() &
        (put_spread <= 20) &
        (atm["PE_Volume"] > 0) &
        (atm["PE_OI"] > 0)
    )

    results.append({
        "Expiry": expiry,
        "ATM rows": len(atm),
        "Usable ATM calls": int(usable_calls.sum()),
        "Usable ATM puts": int(usable_puts.sum())
    })

print(pd.DataFrame(results).to_string(index=False))
```

```python
# combine the three threshold snapshots for three expiry contracts

datasets = {
    "06 October": (df, 22421.95),
    "13 October": (df_13, 22321.80),
    "19 October": (df_19, 22520.45)
}

cleaned_datasets = {}

for expiry, (original_data, spot) in datasets.items():

    data = original_data.copy()

    # Add expiry label and spot reference
    data["Expiry"] = expiry
    data["Spot"] = spot

    # Calculate moneyness
    data["Moneyness"] = data["Strike"] / spot

    # Calculate relative bid-ask spreads
    call_mid = (data["CE_Bid"] + data["CE_Ask"]) / 2
    put_mid = (data["PE_Bid"] + data["PE_Ask"]) / 2

    data["CE_Relative_Spread"] = (
        (data["CE_Ask"] - data["CE_Bid"]) / call_mid * 100
    ).where(
        (data["CE_Bid"] > 0) &
        (data["CE_Ask"] >= data["CE_Bid"]) &
        (call_mid > 0)
    )

    data["PE_Relative_Spread"] = (
        (data["PE_Ask"] - data["PE_Bid"]) / put_mid * 100
    ).where(
        (data["PE_Bid"] > 0) &
        (data["PE_Ask"] >= data["PE_Bid"]) &
        (put_mid > 0)
    )

    # Keep observations passing the 20% spread and liquidity filters
    call_mask = (
        data["CE_IV"].notna() &
        data["CE_Relative_Spread"].notna() &
        (data["CE_Relative_Spread"] <= 20) &
        (data["CE_Volume"] > 0) &
        (data["CE_OI"] > 0)
    )

    put_mask = (
        data["PE_IV"].notna() &
        data["PE_Relative_Spread"].notna() &
        (data["PE_Relative_Spread"] <= 20) &
        (data["PE_Volume"] > 0) &
        (data["PE_OI"] > 0)
    )

    # Keep calls and puts separately
    calls = data.loc[call_mask].copy()
    puts = data.loc[put_mask].copy()

    calls["Option_Type"] = "Call"
    calls["IV"] = calls["CE_IV"]

    puts["Option_Type"] = "Put"
    puts["IV"] = puts["PE_IV"]

    # Combine both sides
    cleaned = pd.concat(
        [calls, puts],
        ignore_index=True
    )

    cleaned_datasets[expiry] = cleaned

    print(
        expiry,
        "| Calls:", len(calls),
        "| Puts:", len(puts),
        "| Combined:", len(cleaned)
    )

# Combine all three expiries
iv_data = pd.concat(
    cleaned_datasets.values(),
    ignore_index=True
)

print("\nCombined dataset shape:", iv_data.shape)
print("\nSelected columns:")
print(
    iv_data[
        ["Expiry", "Spot", "Strike", "Option_Type",
         "IV", "Moneyness"]
    ].head(10).to_string(index=False)
)
```

```python
# Work on a copy so iv_data remains unchanged
check_data = iv_data.copy()

check_data["Moneyness_Region"] = "Near ATM"

# Calls
is_call = check_data["Option_Type"] == "Call"

check_data.loc[
    is_call & (check_data["Moneyness"] < 0.95),
    "Moneyness_Region"
] = "Far ITM"

check_data.loc[
    is_call &
    (check_data["Moneyness"] >= 0.95) &
    (check_data["Moneyness"] < 0.98),
    "Moneyness_Region"
] = "ITM"

check_data.loc[
    is_call &
    (check_data["Moneyness"] > 1.02) &
    (check_data["Moneyness"] <= 1.05),
    "Moneyness_Region"
] = "OTM"

check_data.loc[
    is_call & (check_data["Moneyness"] > 1.05),
    "Moneyness_Region"
] = "Far OTM"

# Puts: reverse the ITM/OTM direction
is_put = check_data["Option_Type"] == "Put"

check_data.loc[
    is_put & (check_data["Moneyness"] > 1.05),
    "Moneyness_Region"
] = "Far ITM"

check_data.loc[
    is_put &
    (check_data["Moneyness"] > 1.02) &
    (check_data["Moneyness"] <= 1.05),
    "Moneyness_Region"
] = "ITM"

check_data.loc[
    is_put &
    (check_data["Moneyness"] >= 0.95) &
    (check_data["Moneyness"] < 0.98),
    "Moneyness_Region"
] = "OTM"

check_data.loc[
    is_put & (check_data["Moneyness"] < 0.95),
    "Moneyness_Region"
] = "Far OTM"

# Examine observations with IV above 30%
high_iv_corrected = check_data[check_data["IV"] > 30]

print("High-IV observations by correctly classified region:")
print(
    high_iv_corrected.groupby(
        ["Option_Type", "Moneyness_Region"]
    ).size().to_string()
)

print("\nHigh-IV observations near ATM:")
print(
    high_iv_corrected[
        high_iv_corrected["Moneyness_Region"] == "Near ATM"
    ][["Expiry", "Option_Type", "Strike", "IV", "Moneyness"]]
    .to_string(index=False)
)
```

```python
# step 4 establish a baseline model

from sklearn.model_selection import GroupShuffleSplit
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# use the expiry as the group so one expiry stays out of training

splitter = GroupShuffleSplit(n_splits = 1, test_size = 1, random_state = 42)

train_idx, test_idx = next(splitter.split(x_encoded, y, groups = model_data['Expiry']))

x_train = x_encoded.iloc[train_idx]
x_test = x_encoded.iloc[test_idx]

y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]

# Baseline model: predict the training mean IV

baseline = DummyRegressor(strategy = 'mean')
baseline.fit(x_train, y_train)

y_pred_baseline = baseline.predict(x_test)

# Evaluate baseline performance

mae = mean_absolute_error(y_test, y_pred_baseline)
rmse = np.sqrt(mean_squared_error(y_test, y_pred_baseline))
r2 = r2_score(y_test, y_pred_baseline)

print('Training observations:', len(x_train))
print('Testing observations:', len(x_test))

print('\nHeld-out expiry:')
print(model_data.iloc[test_idx]['Expiry'].unique())

print('\nBaseline model results')
print('Training mean IV:', round(y_train.mean(), 2))
print('MAE:', round(mae, 3))
print('RMSE:', round(rmse, 3))
print("R²:", round(r2, 3))
```

```python
#  step 2 create a reusable api fetching function

import os
import requests
from dotenv import load_dotenv

# Load the token saved in your .env file
load_dotenv(".env", override=True)

UPSTOX_TOKEN = os.getenv("UPSTOX_ACCESS_TOKEN")

if not UPSTOX_TOKEN:
    raise ValueError(
        "Token not found. Check that your .env file exists."
    )

def fetch_option_chain(expiry_date):
    """
    Fetch NIFTY 50 option-chain data for one expiry date.
    Example expiry_date: '2026-10-19'
    """

    url = "https://api.upstox.com/v2/option/chain"

    headers = {
        "Accept": "application/json",
        "Authorization": f"Bearer {UPSTOX_TOKEN}"
    }

    params = {
        "instrument_key": "NSE_INDEX|Nifty 50",
        "expiry_date": expiry_date
    }

    result = requests.get(
        url,
        headers=headers,
        params=params,
        timeout=20
    )

    result.raise_for_status()

    payload = result.json()

    if payload.get("status") != "success":
        raise RuntimeError("Upstox did not return a successful response.")

    return payload["data"]

print("Option-chain function is ready.")
```

```python
from datetime import datetime
import pandas as pd

def convert_option_chain(rows):
    timestamp = datetime.now().astimezone().isoformat()
    records = []

    for row in rows:
        call = row.get("call_options") or {}
        put = row.get("put_options") or {}

        call_market = call.get("market_data") or {}
        put_market = put.get("market_data") or {}

        call_greeks = call.get("option_greeks") or {}
        put_greeks = put.get("option_greeks") or {}

        records.append({
            "timestamp": timestamp,
            "expiry": row.get("expiry"),
            "spot": row.get("underlying_spot_price"),
            "strike": row.get("strike_price"),
            "pcr": row.get("pcr"),
            "CE_LTP": call_market.get("ltp"),
            "CE_Volume": call_market.get("volume"),
            "CE_OI": call_market.get("oi"),
            "CE_Bid": call_market.get("bid_price"),
            "CE_Ask": call_market.get("ask_price"),
            "CE_IV": call_greeks.get("iv"),
            "PE_LTP": put_market.get("ltp"),
            "PE_Volume": put_market.get("volume"),
            "PE_OI": put_market.get("oi"),
            "PE_Bid": put_market.get("bid_price"),
            "PE_Ask": put_market.get("ask_price"),
            "PE_IV": put_greeks.get("iv"),
        })

    return pd.DataFrame(records)

print("convert_option_chain is defined.")
```

```python
# step 7 create a cleaned iv dataset

import pandas as pd

df = test_snapshot.copy()

# Treat zero IV as unavailable
df["CE_IV"] = df["CE_IV"].replace(0, float("nan"))
df["PE_IV"] = df["PE_IV"].replace(0, float("nan"))

# Convert calls and puts into one consistent table
common_cols = ["timestamp", "expiry", "spot", "strike", "pcr"]

calls = df[common_cols + [
    "CE_LTP", "CE_Volume", "CE_OI", "CE_Bid", "CE_Ask", "CE_IV"
]].copy()

calls.columns = common_cols + [
    "LTP", "Volume", "OI", "Bid", "Ask", "IV"
]
calls["Option_Type"] = "CE"

puts = df[common_cols + [
    "PE_LTP", "PE_Volume", "PE_OI", "PE_Bid", "PE_Ask", "PE_IV"
]].copy()

puts.columns = common_cols + [
    "LTP", "Volume", "OI", "Bid", "Ask", "IV"
]
puts["Option_Type"] = "PE"

# Combine calls and puts
iv_data = pd.concat([calls, puts], ignore_index=True)

# Keep usable IV targets and active options
iv_data = iv_data[
    iv_data["IV"].notna()
    & (iv_data["Volume"] > 0)
    & (iv_data["OI"] > 0)
].copy()

# Calculate moneyness
iv_data["Moneyness"] = iv_data["strike"] / iv_data["spot"]

print("Cleaned dataset shape:", iv_data.shape)
print("\nOption types:")
print(iv_data["Option_Type"].value_counts())

display(iv_data.head())
```

```python
# check the bid ask spread quality

# Step 8: Check quote quality

iv_data["Mid_Price"] = (iv_data["Bid"] + iv_data["Ask"]) / 2

iv_data["Relative_Spread"] = (
    (iv_data["Ask"] - iv_data["Bid"]) / iv_data["Mid_Price"]
)

# Identify quotes that may be unreliable
iv_data["Quote_Check"] = "Review"

valid_quotes = (
    (iv_data["Bid"] > 0)
    & (iv_data["Ask"] > 0)
    & (iv_data["Ask"] >= iv_data["Bid"])
    & (iv_data["Mid_Price"] > 0)
)

iv_data.loc[valid_quotes, "Quote_Check"] = "Valid quote"

print("Total observations:", len(iv_data))
print("Valid bid-ask quotes:", valid_quotes.sum())
print("Quotes needing review:", (~valid_quotes).sum())

print("\nRelative spread summary:")
display(iv_data["Relative_Spread"].describe())

print("\nLargest relative spreads:")
display(
    iv_data.sort_values("Relative_Spread", ascending=False)[
        ["Option_Type", "strike", "IV", "Bid", "Ask", "Relative_Spread"]
    ].head(10)
)
```

```python
# Filter wide spreads

# Step 9: Apply an initial spread-quality filter

spread_threshold = 0.20

iv_data["Spread_Quality"] = (
    iv_data["Relative_Spread"] <= spread_threshold
)

filtered_data = iv_data[iv_data["Spread_Quality"]].copy()

print("Original observations:", len(iv_data))
print("Retained observations:", len(filtered_data))
print("Removed for wide spreads:", len(iv_data) - len(filtered_data))

print("\nObservations by option type:")
display(filtered_data["Option_Type"].value_counts())

print("\nRemaining spread summary:")
display(filtered_data["Relative_Spread"].describe())
```

```python
# prepare the automated collector

import time
import pandas as pd
from pathlib import Path
from datetime import datetime
from zoneinfo import ZoneInfo

# Collection settings
INTERVAL_SECONDS = 5 * 60
TARGET_SNAPSHOTS = 1000

RAW_FOLDER = Path("data/raw")
RAW_FOLDER.mkdir(parents=True, exist_ok=True)

IST = ZoneInfo("Asia/Kolkata")


def collect_one_snapshot(expiry_date):
    """Fetch and save one option-chain snapshot."""

    collected_at = datetime.now(IST)

    # Fetch option-chain data using your existing function
    rows = fetch_option_chain(expiry_date)

    if not rows:
        raise ValueError("The API returned no option-chain rows.")

    # Convert the response using your existing conversion function
    snapshot = convert_option_chain(rows)

    # Use the actual collection time for the saved snapshot
    snapshot["timestamp"] = collected_at.isoformat()

    # Save a uniquely named CSV file
    filename_time = collected_at.strftime("%Y%m%d_%H%M%S")
    output_file = RAW_FOLDER / f"nifty_options_{filename_time}.csv"

    snapshot.to_csv(output_file, index=False)

    return {
        "timestamp": collected_at.isoformat(),
        "rows": len(snapshot),
        "file": str(output_file),
    }


print("Collector function defined successfully.")
print("No API request has been made.")
print("Output folder:", RAW_FOLDER.resolve())
```

```python
# Step 3C — Add market-hours and expiry safeguards

import time
from datetime import datetime, time as dt_time
from zoneinfo import ZoneInfo
from pathlib import Path

IST = ZoneInfo("Asia/Kolkata")

INTERVAL_SECONDS = 300
TARGET_SNAPSHOTS = 1000
RAW_FOLDER = Path("data/raw")
RAW_FOLDER.mkdir(parents=True, exist_ok=True)


def is_market_open():
    now = datetime.now(IST)
    return (
        now.weekday() < 5
        and dt_time(9, 15) <= now.time().replace(tzinfo=None)
        < dt_time(15, 30)
    )


def run_collector(expiry_date, target=TARGET_SNAPSHOTS):
    successful = 0

    print("Collector started. Press Ctrl+C to stop safely.")

    try:
        while successful < target:
            now = datetime.now(IST)

            if not is_market_open():
                print("Outside market hours. Collector stopped.")
                break

            try:
                result = collect_one_snapshot(expiry_date)
                successful += 1

                print(
                    f"{successful}/{target} | "
                    f"{result['timestamp']} | "
                    f"Rows: {result['rows']} | "
                    f"Saved: {result['file']}"
                )

            except Exception as exc:
                print(f"Collection error at {now.isoformat()}: {exc}")

            if successful < target:
                time.sleep(INTERVAL_SECONDS)

    except KeyboardInterrupt:
        print("Stopped by user. Saved snapshots are preserved.")

    print("Successful snapshots this run:", successful)
```

```python
# Step 5 — Add duplicate protection

from pathlib import Path
import pandas as pd

RAW_FOLDER = Path("data/raw")
RAW_FOLDER.mkdir(parents=True, exist_ok=True)


def snapshot_already_saved(timestamp, expiry_date):
    """Check whether a snapshot for this timestamp and expiry already exists."""
    for file in RAW_FOLDER.glob("nifty_options_*.csv"):
        try:
            existing = pd.read_csv(
                file, usecols=["timestamp", "expiry"]
            )

            if (
                existing["timestamp"].eq(timestamp).any()
                and existing["expiry"].eq(expiry_date).any()
            ):
                return True

        except (ValueError, OSError, pd.errors.ParserError):
            continue

    return False


print("Duplicate-check function defined.")
print("Raw folder:", RAW_FOLDER.resolve())
```

```python
from datetime import datetime
from zoneinfo import ZoneInfo
from pathlib import Path

IST = ZoneInfo("Asia/Kolkata")
RAW_FOLDER = Path("data/raw")
RAW_FOLDER.mkdir(parents=True, exist_ok=True)

def collect_one_snapshot(expiry_date):
    collected_at = datetime.now(IST)

    # Only collect during regular weekday market hours.
    market_time = collected_at.time().replace(tzinfo=None)
    if (
        collected_at.weekday() >= 5
        or not (datetime.strptime("09:15", "%H:%M").time()
                <= market_time
                < datetime.strptime("15:30", "%H:%M").time())
    ):
        raise RuntimeError("Market is closed; no API request was made.")

    # Fetch the option chain.
    rows = fetch_option_chain(expiry_date)
    if not rows:
        raise ValueError("No option-chain rows returned.")

    snapshot = convert_option_chain(rows)
    snapshot["timestamp"] = collected_at.isoformat()

    # Use microseconds to reduce the risk of filename collisions.
    filename_time = collected_at.strftime("%Y%m%d_%H%M%S_%f")
    output_file = RAW_FOLDER / f"nifty_options_{filename_time}.csv"

    if output_file.exists():
        raise FileExistsError(f"Refusing to overwrite {output_file}")

    snapshot.to_csv(output_file, index=False)

    # Confirm the saved file is readable and has the expected row count.
    saved = pd.read_csv(output_file)
    if len(saved) != len(snapshot):
        output_file.unlink(missing_ok=True)
        raise IOError("Saved row count does not match the collected snapshot.")

    log_collection(
        collected_at.isoformat(),
        expiry_date,
        len(snapshot),
        output_file.name,
        "SUCCESS"
    )

    return {
        "timestamp": collected_at.isoformat(),
        "rows": len(snapshot),
        "file": str(output_file)
    }

print("Safer collector function defined.")
print("No API request was made.")
```

```python
import time
import pandas as pd
from datetime import datetime, time as dt_time, timedelta
from zoneinfo import ZoneInfo
from pathlib import Path

IST = ZoneInfo("Asia/Kolkata")
RAW_FOLDER = Path("data/raw")
LOG_FILE = RAW_FOLDER / "collection_log.csv"

INTERVAL_SECONDS = 300
TARGET_SNAPSHOTS = 60
EXPIRY_DATE = "2026-10-19"


def market_is_open():
    now = datetime.now(IST)
    current_time = now.time().replace(tzinfo=None)

    return (
        now.weekday() < 5
        and dt_time(9, 15) <= current_time < dt_time(15, 30)
    )


def successful_snapshot_count():
    if not LOG_FILE.exists():
        return 0

    log = pd.read_csv(LOG_FILE)

    if log.empty:
        return 0

    successful = log[log["status"].eq("SUCCESS")]
    return successful["filename"].nunique()


def run_automatic_collector(target=TARGET_SNAPSHOTS):
    print("Automatic collector started.")
    print("Target successful snapshots:", target)
    print("Press Ctrl+C to stop safely.")

    next_collection = time.monotonic()

    try:
        while successful_snapshot_count() < target:
            if not market_is_open():
                print("Market is closed. Collector paused.")
                print("Stop this run and restart it during market hours.")
                break

            wait_seconds = next_collection - time.monotonic()

            if wait_seconds > 0:
                time.sleep(wait_seconds)

            if not market_is_open():
                print("Market closed before next collection.")
                break

            try:
                result = collect_one_snapshot(EXPIRY_DATE)

                count = successful_snapshot_count()

                print(
                    f"{count}/{target} | "
                    f"{result['timestamp']} | "
                    f"Rows: {result['rows']} | "
                    f"Saved: {result['file']}"
                )

            except KeyboardInterrupt:
                raise

            except Exception as exc:
                print(
                    f"Collection failed at "
                    f"{datetime.now(IST).isoformat()}: "
                    f"{type(exc).__name__}: {exc}"
                )

            next_collection = time.monotonic() + INTERVAL_SECONDS

    except KeyboardInterrupt:
        print("Stopped by user. Existing files are preserved.")

    print("Successful snapshots recorded:", successful_snapshot_count())
    print("Collector finished.")


print("Collector function defined.")
print("No API request was made.")
```

```python
import os
import requests
from dotenv import load_dotenv

load_dotenv(".env", override=True)

UPSTOX_TOKEN = os.getenv("UPSTOX_ACCESS_TOKEN")

if not UPSTOX_TOKEN:
    raise ValueError(
        "Token not found. Check that your .env file exists "
        "in the current working directory."
    )

def fetch_option_chain(expiry_date):
    url = "https://api.upstox.com/v2/option/chain"

    headers = {
        "Accept": "application/json",
        "Authorization": f"Bearer {UPSTOX_TOKEN}"
    }

    params = {
        "instrument_key": "NSE_INDEX|Nifty 50",
        "expiry_date": expiry_date
    }

    response = requests.get(
        url,
        headers=headers,
        params=params,
        timeout=30
    )
    response.raise_for_status()

    payload = response.json()

    if payload.get("status") != "success":
        raise RuntimeError(
            "Upstox did not return a successful response."
        )

    return payload["data"]

print("fetch_option_chain() is ready.")
```

```python
import os
import csv
import time
import requests
import pandas as pd

from pathlib import Path
from datetime import datetime, time as dt_time
from zoneinfo import ZoneInfo
from dotenv import load_dotenv


# =====================================================
# 1. CONFIGURATION
# =====================================================

IST = ZoneInfo("Asia/Kolkata")
EXPIRY_DATE = "2026-10-19"
TARGET_NEW_SNAPSHOTS = 60
INTERVAL_SECONDS = 300

BASE_DIR = Path.home() / "Downloads"
RAW_FOLDER = BASE_DIR / "data" / "raw"
RAW_FOLDER.mkdir(parents=True, exist_ok=True)

LOG_FILE = RAW_FOLDER / "collection_log.csv"
ENV_FILE = BASE_DIR / ".env"

load_dotenv(ENV_FILE, override=True)
UPSTOX_TOKEN = os.getenv("UPSTOX_ACCESS_TOKEN")

if not UPSTOX_TOKEN:
    raise ValueError(
        "Upstox token not found. Check the UPSTOX_ACCESS_TOKEN "
        "entry in your Downloads/.env file."
    )


# =====================================================
# 2. LOGGING
# =====================================================

LOG_COLUMNS = ["timestamp", "expiry", "rows", "filename", "status"]

if not LOG_FILE.exists() or LOG_FILE.stat().st_size == 0:
    with open(LOG_FILE, "w", newline="", encoding="utf-8") as f:
        csv.writer(f).writerow(LOG_COLUMNS)


def log_collection(timestamp, expiry, rows, filename, status):
    with open(LOG_FILE, "a", newline="", encoding="utf-8") as f:
        csv.writer(f).writerow(
            [timestamp, expiry, rows, filename, status]
        )


def successful_snapshot_count():
    if not LOG_FILE.exists() or LOG_FILE.stat().st_size == 0:
        return 0

    log = pd.read_csv(LOG_FILE)

    required = {"status", "filename"}
    if not required.issubset(log.columns):
        raise ValueError(
            "Collection log columns are incorrect. "
            "Existing log has been preserved."
        )

    successful = log[
        log["status"].astype(str).str.upper().eq("SUCCESS")
    ]

    return successful["filename"].dropna().nunique()


# =====================================================
# 3. API FUNCTION
# =====================================================

def fetch_option_chain(expiry_date):
    url = "https://api.upstox.com/v2/option/chain"

    headers = {
        "Accept": "application/json",
        "Authorization": f"Bearer {UPSTOX_TOKEN}"
    }

    params = {
        "instrument_key": "NSE_INDEX|Nifty 50",
        "expiry_date": expiry_date
    }

    response = requests.get(
        url,
        headers=headers,
        params=params,
        timeout=30
    )
    response.raise_for_status()

    payload = response.json()

    if payload.get("status") != "success":
        raise RuntimeError("Upstox returned an unsuccessful response.")

    return payload["data"]


# =====================================================
# 4. CONVERSION FUNCTION — FIXES YOUR NameError
# =====================================================

def convert_option_chain(rows):
    timestamp = datetime.now(IST).isoformat()
    records = []

    for row in rows:
        call = row.get("call_options") or {}
        put = row.get("put_options") or {}

        call_market = call.get("market_data") or {}
        put_market = put.get("market_data") or {}

        call_greeks = call.get("option_greeks") or {}
        put_greeks = put.get("option_greeks") or {}

        records.append({
            "timestamp": timestamp,
            "expiry": row.get("expiry"),
            "spot": row.get("underlying_spot_price"),
            "strike": row.get("strike_price"),
            "pcr": row.get("pcr"),

            "CE_LTP": call_market.get("ltp"),
            "CE_Volume": call_market.get("volume"),
            "CE_OI": call_market.get("oi"),
            "CE_Bid": call_market.get("bid_price"),
            "CE_Ask": call_market.get("ask_price"),
            "CE_IV": call_greeks.get("iv"),

            "PE_LTP": put_market.get("ltp"),
            "PE_Volume": put_market.get("volume"),
            "PE_OI": put_market.get("oi"),
            "PE_Bid": put_market.get("bid_price"),
            "PE_Ask": put_market.get("ask_price"),
            "PE_IV": put_greeks.get("iv")
        })

    return pd.DataFrame(records)


# =====================================================
# 5. MARKET-HOURS CHECK
# =====================================================

def market_is_open():
    now = datetime.now(IST)
    current_time = now.time().replace(tzinfo=None)

    return (
        now.weekday() < 5
        and dt_time(9, 15) <= current_time < dt_time(15, 30)
    )


# =====================================================
# 6. COLLECT AND SAVE ONE SNAPSHOT
# =====================================================

def collect_one_snapshot(expiry_date):
    collected_at = datetime.now(IST)

    if not market_is_open():
        raise RuntimeError("Market is closed.")

    rows = fetch_option_chain(expiry_date)

    if not rows:
        raise ValueError("The API returned no option-chain rows.")

    snapshot = convert_option_chain(rows)
    snapshot["timestamp"] = collected_at.isoformat()

    filename = (
        "nifty_options_"
        + collected_at.strftime("%Y%m%d_%H%M%S_%f")
        + ".csv"
    )

    output_file = RAW_FOLDER / filename

    if output_file.exists():
        raise FileExistsError(
            "Refusing to overwrite an existing snapshot."
        )

    snapshot.to_csv(output_file, index=False)

    try:
        saved = pd.read_csv(output_file)

        if len(saved) != len(snapshot):
            raise IOError("Saved row count does not match collected rows.")

        log_collection(
            collected_at.isoformat(),
            expiry_date,
            len(snapshot),
            filename,
            "SUCCESS"
        )

    except Exception:
        output_file.unlink(missing_ok=True)
        raise

    return {
        "timestamp": collected_at.isoformat(),
        "rows": len(snapshot),
        "file": str(output_file)
    }


# =====================================================
# 7. COLLECTER — 60 ADDITIONAL SUCCESSFUL SNAPSHOTS
# =====================================================

def run_corrected_collector(target=TARGET_NEW_SNAPSHOTS):
    starting_count = successful_snapshot_count()
    target_total = starting_count + target

    print("Starting successful count:", starting_count)
    print("Additional snapshots required:", target)
    print("Target total:", target_total)
    print("Expiry:", EXPIRY_DATE)
    print("Output folder:", RAW_FOLDER)
    print("Press Ctrl+C to stop safely.\n")

    try:
        while successful_snapshot_count() < target_total:

            if not market_is_open():
                print("Market closed. Collection stopped.")
                break

            # Wait for the next 5-minute clock boundary.
            now = datetime.now(IST)
            seconds_into_interval = (
                (now.minute % 5) * 60
                + now.second
                + now.microsecond / 1_000_000
            )

            wait_seconds = INTERVAL_SECONDS - seconds_into_interval
            if wait_seconds > 0:
                time.sleep(wait_seconds)

            if not market_is_open():
                print("Market closed before the next collection.")
                break

            try:
                result = collect_one_snapshot(EXPIRY_DATE)
                total = successful_snapshot_count()

                print(
                    f"SUCCESS | Total: {total}/{target_total} | "
                    f"Rows: {result['rows']} | "
                    f"Time: {result['timestamp']} | "
                    f"File: {result['file']}"
                )

            except KeyboardInterrupt:
                raise

            except Exception as exc:
                error_time = datetime.now(IST).isoformat()

                log_collection(
                    error_time,
                    EXPIRY_DATE,
                    0,
                    "",
                    "ERROR"
                )

                print(
                    f"Collection error: {type(exc).__name__}: {exc}"
                )

        print(
            "\nFinal successful snapshot count:",
            successful_snapshot_count()
        )

    except KeyboardInterrupt:
        print("\nStopped by user. Existing data is preserved.")
        print(
            "Successful snapshot count:",
            successful_snapshot_count()
        )


# =====================================================
# 8. VERIFY AND START
# =====================================================

print("Conversion function:", callable(convert_option_chain))
print("API function:", callable(fetch_option_chain))
print("Snapshot collector:", callable(collect_one_snapshot))
print("Log file:", LOG_FILE)

if market_is_open():
    run_corrected_collector()
else:
    print("Market is closed. No collection was started.")
```

```python
# ============================================================
# NIFTY 50 OPTIONS DATA COLLECTOR
# Target: 81 successful snapshots in total
# Interval: 5 minutes
# Expiry: 19 October 2026
# ============================================================

import os
import csv
import time
import requests
import pandas as pd

from pathlib import Path
from datetime import datetime, time as dt_time
from zoneinfo import ZoneInfo
from dotenv import load_dotenv


# ------------------------------------------------------------
# 1. CONFIGURATION
# ------------------------------------------------------------

IST = ZoneInfo("Asia/Kolkata")

EXPIRY_DATE = "2026-10-19"
TARGET_TOTAL = 81
INTERVAL_SECONDS = 300

BASE_DIR = Path.home() / "Downloads"
RAW_FOLDER = BASE_DIR / "data" / "raw"
LOG_FILE = RAW_FOLDER / "collection_log.csv"
ENV_FILE = BASE_DIR / ".env"

RAW_FOLDER.mkdir(parents=True, exist_ok=True)

# Load the existing token. Never print or share it.
load_dotenv(ENV_FILE, override=True)
UPSTOX_TOKEN = os.getenv("UPSTOX_ACCESS_TOKEN")

if not UPSTOX_TOKEN:
    raise ValueError(
        "Upstox token not found. Check the UPSTOX_ACCESS_TOKEN "
        "entry in C:\\Users\\DIVIYA\\Downloads\\.env"
    )


# ------------------------------------------------------------
# 2. COLLECTION LOG
# ------------------------------------------------------------

LOG_COLUMNS = [
    "timestamp",
    "expiry",
    "rows",
    "filename",
    "status"
]

if not LOG_FILE.exists() or LOG_FILE.stat().st_size == 0:
    with open(LOG_FILE, "w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerow(LOG_COLUMNS)

# Confirm the existing log has the expected structure.
log_header = pd.read_csv(LOG_FILE, nrows=0).columns.tolist()

if log_header != LOG_COLUMNS:
    raise ValueError(
        f"Unexpected collection log columns: {log_header}. "
        "The existing log was not overwritten."
    )


def log_collection(timestamp, expiry, rows, filename, status):
    with open(LOG_FILE, "a", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerow([
            timestamp,
            expiry,
            rows,
            filename,
            status
        ])


def successful_snapshot_count():
    if not LOG_FILE.exists() or LOG_FILE.stat().st_size == 0:
        return 0

    log = pd.read_csv(LOG_FILE)

    if not {"status", "filename"}.issubset(log.columns):
        raise ValueError("Collection log is missing required columns.")

    successful = log[
        log["status"].astype(str).str.upper().eq("SUCCESS")
    ]

    # Count unique successful CSV filenames, not individual strike rows.
    return successful["filename"].dropna().nunique()


# ------------------------------------------------------------
# 3. FETCH OPTION CHAIN FROM UPSTOX
# ------------------------------------------------------------

def fetch_option_chain(expiry_date):
    url = "https://api.upstox.com/v2/option/chain"

    headers = {
        "Accept": "application/json",
        "Authorization": f"Bearer {UPSTOX_TOKEN}"
    }

    params = {
        "instrument_key": "NSE_INDEX|Nifty 50",
        "expiry_date": expiry_date
    }

    response = requests.get(
        url,
        headers=headers,
        params=params,
        timeout=30
    )

    response.raise_for_status()
    payload = response.json()

    if payload.get("status") != "success":
        raise RuntimeError(
            "Upstox did not return a successful option-chain response."
        )

    return payload["data"]


# ------------------------------------------------------------
# 4. CONVERT API RESPONSE INTO A DATAFRAME
# ------------------------------------------------------------

def convert_option_chain(rows):
    timestamp = datetime.now(IST).isoformat()
    records = []

    for row in rows:
        call = row.get("call_options") or {}
        put = row.get("put_options") or {}

        call_market = call.get("market_data") or {}
        put_market = put.get("market_data") or {}

        call_greeks = call.get("option_greeks") or {}
        put_greeks = put.get("option_greeks") or {}

        records.append({
            "timestamp": timestamp,
            "expiry": row.get("expiry"),
            "spot": row.get("underlying_spot_price"),
            "strike": row.get("strike_price"),
            "pcr": row.get("pcr"),

            "CE_LTP": call_market.get("ltp"),
            "CE_Volume": call_market.get("volume"),
            "CE_OI": call_market.get("oi"),
            "CE_Bid": call_market.get("bid_price"),
            "CE_Ask": call_market.get("ask_price"),
            "CE_IV": call_greeks.get("iv"),

            "PE_LTP": put_market.get("ltp"),
            "PE_Volume": put_market.get("volume"),
            "PE_OI": put_market.get("oi"),
            "PE_Bid": put_market.get("bid_price"),
            "PE_Ask": put_market.get("ask_price"),
            "PE_IV": put_greeks.get("iv")
        })

    return pd.DataFrame(records)


# ------------------------------------------------------------
# 5. MARKET HOURS
# ------------------------------------------------------------

def market_is_open():
    now = datetime.now(IST)
    current_time = now.time().replace(tzinfo=None)

    return (
        now.weekday() < 5
        and dt_time(9, 15) <= current_time < dt_time(15, 30)
    )


# ------------------------------------------------------------
# 6. COLLECT AND SAVE ONE SNAPSHOT
# ------------------------------------------------------------

def collect_one_snapshot(expiry_date):
    collected_at = datetime.now(IST)

    if not market_is_open():
        raise RuntimeError("Market is closed.")

    rows = fetch_option_chain(expiry_date)

    if not rows:
        raise ValueError("The option-chain API returned no rows.")

    snapshot = convert_option_chain(rows)

    # Use the snapshot collection timestamp consistently.
    snapshot["timestamp"] = collected_at.isoformat()

    filename = (
        "nifty_options_"
        + collected_at.strftime("%Y%m%d_%H%M%S_%f")
        + ".csv"
    )

    output_file = RAW_FOLDER / filename

    # Never overwrite an existing file.
    if output_file.exists():
        raise FileExistsError(
            f"Refusing to overwrite existing file: {output_file}"
        )

    snapshot.to_csv(output_file, index=False)

    try:
        saved = pd.read_csv(output_file)

        if len(saved) != len(snapshot):
            raise IOError("Saved row count does not match collected row count.")

        log_collection(
            collected_at.isoformat(),
            expiry_date,
            len(snapshot),
            filename,
            "SUCCESS"
        )

    except Exception:
        output_file.unlink(missing_ok=True)
        raise

    return {
        "timestamp": collected_at.isoformat(),
        "rows": len(snapshot),
        "file": str(output_file)
    }


# ------------------------------------------------------------
# 7. COLLECT UNTIL THE TOTAL REACHES 81
# ------------------------------------------------------------

def run_collector_until_total(target_total=TARGET_TOTAL):

    starting_count = successful_snapshot_count()
    remaining = max(0, target_total - starting_count)

    print("=" * 60)
    print("NIFTY 50 OPTION-CHAIN COLLECTOR")
    print("=" * 60)
    print("Current successful snapshots:", starting_count)
    print("Target total:", target_total)
    print("Additional snapshots required:", remaining)
    print("Expiry:", EXPIRY_DATE)
    print("Interval: 5 minutes")
    print("Save folder:", RAW_FOLDER)
    print("Log file:", LOG_FILE)
    print("=" * 60)

    if remaining == 0:
        print("Target already reached. Nothing to collect.")
        return

    now = datetime.now(IST)

    # If started before 09:15 on a weekday, wait for market opening.
    if now.weekday() < 5 and now.time().replace(tzinfo=None) < dt_time(9, 15):
        market_open_time = now.replace(
            hour=9, minute=15, second=0, microsecond=0
        )

        wait_until_open = (
            market_open_time - now
        ).total_seconds()

        print("Waiting for the market to open at 09:15 IST...")
        time.sleep(max(0, wait_until_open))

    try:
        while successful_snapshot_count() < target_total:

            now = datetime.now(IST)
            current_time = now.time().replace(tzinfo=None)

            if now.weekday() >= 5:
                print("Weekend detected. Collection stopped.")
                break

            if current_time >= dt_time(15, 30):
                print("Market closed. Collection stopped.")
                break

            if current_time < dt_time(9, 15):
                print("Waiting for market opening...")
                time.sleep(5)
                continue

            # Align collection to the next 5-minute boundary.
            now = datetime.now(IST)

            seconds_into_interval = (
                (now.minute % 5) * 60
                + now.second
                + now.microsecond / 1_000_000
            )

            wait_seconds = (
                INTERVAL_SECONDS - seconds_into_interval
            ) % INTERVAL_SECONDS

            # If already on a boundary, collect without an extra 5-minute wait.
            if wait_seconds > 1:
                time.sleep(wait_seconds)

            if not market_is_open():
                print("Market closed before collection. Stopping.")
                break

            try:
                result = collect_one_snapshot(EXPIRY_DATE)
                current_count = successful_snapshot_count()

                print(
                    f"SUCCESS | Total: {current_count}/{target_total} | "
                    f"Rows: {result['rows']} | "
                    f"Time: {result['timestamp']} | "
                    f"File: {result['file']}"
                )

            except Exception as exc:
                error_time = datetime.now(IST).isoformat()

                # Record the failure without increasing the success count.
                log_collection(
                    error_time,
                    EXPIRY_DATE,
                    0,
                    "",
                    "ERROR"
                )

                print(
                    f"Collection error: {type(exc).__name__}: {exc}"
                )

        print("\nCollection session ended.")
        print(
            "Final successful snapshot count:",
            successful_snapshot_count()
        )
        print("Existing CSV files have been preserved.")

    except KeyboardInterrupt:
        print("\nCollection interrupted by user.")
        print(
            "Successful snapshots recorded:",
            successful_snapshot_count()
        )
        print("Existing CSV files have been preserved.")


# ------------------------------------------------------------
# 8. START COLLECTION
# ------------------------------------------------------------

print("Conversion function:", callable(convert_option_chain))
print("API function:", callable(fetch_option_chain))
print("Snapshot collector:", callable(collect_one_snapshot))

run_collector_until_total(target_total=81)
```

```python
from pathlib import Path
import pandas as pd

# ============================================================
# STEP 1: COMBINE ALL SUCCESSFUL SNAPSHOTS
# ============================================================

BASE_DIR = Path.home() / "Downloads"
RAW_FOLDER = BASE_DIR / "data" / "raw"
PROCESSED_FOLDER = BASE_DIR / "data" / "processed"
LOG_FILE = RAW_FOLDER / "collection_log.csv"

PROCESSED_FOLDER.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 1. READ COLLECTION LOG
# ------------------------------------------------------------

log = pd.read_csv(LOG_FILE)

print("=" * 60)
print("COLLECTION LOG")
print("=" * 60)

print("Total log entries:", len(log))

# Keep only successful collections
successful = log[
    log["status"].astype(str).str.upper().eq("SUCCESS")
].copy()

# Remove blank filenames
successful = successful[
    successful["filename"].notna()
].copy()

# Keep only actual NIFTY option snapshot files
successful = successful[
    successful["filename"]
    .astype(str)
    .str.startswith("nifty_options_")
].copy()

# Avoid loading the same file twice
successful = successful.drop_duplicates(
    subset="filename"
)

print("Successful unique snapshots:", len(successful))

# ------------------------------------------------------------
# 2. LOAD ALL SNAPSHOT FILES
# ------------------------------------------------------------

frames = []
missing_files = []
bad_files = []

for filename in successful["filename"]:

    file_path = RAW_FOLDER / filename

    if not file_path.exists():
        missing_files.append(filename)
        continue

    try:
        df = pd.read_csv(file_path)

        if df.empty:
            bad_files.append(filename)
            continue

        df["source_file"] = filename

        frames.append(df)

    except Exception as e:
        print("Problem reading:", filename)
        print(e)

        bad_files.append(filename)

# ------------------------------------------------------------
# 3. COMBINE DATA
# ------------------------------------------------------------

if len(frames) == 0:
    raise ValueError("No valid snapshot files were found.")

master = pd.concat(
    frames,
    ignore_index=True
)

# ------------------------------------------------------------
# 4. FIX DATA TYPES
# ------------------------------------------------------------

master["timestamp"] = pd.to_datetime(
    master["timestamp"],
    errors="coerce",
    utc=True
)

numeric_columns = [
    "spot",
    "strike",
    "pcr",
    "CE_LTP",
    "CE_Volume",
    "CE_OI",
    "CE_Bid",
    "CE_Ask",
    "CE_IV",
    "PE_LTP",
    "PE_Volume",
    "PE_OI",
    "PE_Bid",
    "PE_Ask",
    "PE_IV"
]

for col in numeric_columns:
    if col in master.columns:
        master[col] = pd.to_numeric(
            master[col],
            errors="coerce"
        )

# ------------------------------------------------------------
# 5. SORT DATA
# ------------------------------------------------------------

master = master.sort_values(
    ["timestamp", "strike"]
).reset_index(drop=True)

# ------------------------------------------------------------
# 6. SAVE MASTER DATASET
# ------------------------------------------------------------

MASTER_FILE = (
    PROCESSED_FOLDER /
    "nifty_options_master.csv"
)

master.to_csv(
    MASTER_FILE,
    index=False
)

# ------------------------------------------------------------
# 7. DATA QUALITY REPORT
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("MASTER DATASET REPORT")
print("=" * 60)

print("Successful snapshots in log:", len(successful))
print("Files successfully loaded:", len(frames))
print("Missing files:", len(missing_files))
print("Bad/empty files:", len(bad_files))

print("\nTotal rows:", len(master))

print(
    "Unique timestamps:",
    master["timestamp"].nunique()
)

print(
    "Unique strikes:",
    master["strike"].nunique()
)

print(
    "Earliest timestamp:",
    master["timestamp"].min()
)

print(
    "Latest timestamp:",
    master["timestamp"].max()
)

print("\nExpiry distribution:")

if "expiry" in master.columns:
    print(
        master["expiry"]
        .value_counts(dropna=False)
    )

# ------------------------------------------------------------
# 8. MISSING VALUES
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("MISSING VALUES")
print("=" * 60)

important_columns = [
    "spot",
    "strike",
    "CE_IV",
    "PE_IV",
    "CE_LTP",
    "PE_LTP",
    "CE_Volume",
    "PE_Volume",
    "CE_OI",
    "PE_OI"
]

for col in important_columns:

    if col in master.columns:

        missing = master[col].isna().sum()

        percentage = (
            master[col].isna().mean() * 100
        )

        print(
            f"{col:<12} "
            f"Missing: {missing:<6} "
            f"({percentage:.2f}%)"
        )

# ------------------------------------------------------------
# 9. DUPLICATE CHECK
# ------------------------------------------------------------

duplicates = master.duplicated(
    subset=["timestamp", "strike"]
).sum()

print("\nDuplicate timestamp + strike rows:",
      duplicates)

# ------------------------------------------------------------
# 10. IV RANGE
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("IMPLIED VOLATILITY SUMMARY")
print("=" * 60)

print("\nCALL IV")
print(master["CE_IV"].describe())

print("\nPUT IV")
print(master["PE_IV"].describe())

# ------------------------------------------------------------
# 11. SPOT RANGE
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("NIFTY SPOT")
print("=" * 60)

print(master["spot"].describe())

# ------------------------------------------------------------
# 12. FINAL MESSAGE
# ------------------------------------------------------------

print("\n" + "=" * 60)

print("MASTER DATASET SAVED SUCCESSFULLY")

print("Location:")
print(MASTER_FILE)

print("=" * 60)
```

```python
# prepare ml data set

import pandas as pd
import numpy as np
from pathlib import Path

# cleaning

BASE_DIR = Path.home() / "Downloads"
PROCESSED_FOLDER = BASE_DIR / "data" / "processed"

MASTER_FILE = PROCESSED_FOLDER / "nifty_options_master.csv"

# ------------------------------------------------------------
# 1. LOAD MASTER DATA
# ------------------------------------------------------------

df = pd.read_csv(MASTER_FILE)

df["timestamp"] = pd.to_datetime(
    df["timestamp"],
    errors="coerce",
    utc=True
)

df["expiry"] = pd.to_datetime(
    df["expiry"],
    errors="coerce"
)

print("Original master shape:", df.shape)

# ------------------------------------------------------------
# 2. CREATE CALL DATASET
# ------------------------------------------------------------

calls = df[
    [
        "timestamp",
        "expiry",
        "spot",
        "strike",
        "pcr",
        "CE_LTP",
        "CE_Volume",
        "CE_OI",
        "CE_Bid",
        "CE_Ask",
        "CE_IV"
    ]
].copy()

calls = calls.rename(
    columns={
        "CE_LTP": "LTP",
        "CE_Volume": "Volume",
        "CE_OI": "OI",
        "CE_Bid": "Bid",
        "CE_Ask": "Ask",
        "CE_IV": "IV"
    }
)

calls["option_type"] = "CE"

# ------------------------------------------------------------
# 3. CREATE PUT DATASET
# ------------------------------------------------------------

puts = df[
    [
        "timestamp",
        "expiry",
        "spot",
        "strike",
        "pcr",
        "PE_LTP",
        "PE_Volume",
        "PE_OI",
        "PE_Bid",
        "PE_Ask",
        "PE_IV"
    ]
].copy()

puts = puts.rename(
    columns={
        "PE_LTP": "LTP",
        "PE_Volume": "Volume",
        "PE_OI": "OI",
        "PE_Bid": "Bid",
        "PE_Ask": "Ask",
        "PE_IV": "IV"
    }
)

puts["option_type"] = "PE"

# ------------------------------------------------------------
# 4. COMBINE CALLS AND PUTS
# ------------------------------------------------------------

options = pd.concat(
    [calls, puts],
    ignore_index=True
)

print("After CE + PE restructuring:", options.shape)

# ------------------------------------------------------------
# 5. CONVERT NUMERIC COLUMNS
# ------------------------------------------------------------

numeric_columns = [
    "spot",
    "strike",
    "pcr",
    "LTP",
    "Volume",
    "OI",
    "Bid",
    "Ask",
    "IV"
]

for col in numeric_columns:
    options[col] = pd.to_numeric(
        options[col],
        errors="coerce"
    )

# ------------------------------------------------------------
# 6. REMOVE ROWS WITHOUT ESSENTIAL INFORMATION
# ------------------------------------------------------------

before = len(options)

options = options.dropna(
    subset=[
        "timestamp",
        "spot",
        "strike",
        "IV"
    ]
).copy()

print(
    "Rows removed because essential data was missing:",
    before - len(options)
)

# IV must be positive
options = options[
    options["IV"] > 0
].copy()

# ------------------------------------------------------------
# 7. CREATE MONEYNESS
# ------------------------------------------------------------

options["moneyness"] = (
    options["strike"] /
    options["spot"]
)

# Distance of strike from ATM
options["distance_from_ATM"] = abs(
    options["strike"] -
    options["spot"]
)

# Percentage distance from ATM
options["distance_from_ATM_pct"] = (
    options["distance_from_ATM"] /
    options["spot"]
) * 100

# ------------------------------------------------------------
# 8. CREATE BID-ASK SPREAD
# ------------------------------------------------------------

options["bid_ask_spread"] = (
    options["Ask"] -
    options["Bid"]
)

options["mid_price"] = (
    options["Ask"] +
    options["Bid"]
) / 2

options["relative_spread_pct"] = np.where(
    options["mid_price"] > 0,
    (
        options["bid_ask_spread"] /
        options["mid_price"]
    ) * 100,
    np.nan
)

# ------------------------------------------------------------
# 9. CREATE TIME-TO-EXPIRY
# ------------------------------------------------------------

# Convert timestamp to Indian time
options["timestamp_ist"] = (
    options["timestamp"]
    .dt.tz_convert("Asia/Kolkata")
)

# Days remaining until expiry
timestamp_naive = (
    options["timestamp_ist"]
    .dt.tz_localize(None)
)

options["days_to_expiry"] = (
    options["expiry"] -
    timestamp_naive.dt.normalize()
).dt.total_seconds() / 86400

# ------------------------------------------------------------
# 10. OPTION TYPE AS NUMERIC VARIABLE
# ------------------------------------------------------------

options["option_type_code"] = (
    options["option_type"]
    .map({
        "CE": 0,
        "PE": 1
    })
)

# ------------------------------------------------------------
# 11. SORT CORRECTLY
# ------------------------------------------------------------

options = options.sort_values(
    [
        "option_type",
        "strike",
        "timestamp"
    ]
).reset_index(drop=True)

# ------------------------------------------------------------
# 12. BASIC QUALITY CHECK
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("ML DATASET PREPARATION REPORT")
print("=" * 60)

print("Final rows:", len(options))

print(
    "Unique timestamps:",
    options["timestamp"].nunique()
)

print(
    "Unique strikes:",
    options["strike"].nunique()
)

print("\nOption type counts:")
print(
    options["option_type"]
    .value_counts()
)

print("\nMoneyness summary:")
print(
    options["moneyness"]
    .describe()
)

print("\nIV summary:")
print(
    options["IV"]
    .describe()
)

print("\nMissing values:")
print(
    options[
        [
            "IV",
            "LTP",
            "Volume",
            "OI",
            "Bid",
            "Ask",
            "moneyness",
            "relative_spread_pct"
        ]
    ]
    .isna()
    .sum()
)

# ------------------------------------------------------------
# 13. SAVE
# ------------------------------------------------------------

OUTPUT_FILE = (
    PROCESSED_FOLDER /
    "nifty_options_features.csv"
)

options.to_csv(
    OUTPUT_FILE,
    index=False
)

print("\nDataset saved:")
print(OUTPUT_FILE)
```

```python
# ============================================================
# STEP 3: CREATE EXACT 5-MINUTE-AHEAD IV TARGET
# ============================================================

import pandas as pd
from pathlib import Path

# ------------------------------------------------------------
# 1. LOAD FEATURE DATASET
# ------------------------------------------------------------

BASE_DIR = Path.home() / "Downloads"
PROCESSED_FOLDER = BASE_DIR / "data" / "processed"

FEATURE_FILE = (
    PROCESSED_FOLDER /
    "nifty_options_features.csv"
)

options = pd.read_csv(FEATURE_FILE)

options["timestamp"] = pd.to_datetime(
    options["timestamp"],
    errors="coerce",
    utc=True
)

print("Original feature rows:", len(options))

# ------------------------------------------------------------
# 2. REMOVE DUPLICATE CONTRACT-TIME OBSERVATIONS
# ------------------------------------------------------------

before = len(options)

options = options.drop_duplicates(
    subset=[
        "timestamp",
        "expiry",
        "strike",
        "option_type"
    ],
    keep="last"
).copy()

print(
    "Duplicate contract-time rows removed:",
    before - len(options)
)

# ------------------------------------------------------------
# 3. CREATE THE REQUIRED FUTURE TIMESTAMP
# ------------------------------------------------------------

# Example:
# 10:00 -> target time = 10:05
# 10:05 -> target time = 10:10

options["target_timestamp"] = (
    options["timestamp"]
    + pd.Timedelta(minutes=5)
)

# ------------------------------------------------------------
# 4. CREATE LOOKUP TABLE FOR FUTURE IV
# ------------------------------------------------------------

future = options[
    [
        "timestamp",
        "expiry",
        "strike",
        "option_type",
        "IV"
    ]
].copy()

future = future.rename(
    columns={
        "timestamp": "target_timestamp",
        "IV": "IV_t_plus_5"
    }
)

# ------------------------------------------------------------
# 5. MATCH SAME OPTION EXACTLY 5 MINUTES LATER
# ------------------------------------------------------------

ml = options.merge(
    future,
    on=[
        "target_timestamp",
        "expiry",
        "strike",
        "option_type"
    ],
    how="left"
)

# ------------------------------------------------------------
# 6. CHECK HOW MANY EXACT MATCHES EXIST
# ------------------------------------------------------------

matched = ml["IV_t_plus_5"].notna().sum()
unmatched = ml["IV_t_plus_5"].isna().sum()

print("\nExact 5-minute target matches:", matched)
print("Rows without exact 5-minute target:", unmatched)

# ------------------------------------------------------------
# 7. KEEP ONLY ROWS WITH A REAL 5-MINUTE TARGET
# ------------------------------------------------------------

ml = ml.dropna(
    subset=["IV_t_plus_5"]
).copy()

# ------------------------------------------------------------
# 8. CREATE IV CHANGE
# ------------------------------------------------------------

ml["IV_change_5m"] = (
    ml["IV_t_plus_5"]
    - ml["IV"]
)

ml["IV_change_pct_5m"] = (
    (
        ml["IV_t_plus_5"]
        - ml["IV"]
    )
    / ml["IV"]
) * 100

# ------------------------------------------------------------
# 9. VERIFY THAT TARGET REALLY IS 5 MINUTES AHEAD
# ------------------------------------------------------------

ml["time_difference_minutes"] = (
    (
        ml["target_timestamp"]
        - ml["timestamp"]
    )
    .dt.total_seconds()
    / 60
)

print("\nTime difference check:")

print(
    ml["time_difference_minutes"]
    .value_counts()
    .sort_index()
)

# ------------------------------------------------------------
# 10. CHECK FOR ACCIDENTAL OVERNIGHT MATCHING
# ------------------------------------------------------------

current_date = (
    ml["timestamp"]
    .dt.tz_convert("Asia/Kolkata")
    .dt.date
)

target_date = (
    ml["target_timestamp"]
    .dt.tz_convert("Asia/Kolkata")
    .dt.date
)

overnight_matches = (
    current_date != target_date
).sum()

print(
    "\nAccidental overnight matches:",
    overnight_matches
)

# ------------------------------------------------------------
# 11. SORT CHRONOLOGICALLY
# ------------------------------------------------------------

ml = ml.sort_values(
    [
        "timestamp",
        "strike",
        "option_type"
    ]
).reset_index(drop=True)

# ------------------------------------------------------------
# 12. REPORT
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("5-MINUTE TARGET REPORT")
print("=" * 60)

print("ML observations:", len(ml))

print(
    "Unique input timestamps:",
    ml["timestamp"].nunique()
)

print(
    "Unique target timestamps:",
    ml["target_timestamp"].nunique()
)

print("\nOption types:")

print(
    ml["option_type"]
    .value_counts()
)

print("\nCurrent IV:")
print(
    ml["IV"]
    .describe()
)

print("\n5-minute future IV:")
print(
    ml["IV_t_plus_5"]
    .describe()
)

print("\n5-minute IV change:")
print(
    ml["IV_change_5m"]
    .describe()
)

# ------------------------------------------------------------
# 13. SHOW SAMPLE
# ------------------------------------------------------------

print("\nSample target matching:")

print(
    ml[
        [
            "timestamp",
            "target_timestamp",
            "strike",
            "option_type",
            "IV",
            "IV_t_plus_5",
            "IV_change_5m"
        ]
    ].head(10).to_string(index=False)
)

# ------------------------------------------------------------
# 14. SAVE ML DATASET
# ------------------------------------------------------------

OUTPUT_FILE = (
    PROCESSED_FOLDER /
    "nifty_options_ml_5min.csv"
)

ml.to_csv(
    OUTPUT_FILE,
    index=False
)

print("\n" + "=" * 60)

print("5-MINUTE ML DATASET SAVED")

print(OUTPUT_FILE)

print("=" * 60)
```

```python
# ============================================================
# STEP 4: CHRONOLOGICAL TRAIN/TEST SPLIT + BASELINE MODEL
# ============================================================

import pandas as pd
import numpy as np

from pathlib import Path

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ------------------------------------------------------------
# 1. LOAD 5-MINUTE ML DATASET
# ------------------------------------------------------------

BASE_DIR = Path.home() / "Downloads"
PROCESSED_FOLDER = BASE_DIR / "data" / "processed"

ML_FILE = (
    PROCESSED_FOLDER /
    "nifty_options_ml_5min.csv"
)

df = pd.read_csv(ML_FILE)

df["time_5m"] = pd.to_datetime(
    df["time_5m"],
    errors="coerce",
    utc=True
)

df["target_time_5m"] = pd.to_datetime(
    df["target_time_5m"],
    errors="coerce",
    utc=True
)

df = df.sort_values(
    "time_5m"
).reset_index(drop=True)

print("Total ML observations:", len(df))
print("Unique timestamps:", df["time_5m"].nunique())


# ------------------------------------------------------------
# 2. GET UNIQUE TIMES
# ------------------------------------------------------------

unique_times = (
    df["time_5m"]
    .drop_duplicates()
    .sort_values()
    .reset_index(drop=True)
)

print("Total unique times:", len(unique_times))


# ------------------------------------------------------------
# 3. CHRONOLOGICAL 80/20 SPLIT
# ------------------------------------------------------------

split_position = int(
    len(unique_times) * 0.80
)

train_times = unique_times.iloc[
    :split_position
]

test_times = unique_times.iloc[
    split_position:
]

train = df[
    df["time_5m"].isin(train_times)
].copy()

test = df[
    df["time_5m"].isin(test_times)
].copy()


# ------------------------------------------------------------
# 4. CHECK SPLIT
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("CHRONOLOGICAL SPLIT")
print("=" * 60)

print("Training rows:", len(train))
print("Testing rows:", len(test))

print(
    "Training timestamps:",
    train["time_5m"].nunique()
)

print(
    "Testing timestamps:",
    test["time_5m"].nunique()
)

print("\nTraining period:")

print(
    train["time_5m"].min(),
    "to",
    train["time_5m"].max()
)

print("\nTesting period:")

print(
    test["time_5m"].min(),
    "to",
    test["time_5m"].max()
)


# ------------------------------------------------------------
# 5. CHECK FOR TIME LEAKAGE
# ------------------------------------------------------------

overlap = set(
    train["time_5m"]
).intersection(
    set(test["time_5m"])
)

print(
    "\nOverlapping train/test timestamps:",
    len(overlap)
)


# ------------------------------------------------------------
# 6. BASELINE FORECAST
# ------------------------------------------------------------

# Baseline assumption:
#
# Future IV = Current IV

test["baseline_prediction"] = test["IV"]


# ------------------------------------------------------------
# 7. ACTUAL AND PREDICTED VALUES
# ------------------------------------------------------------

y_true = test["IV_t_plus_5"]

y_pred_baseline = test[
    "baseline_prediction"
]


# ------------------------------------------------------------
# 8. CALCULATE BASELINE PERFORMANCE
# ------------------------------------------------------------

baseline_mae = mean_absolute_error(
    y_true,
    y_pred_baseline
)

baseline_mse = mean_squared_error(
    y_true,
    y_pred_baseline
)

baseline_rmse = np.sqrt(
    baseline_mse
)

baseline_r2 = r2_score(
    y_true,
    y_pred_baseline
)


# ------------------------------------------------------------
# 9. PRINT RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("NAIVE BASELINE PERFORMANCE")
print("=" * 60)

print(
    f"MAE  : {baseline_mae:.4f}"
)

print(
    f"MSE  : {baseline_mse:.4f}"
)

print(
    f"RMSE : {baseline_rmse:.4f}"
)

print(
    f"R²   : {baseline_r2:.4f}"
)


# ------------------------------------------------------------
# 10. BASELINE ERROR
# ------------------------------------------------------------

test["baseline_error"] = (
    test["IV_t_plus_5"]
    - test["baseline_prediction"]
)

print("\nBaseline error summary:")

print(
    test["baseline_error"]
    .describe()
)


# ------------------------------------------------------------
# 11. PERFORMANCE BY OPTION TYPE
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("BASELINE PERFORMANCE BY OPTION TYPE")
print("=" * 60)

for option_type in ["CE", "PE"]:

    subset = test[
        test["option_type"] == option_type
    ]

    if len(subset) == 0:
        continue

    actual = subset[
        "IV_t_plus_5"
    ]

    prediction = subset[
        "IV"
    ]

    mae = mean_absolute_error(
        actual,
        prediction
    )

    rmse = np.sqrt(
        mean_squared_error(
            actual,
            prediction
        )
    )

    r2 = r2_score(
        actual,
        prediction
    )

    print(f"\n{option_type}")

    print(
        f"Observations : {len(subset)}"
    )

    print(
        f"MAE          : {mae:.4f}"
    )

    print(
        f"RMSE         : {rmse:.4f}"
    )

    print(
        f"R²           : {r2:.4f}"
    )


# ------------------------------------------------------------
# 12. SAVE TRAIN AND TEST DATA
# ------------------------------------------------------------

TRAIN_FILE = (
    PROCESSED_FOLDER /
    "nifty_iv_train.csv"
)

TEST_FILE = (
    PROCESSED_FOLDER /
    "nifty_iv_test.csv"
)

train.to_csv(
    TRAIN_FILE,
    index=False
)

test.to_csv(
    TEST_FILE,
    index=False
)

print("\n" + "=" * 60)

print("TRAIN/TEST DATA SAVED")

print("\nTraining:")
print(TRAIN_FILE)

print("\nTesting:")
print(TEST_FILE)

print("=" * 60)
```

```python
# ============================================================
# STEP 5: LINEAR REGRESSION MODEL
# Predict 5-minute-ahead Implied Volatility
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ------------------------------------------------------------
# 1. LOAD TRAIN AND TEST DATA
# ------------------------------------------------------------

BASE_DIR = Path.home() / "Downloads"
PROCESSED_FOLDER = BASE_DIR / "data" / "processed"

TRAIN_FILE = PROCESSED_FOLDER / "nifty_iv_train.csv"
TEST_FILE = PROCESSED_FOLDER / "nifty_iv_test.csv"

train = pd.read_csv(TRAIN_FILE)
test = pd.read_csv(TEST_FILE)

print("Training rows:", len(train))
print("Testing rows:", len(test))


# ------------------------------------------------------------
# 2. SELECT FEATURES
# ------------------------------------------------------------

features = [
    "IV",
    "spot",
    "strike",
    "moneyness",
    "distance_from_ATM_pct",
    "LTP",
    "Volume",
    "OI",
    "Bid",
    "Ask",
    "bid_ask_spread",
    "relative_spread_pct",
    "pcr",
    "days_to_expiry",
    "option_type_code"
]

target = "IV_t_plus_5"

print("\nNumber of features:", len(features))
print("\nFeatures used:")

for feature in features:
    print("-", feature)


# ------------------------------------------------------------
# 3. CHECK THAT WE ARE NOT USING FUTURE INFORMATION
# ------------------------------------------------------------

forbidden_features = [
    "IV_t_plus_5",
    "IV_change_5m",
    "IV_change_pct_5m",
    "target_time_5m",
    "baseline_error"
]

leakage = [
    col for col in forbidden_features
    if col in features
]

print("\nFuture-information features accidentally included:",
      leakage)

if leakage:
    raise ValueError(
        "STOP: Future information is being used as a feature."
    )


# ------------------------------------------------------------
# 4. CREATE X AND Y
# ------------------------------------------------------------

X_train = train[features].copy()
y_train = train[target].copy()

X_test = test[features].copy()
y_test = test[target].copy()


# ------------------------------------------------------------
# 5. HANDLE ANY REMAINING MISSING VALUES
# ------------------------------------------------------------

# Use training medians only.
# This avoids using information from the test period.

train_medians = X_train.median(numeric_only=True)

X_train = X_train.fillna(train_medians)
X_test = X_test.fillna(train_medians)

print("\nMissing values after preparation:")

print("Training:", X_train.isna().sum().sum())
print("Testing :", X_test.isna().sum().sum())


# ------------------------------------------------------------
# 6. TRAIN LINEAR REGRESSION
# ------------------------------------------------------------

linear_model = LinearRegression()

linear_model.fit(
    X_train,
    y_train
)


# ------------------------------------------------------------
# 7. PREDICT FUTURE IV
# ------------------------------------------------------------

linear_prediction = linear_model.predict(
    X_test
)


# ------------------------------------------------------------
# 8. CALCULATE LINEAR REGRESSION METRICS
# ------------------------------------------------------------

linear_mae = mean_absolute_error(
    y_test,
    linear_prediction
)

linear_mse = mean_squared_error(
    y_test,
    linear_prediction
)

linear_rmse = np.sqrt(
    linear_mse
)

linear_r2 = r2_score(
    y_test,
    linear_prediction
)


# ------------------------------------------------------------
# 9. CALCULATE BASELINE AGAIN
# ------------------------------------------------------------

baseline_prediction = test["IV"]

baseline_mae = mean_absolute_error(
    y_test,
    baseline_prediction
)

baseline_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        baseline_prediction
    )
)

baseline_r2 = r2_score(
    y_test,
    baseline_prediction
)


# ------------------------------------------------------------
# 10. COMPARE BASELINE VS LINEAR REGRESSION
# ------------------------------------------------------------

comparison = pd.DataFrame({
    "Model": [
        "Naive Baseline",
        "Linear Regression"
    ],

    "MAE": [
        baseline_mae,
        linear_mae
    ],

    "RMSE": [
        baseline_rmse,
        linear_rmse
    ],

    "R2": [
        baseline_r2,
        linear_r2
    ]
})

print("\n" + "=" * 60)
print("MODEL COMPARISON")
print("=" * 60)

print(
    comparison.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# 11. IMPROVEMENT OVER BASELINE
# ------------------------------------------------------------

mae_improvement = (
    (baseline_mae - linear_mae)
    / baseline_mae
) * 100

rmse_improvement = (
    (baseline_rmse - linear_rmse)
    / baseline_rmse
) * 100

print("\n" + "=" * 60)
print("IMPROVEMENT OVER BASELINE")
print("=" * 60)

print(
    f"MAE improvement  : {mae_improvement:.2f}%"
)

print(
    f"RMSE improvement : {rmse_improvement:.2f}%"
)


# ------------------------------------------------------------
# 12. PERFORMANCE BY OPTION TYPE
# ------------------------------------------------------------

results = test.copy()

results["linear_prediction"] = linear_prediction

print("\n" + "=" * 60)
print("LINEAR REGRESSION BY OPTION TYPE")
print("=" * 60)

for option_type in ["CE", "PE"]:

    subset = results[
        results["option_type"] == option_type
    ]

    if len(subset) == 0:
        continue

    actual = subset["IV_t_plus_5"]

    predicted = subset["linear_prediction"]

    mae = mean_absolute_error(
        actual,
        predicted
    )

    rmse = np.sqrt(
        mean_squared_error(
            actual,
            predicted
        )
    )

    r2 = r2_score(
        actual,
        predicted
    )

    print(f"\n{option_type}")

    print(
        f"Observations : {len(subset)}"
    )

    print(
        f"MAE          : {mae:.4f}"
    )

    print(
        f"RMSE         : {rmse:.4f}"
    )

    print(
        f"R²           : {r2:.4f}"
    )


# ------------------------------------------------------------
# 13. FEATURE COEFFICIENTS
# ------------------------------------------------------------

coefficients = pd.DataFrame({
    "Feature": features,
    "Coefficient": linear_model.coef_
})

coefficients["Absolute_Coefficient"] = (
    coefficients["Coefficient"].abs()
)

coefficients = coefficients.sort_values(
    "Absolute_Coefficient",
    ascending=False
)

print("\n" + "=" * 60)
print("LINEAR REGRESSION COEFFICIENTS")
print("=" * 60)

print(
    coefficients[
        ["Feature", "Coefficient"]
    ].to_string(index=False)
)


# ------------------------------------------------------------
# 14. SAVE PREDICTIONS
# ------------------------------------------------------------

RESULT_FILE = (
    PROCESSED_FOLDER /
    "linear_regression_predictions.csv"
)

results.to_csv(
    RESULT_FILE,
    index=False
)

print("\nPredictions saved:")
print(RESULT_FILE)
```

```python
# ============================================================
# STEP 6: RANDOM FOREST REGRESSION
# 5-MINUTE-AHEAD IV FORECAST
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ------------------------------------------------------------
# 1. LOAD TRAIN / TEST DATA
# ------------------------------------------------------------

BASE_DIR = Path.home() / "Downloads"
PROCESSED_FOLDER = BASE_DIR / "data" / "processed"

TRAIN_FILE = PROCESSED_FOLDER / "nifty_iv_train.csv"
TEST_FILE = PROCESSED_FOLDER / "nifty_iv_test.csv"

train = pd.read_csv(TRAIN_FILE)
test = pd.read_csv(TEST_FILE)

print("Training rows:", len(train))
print("Testing rows :", len(test))


# ------------------------------------------------------------
# 2. FEATURES
# ------------------------------------------------------------

features = [
    "IV",
    "spot",
    "strike",
    "moneyness",
    "distance_from_ATM_pct",
    "LTP",
    "Volume",
    "OI",
    "Bid",
    "Ask",
    "bid_ask_spread",
    "relative_spread_pct",
    "pcr",
    "days_to_expiry",
    "option_type_code"
]

target = "IV_t_plus_5"


# ------------------------------------------------------------
# 3. SAFETY CHECK — NO FUTURE INFORMATION
# ------------------------------------------------------------

forbidden = [
    "IV_t_plus_5",
    "IV_change_5m",
    "IV_change_pct_5m",
    "target_time_5m",
    "baseline_error"
]

leakage = [
    x for x in forbidden
    if x in features
]

print("\nFuture-information features included:", leakage)

if leakage:
    raise ValueError("Future information detected.")


# ------------------------------------------------------------
# 4. PREPARE TRAIN / TEST
# ------------------------------------------------------------

X_train = train[features].copy()
X_test = test[features].copy()

y_train = train[target].copy()
y_test = test[target].copy()

# Training medians only
medians = X_train.median(numeric_only=True)

X_train = X_train.fillna(medians)
X_test = X_test.fillna(medians)

print(
    "Missing training values:",
    X_train.isna().sum().sum()
)

print(
    "Missing testing values:",
    X_test.isna().sum().sum()
)


# ------------------------------------------------------------
# 5. CREATE RANDOM FOREST
# ------------------------------------------------------------

rf_model = RandomForestRegressor(
    n_estimators=300,
    max_depth=12,
    min_samples_leaf=3,
    max_features="sqrt",
    random_state=42,
    n_jobs=-1
)


# ------------------------------------------------------------
# 6. TRAIN
# ------------------------------------------------------------

print("\nTraining Random Forest...")

rf_model.fit(
    X_train,
    y_train
)

print("Training completed.")


# ------------------------------------------------------------
# 7. PREDICT
# ------------------------------------------------------------

rf_prediction = rf_model.predict(
    X_test
)


# ------------------------------------------------------------
# 8. RANDOM FOREST METRICS
# ------------------------------------------------------------

rf_mae = mean_absolute_error(
    y_test,
    rf_prediction
)

rf_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        rf_prediction
    )
)

rf_r2 = r2_score(
    y_test,
    rf_prediction
)


# ------------------------------------------------------------
# 9. BASELINE METRICS
# ------------------------------------------------------------

baseline_prediction = test["IV"]

baseline_mae = mean_absolute_error(
    y_test,
    baseline_prediction
)

baseline_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        baseline_prediction
    )
)

baseline_r2 = r2_score(
    y_test,
    baseline_prediction
)


# ------------------------------------------------------------
# 10. LINEAR REGRESSION RESULTS
# From our previous model
# ------------------------------------------------------------

linear_mae = 6.169280
linear_rmse = 8.144062
linear_r2 = 0.806990


# ------------------------------------------------------------
# 11. MODEL COMPARISON
# ------------------------------------------------------------

comparison = pd.DataFrame({

    "Model": [
        "Naive Baseline",
        "Linear Regression",
        "Random Forest"
    ],

    "MAE": [
        baseline_mae,
        linear_mae,
        rf_mae
    ],

    "RMSE": [
        baseline_rmse,
        linear_rmse,
        rf_rmse
    ],

    "R2": [
        baseline_r2,
        linear_r2,
        rf_r2
    ]
})

print("\n" + "=" * 65)
print("MODEL COMPARISON")
print("=" * 65)

print(
    comparison.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# 12. IMPROVEMENT OVER BASELINE
# ------------------------------------------------------------

mae_improvement = (
    (baseline_mae - rf_mae)
    / baseline_mae
) * 100

rmse_improvement = (
    (baseline_rmse - rf_rmse)
    / baseline_rmse
) * 100

print("\n" + "=" * 65)
print("RANDOM FOREST VS BASELINE")
print("=" * 65)

print(
    f"MAE improvement  : {mae_improvement:.2f}%"
)

print(
    f"RMSE improvement : {rmse_improvement:.2f}%"
)


# ------------------------------------------------------------
# 13. RESULTS DATAFRAME
# ------------------------------------------------------------

results = test.copy()

results["RF_prediction"] = rf_prediction

results["RF_error"] = (
    results["IV_t_plus_5"]
    - results["RF_prediction"]
)


# ------------------------------------------------------------
# 14. PERFORMANCE BY OPTION TYPE
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("RANDOM FOREST PERFORMANCE BY OPTION TYPE")
print("=" * 65)

for option_type in ["CE", "PE"]:

    subset = results[
        results["option_type"] == option_type
    ]

    if len(subset) == 0:
        continue

    actual = subset["IV_t_plus_5"]

    prediction = subset["RF_prediction"]

    mae = mean_absolute_error(
        actual,
        prediction
    )

    rmse = np.sqrt(
        mean_squared_error(
            actual,
            prediction
        )
    )

    r2 = r2_score(
        actual,
        prediction
    )

    print(f"\n{option_type}")

    print(
        f"Observations : {len(subset)}"
    )

    print(
        f"MAE          : {mae:.4f}"
    )

    print(
        f"RMSE         : {rmse:.4f}"
    )

    print(
        f"R²           : {r2:.4f}"
    )


# ------------------------------------------------------------
# 15. FEATURE IMPORTANCE
# ------------------------------------------------------------

importance = pd.DataFrame({

    "Feature": features,

    "Importance":
        rf_model.feature_importances_

})

importance = importance.sort_values(
    "Importance",
    ascending=False
)

print("\n" + "=" * 65)
print("RANDOM FOREST FEATURE IMPORTANCE")
print("=" * 65)

print(
    importance.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# 16. SAVE RESULTS
# ------------------------------------------------------------

RESULT_FILE = (
    PROCESSED_FOLDER /
    "random_forest_predictions.csv"
)

results.to_csv(
    RESULT_FILE,
    index=False
)

print("\nPredictions saved:")
print(RESULT_FILE)
```

```python
# ============================================================
# STEP 9
# RANDOM FOREST MODEL FOR 5-MINUTE IV CHANGE
# ============================================================

import pandas as pd
import numpy as np

from pathlib import Path

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ============================================================
# 1. LOAD NEAR-ATM DATA
# ============================================================

BASE_DIR = Path.home() / "Downloads"

PROCESSED_FOLDER = (
    BASE_DIR /
    "data" /
    "processed"
)

FILE = (
    PROCESSED_FOLDER /
    "nifty_options_ml_5min_near_atm.csv"
)

df = pd.read_csv(FILE)


# Convert timestamp
df["time_5m"] = pd.to_datetime(
    df["time_5m"],
    utc=True
)


print("=" * 70)
print("DATASET")
print("=" * 70)

print("Rows:", len(df))

print(
    "Unique timestamps:",
    df["time_5m"].nunique()
)


# ============================================================
# 2. FEATURES
# ============================================================

features = [

    "IV",

    "spot",

    "strike",

    "moneyness",

    "distance_from_ATM_pct",

    "LTP",

    "Volume",

    "OI",

    "Bid",

    "Ask",

    "bid_ask_spread",

    "relative_spread_pct",

    "pcr",

    "days_to_expiry",

    "option_type_code"
]


# ============================================================
# 3. TARGET
# ============================================================

# Instead of predicting future IV directly,
# we predict the CHANGE in IV.

target = "IV_change_5m"


print("\nTarget:")
print(target)


# ============================================================
# 4. CHRONOLOGICAL 80/20 SPLIT
# ============================================================

unique_times = np.array(
    sorted(
        df["time_5m"].unique()
    )
)

split_index = int(
    len(unique_times) * 0.80
)

train_times = (
    unique_times[:split_index]
)

test_times = (
    unique_times[split_index:]
)


train = df[
    df["time_5m"].isin(
        train_times
    )
].copy()


test = df[
    df["time_5m"].isin(
        test_times
    )
].copy()


print("\n" + "=" * 70)
print("CHRONOLOGICAL SPLIT")
print("=" * 70)

print(
    "Train rows:",
    len(train)
)

print(
    "Test rows:",
    len(test)
)

print(
    "Train timestamps:",
    train["time_5m"].nunique()
)

print(
    "Test timestamps:",
    test["time_5m"].nunique()
)

print(
    "\nTrain period:"
)

print(
    train["time_5m"].min(),
    "to",
    train["time_5m"].max()
)

print(
    "\nTest period:"
)

print(
    test["time_5m"].min(),
    "to",
    test["time_5m"].max()
)


# ============================================================
# 5. X AND y
# ============================================================

X_train = train[features].copy()

X_test = test[features].copy()

y_train = train[target].copy()

y_test = test[target].copy()


# ============================================================
# 6. HANDLE MISSING VALUES
# ============================================================

# Median must be calculated ONLY from training data.

train_medians = (
    X_train.median()
)

X_train = (
    X_train.fillna(
        train_medians
    )
)

X_test = (
    X_test.fillna(
        train_medians
    )
)


print("\nMissing X_train:")
print(
    X_train.isna().sum().sum()
)

print(
    "Missing X_test:",
    X_test.isna().sum().sum()
)


# ============================================================
# 7. BASELINE
# ============================================================

# Persistence says:
#
# Future IV = Current IV
#
# Therefore:
#
# Predicted change in IV = 0

baseline_delta = np.zeros(
    len(test)
)

baseline_future_iv = (
    test["IV"].values
)


# ============================================================
# 8. RANDOM FOREST
# ============================================================

rf_delta = RandomForestRegressor(

    n_estimators=300,

    max_depth=12,

    min_samples_leaf=3,

    max_features="sqrt",

    random_state=42,

    n_jobs=-1
)


rf_delta.fit(
    X_train,
    y_train
)


# ============================================================
# 9. PREDICT IV CHANGE
# ============================================================

predicted_delta = (
    rf_delta.predict(
        X_test
    )
)


# ============================================================
# 10. RECONSTRUCT FUTURE IV
# ============================================================

predicted_future_iv = (

    test["IV"].values

    + predicted_delta

)


actual_future_iv = (
    test["IV_t_plus_5"].values
)


# ============================================================
# 11. EVALUATE BASELINE
# ============================================================

baseline_mae = mean_absolute_error(
    actual_future_iv,
    baseline_future_iv
)

baseline_mse = mean_squared_error(
    actual_future_iv,
    baseline_future_iv
)

baseline_rmse = np.sqrt(
    baseline_mse
)

baseline_r2 = r2_score(
    actual_future_iv,
    baseline_future_iv
)


# ============================================================
# 12. EVALUATE RANDOM FOREST ΔIV MODEL
# ============================================================

rf_mae = mean_absolute_error(
    actual_future_iv,
    predicted_future_iv
)

rf_mse = mean_squared_error(
    actual_future_iv,
    predicted_future_iv
)

rf_rmse = np.sqrt(
    rf_mse
)

rf_r2 = r2_score(
    actual_future_iv,
    predicted_future_iv
)


# ============================================================
# 13. RESULTS
# ============================================================

print("\n" + "=" * 70)

print(
    "5-MINUTE IV FORECAST RESULTS"
)

print("=" * 70)


print("\nPERSISTENCE BASELINE")

print(
    f"MAE  : {baseline_mae:.6f}"
)

print(
    f"RMSE : {baseline_rmse:.6f}"
)

print(
    f"R²   : {baseline_r2:.6f}"
)


print("\nRANDOM FOREST ΔIV MODEL")

print(
    f"MAE  : {rf_mae:.6f}"
)

print(
    f"RMSE : {rf_rmse:.6f}"
)

print(
    f"R²   : {rf_r2:.6f}"
)


# ============================================================
# 14. IMPROVEMENT OVER BASELINE
# ============================================================

mae_improvement = (

    (
        baseline_mae
        - rf_mae
    )

    / baseline_mae

    * 100
)


rmse_improvement = (

    (
        baseline_rmse
        - rf_rmse
    )

    / baseline_rmse

    * 100
)


print(
    "\nImprovement over persistence:"
)

print(
    f"MAE improvement  : "
    f"{mae_improvement:.2f}%"
)

print(
    f"RMSE improvement : "
    f"{rmse_improvement:.2f}%"
)


# ============================================================
# 15. DELTA PREDICTION QUALITY
# ============================================================

delta_mae = mean_absolute_error(
    y_test,
    predicted_delta
)

delta_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        predicted_delta
    )
)


print("\n" + "=" * 70)

print(
    "IV CHANGE PREDICTION"
)

print("=" * 70)

print(
    f"ΔIV MAE  : {delta_mae:.6f}"
)

print(
    f"ΔIV RMSE : {delta_rmse:.6f}"
)


# ============================================================
# 16. DIRECTION ACCURACY
# ============================================================

actual_direction = np.sign(
    y_test.values
)

predicted_direction = np.sign(
    predicted_delta
)

direction_accuracy = np.mean(

    actual_direction
    ==
    predicted_direction

) * 100


print(
    f"Direction accuracy: "
    f"{direction_accuracy:.2f}%"
)


# ============================================================
# 17. FEATURE IMPORTANCE
# ============================================================

importance = pd.DataFrame({

    "Feature":
        features,

    "Importance":
        rf_delta.feature_importances_
})


importance = (
    importance
    .sort_values(
        "Importance",
        ascending=False
    )
)


print("\n" + "=" * 70)

print(
    "FEATURE IMPORTANCE"
)

print("=" * 70)

print(
    importance.to_string(
        index=False
    )
)


# ============================================================
# 18. SAVE PREDICTIONS
# ============================================================

results = test[
    [
        "time_5m",
        "option_type",
        "strike",
        "spot",
        "moneyness",
        "IV",
        "IV_t_plus_5",
        "IV_change_5m"
    ]
].copy()


results[
    "predicted_IV_change"
] = predicted_delta


results[
    "baseline_future_IV"
] = baseline_future_iv


results[
    "predicted_future_IV"
] = predicted_future_iv


results[
    "baseline_error"
] = (

    actual_future_iv
    - baseline_future_iv

)


results[
    "RF_delta_error"
] = (

    actual_future_iv
    - predicted_future_iv

)


OUTPUT_FILE = (

    PROCESSED_FOLDER /

    "random_forest_delta_iv_predictions.csv"

)


results.to_csv(
    OUTPUT_FILE,
    index=False
)


print("\n" + "=" * 70)

print(
    "PREDICTIONS SAVED"
)

print("=" * 70)

print(OUTPUT_FILE)
```

```python
# ============================================================
# STEP 10
# XGBOOST MODEL FOR 5-MINUTE IV CHANGE
# ============================================================

import pandas as pd
import numpy as np

from pathlib import Path

from xgboost import XGBRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ============================================================
# 1. LOAD NEAR-ATM DATASET
# ============================================================

BASE_DIR = Path.home() / "Downloads"

PROCESSED_FOLDER = (
    BASE_DIR /
    "data" /
    "processed"
)

FILE = (
    PROCESSED_FOLDER /
    "nifty_options_ml_5min_near_atm.csv"
)

df = pd.read_csv(FILE)


# Convert timestamp
df["time_5m"] = pd.to_datetime(
    df["time_5m"],
    utc=True
)


print("=" * 70)
print("DATASET")
print("=" * 70)

print("Rows:", len(df))

print(
    "Unique timestamps:",
    df["time_5m"].nunique()
)


# ============================================================
# 2. FEATURES
# ============================================================

features = [

    "IV",

    "spot",

    "strike",

    "moneyness",

    "distance_from_ATM_pct",

    "LTP",

    "Volume",

    "OI",

    "Bid",

    "Ask",

    "bid_ask_spread",

    "relative_spread_pct",

    "pcr",

    "days_to_expiry",

    "option_type_code"
]


# ============================================================
# 3. TARGET
# ============================================================

target = "IV_change_5m"


print("\nTarget:")
print(target)


# ============================================================
# 4. CHRONOLOGICAL 80/20 SPLIT
# ============================================================

unique_times = np.array(
    sorted(
        df["time_5m"].unique()
    )
)

split_index = int(
    len(unique_times) * 0.80
)


train_times = (
    unique_times[:split_index]
)

test_times = (
    unique_times[split_index:]
)


train = df[
    df["time_5m"].isin(
        train_times
    )
].copy()


test = df[
    df["time_5m"].isin(
        test_times
    )
].copy()


print("\n" + "=" * 70)
print("CHRONOLOGICAL SPLIT")
print("=" * 70)

print(
    "Train rows:",
    len(train)
)

print(
    "Test rows:",
    len(test)
)

print(
    "Train timestamps:",
    train["time_5m"].nunique()
)

print(
    "Test timestamps:",
    test["time_5m"].nunique()
)

print("\nTrain period:")

print(
    train["time_5m"].min(),
    "to",
    train["time_5m"].max()
)

print("\nTest period:")

print(
    test["time_5m"].min(),
    "to",
    test["time_5m"].max()
)


# ============================================================
# 5. PREPARE X AND y
# ============================================================

X_train = train[
    features
].copy()

X_test = test[
    features
].copy()

y_train = train[
    target
].copy()

y_test = test[
    target
].copy()


# ============================================================
# 6. HANDLE MISSING VALUES
# ============================================================

# Calculate medians ONLY using training data

train_medians = (
    X_train.median()
)

X_train = (
    X_train.fillna(
        train_medians
    )
)

X_test = (
    X_test.fillna(
        train_medians
    )
)


print(
    "\nMissing X_train:",
    X_train.isna().sum().sum()
)

print(
    "Missing X_test:",
    X_test.isna().sum().sum()
)


# ============================================================
# 7. PERSISTENCE BASELINE
# ============================================================

# Persistence assumes:
#
# predicted change = 0
#
# Therefore:
# future IV = current IV

baseline_future_iv = (
    test["IV"].values
)


actual_future_iv = (
    test["IV_t_plus_5"].values
)


# ============================================================
# 8. CREATE XGBOOST MODEL
# ============================================================

xgb_model = XGBRegressor(

    n_estimators=300,

    learning_rate=0.03,

    max_depth=4,

    min_child_weight=5,

    subsample=0.8,

    colsample_bytree=0.8,

    reg_alpha=0.1,

    reg_lambda=1.0,

    objective="reg:squarederror",

    random_state=42,

    n_jobs=-1
)


# ============================================================
# 9. TRAIN MODEL
# ============================================================

print("\nTraining XGBoost...")

xgb_model.fit(
    X_train,
    y_train
)

print(
    "XGBoost training completed."
)


# ============================================================
# 10. PREDICT ΔIV
# ============================================================

predicted_delta = (
    xgb_model.predict(
        X_test
    )
)


# ============================================================
# 11. RECONSTRUCT FUTURE IV
# ============================================================

predicted_future_iv = (

    test["IV"].values

    + predicted_delta

)


# ============================================================
# 12. BASELINE PERFORMANCE
# ============================================================

baseline_mae = mean_absolute_error(
    actual_future_iv,
    baseline_future_iv
)

baseline_rmse = np.sqrt(
    mean_squared_error(
        actual_future_iv,
        baseline_future_iv
    )
)

baseline_r2 = r2_score(
    actual_future_iv,
    baseline_future_iv
)


# ============================================================
# 13. XGBOOST PERFORMANCE
# ============================================================

xgb_mae = mean_absolute_error(
    actual_future_iv,
    predicted_future_iv
)

xgb_rmse = np.sqrt(
    mean_squared_error(
        actual_future_iv,
        predicted_future_iv
    )
)

xgb_r2 = r2_score(
    actual_future_iv,
    predicted_future_iv
)


# ============================================================
# 14. DISPLAY RESULTS
# ============================================================

print("\n" + "=" * 70)

print(
    "5-MINUTE IV FORECAST RESULTS"
)

print("=" * 70)


print("\nPERSISTENCE BASELINE")

print(
    f"MAE  : {baseline_mae:.6f}"
)

print(
    f"RMSE : {baseline_rmse:.6f}"
)

print(
    f"R²   : {baseline_r2:.6f}"
)


print("\nXGBOOST ΔIV MODEL")

print(
    f"MAE  : {xgb_mae:.6f}"
)

print(
    f"RMSE : {xgb_rmse:.6f}"
)

print(
    f"R²   : {xgb_r2:.6f}"
)


# ============================================================
# 15. IMPROVEMENT OVER BASELINE
# ============================================================

mae_improvement = (

    (
        baseline_mae
        - xgb_mae
    )

    / baseline_mae

    * 100
)


rmse_improvement = (

    (
        baseline_rmse
        - xgb_rmse
    )

    / baseline_rmse

    * 100
)


print(
    "\nImprovement over persistence:"
)

print(
    f"MAE improvement  : "
    f"{mae_improvement:.2f}%"
)

print(
    f"RMSE improvement : "
    f"{rmse_improvement:.2f}%"
)


# ============================================================
# 16. ΔIV PERFORMANCE
# ============================================================

delta_mae = mean_absolute_error(
    y_test,
    predicted_delta
)

delta_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        predicted_delta
    )
)


print("\n" + "=" * 70)

print(
    "IV CHANGE PREDICTION"
)

print("=" * 70)

print(
    f"ΔIV MAE  : {delta_mae:.6f}"
)

print(
    f"ΔIV RMSE : {delta_rmse:.6f}"
)


# ============================================================
# 17. DIRECTION ACCURACY
# ============================================================

actual_direction = np.sign(
    y_test.values
)

predicted_direction = np.sign(
    predicted_delta
)


direction_accuracy = np.mean(

    actual_direction
    ==
    predicted_direction

) * 100


print(
    f"Direction accuracy: "
    f"{direction_accuracy:.2f}%"
)


# ============================================================
# 18. FEATURE IMPORTANCE
# ============================================================

importance = pd.DataFrame(
    {
        "Feature":
            features,

        "Importance":
            xgb_model.feature_importances_
    }
)


importance = (
    importance
    .sort_values(
        "Importance",
        ascending=False
    )
)


print("\n" + "=" * 70)

print(
    "FEATURE IMPORTANCE"
)

print("=" * 70)

print(
    importance.to_string(
        index=False
    )
)


# ============================================================
# 19. COMPARE WITH RANDOM FOREST
# ============================================================

# Results from our previous Random Forest ΔIV model

rf_mae = 0.544259
rf_rmse = 1.292109
rf_r2 = 0.984675


comparison = pd.DataFrame(
    {
        "Model": [
            "Persistence Baseline",
            "Random Forest ΔIV",
            "XGBoost ΔIV"
        ],

        "MAE": [
            baseline_mae,
            rf_mae,
            xgb_mae
        ],

        "RMSE": [
            baseline_rmse,
            rf_rmse,
            xgb_rmse
        ],

        "R2": [
            baseline_r2,
            rf_r2,
            xgb_r2
        ]
    }
)


print("\n" + "=" * 70)

print(
    "FINAL MODEL COMPARISON"
)

print("=" * 70)

print(
    comparison.to_string(
        index=False
    )
)


# ============================================================
# 20. IDENTIFY BEST MODEL
# ============================================================

best_model = (
    comparison
    .sort_values("MAE")
    .iloc[0]
)


print("\n" + "=" * 70)

print("BEST MODEL BASED ON MAE")

print("=" * 70)

print(
    "Model:",
    best_model["Model"]
)

print(
    f"MAE: {best_model['MAE']:.6f}"
)

print(
    f"RMSE: {best_model['RMSE']:.6f}"
)

print(
    f"R²: {best_model['R2']:.6f}"
)


# ============================================================
# 21. SAVE XGBOOST PREDICTIONS
# ============================================================

results = test[
    [
        "time_5m",
        "option_type",
        "strike",
        "spot",
        "moneyness",
        "IV",
        "IV_t_plus_5",
        "IV_change_5m"
    ]
].copy()


results[
    "predicted_IV_change"
] = predicted_delta


results[
    "baseline_future_IV"
] = baseline_future_iv


results[
    "predicted_future_IV"
] = predicted_future_iv


results[
    "baseline_error"
] = (

    actual_future_iv
    - baseline_future_iv

)


results[
    "XGB_delta_error"
] = (

    actual_future_iv
    - predicted_future_iv

)


OUTPUT_FILE = (

    PROCESSED_FOLDER /

    "xgboost_delta_iv_predictions.csv"

)


results.to_csv(
    OUTPUT_FILE,
    index=False
)


print("\n" + "=" * 70)

print(
    "XGBOOST PREDICTIONS SAVED"
)

print("=" * 70)

print(OUTPUT_FILE)
```

```python
# ============================================================
# STEP 11: MODEL DIAGNOSTICS AND VISUALIZATION
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path


# ============================================================
# 1. FILE LOCATIONS
# ============================================================

BASE_DIR = Path.home() / "Downloads"

PROCESSED_FOLDER = (
    BASE_DIR /
    "data" /
    "processed"
)

RF_FILE = (
    PROCESSED_FOLDER /
    "random_forest_delta_iv_predictions.csv"
)

XGB_FILE = (
    PROCESSED_FOLDER /
    "xgboost_delta_iv_predictions.csv"
)


# ============================================================
# 2. LOAD PREDICTIONS
# ============================================================

rf = pd.read_csv(RF_FILE)
xgb = pd.read_csv(XGB_FILE)

rf["time_5m"] = pd.to_datetime(
    rf["time_5m"],
    utc=True
)

xgb["time_5m"] = pd.to_datetime(
    xgb["time_5m"],
    utc=True
)


print("=" * 70)
print("MODEL DIAGNOSTIC DATA")
print("=" * 70)

print("Random Forest rows:", len(rf))
print("XGBoost rows:", len(xgb))


# ============================================================
# 3. CALCULATE ABSOLUTE ERRORS
# ============================================================

rf["RF_absolute_error"] = abs(
    rf["IV_t_plus_5"]
    - rf["predicted_future_IV"]
)

rf["baseline_absolute_error"] = abs(
    rf["IV_t_plus_5"]
    - rf["baseline_future_IV"]
)

xgb["XGB_absolute_error"] = abs(
    xgb["IV_t_plus_5"]
    - xgb["predicted_future_IV"]
)


# ============================================================
# 4. GRAPH 1
# ACTUAL FUTURE IV VS RANDOM FOREST PREDICTED IV
# ============================================================

plt.figure(figsize=(8, 6))

plt.scatter(
    rf["IV_t_plus_5"],
    rf["predicted_future_IV"],
    alpha=0.5
)

minimum = min(
    rf["IV_t_plus_5"].min(),
    rf["predicted_future_IV"].min()
)

maximum = max(
    rf["IV_t_plus_5"].max(),
    rf["predicted_future_IV"].max()
)

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--"
)

plt.xlabel("Actual IV after 5 Minutes")

plt.ylabel("Predicted IV after 5 Minutes")

plt.title(
    "Random Forest: Actual vs Predicted 5-Minute IV"
)

plt.grid(alpha=0.3)

plt.show()


# ============================================================
# 5. GRAPH 2
# RANDOM FOREST ERROR DISTRIBUTION
# ============================================================

rf_error = (
    rf["IV_t_plus_5"]
    - rf["predicted_future_IV"]
)

plt.figure(figsize=(8, 6))

plt.hist(
    rf_error,
    bins=40,
    edgecolor="black"
)

plt.axvline(
    0,
    linestyle="--"
)

plt.xlabel("Prediction Error")

plt.ylabel("Frequency")

plt.title(
    "Random Forest 5-Minute IV Prediction Errors"
)

plt.grid(alpha=0.3)

plt.show()


# ============================================================
# 6. MODEL PERFORMANCE TABLE
# ============================================================

baseline_mae = (
    rf["baseline_absolute_error"].mean()
)

rf_mae = (
    rf["RF_absolute_error"].mean()
)

xgb_mae = (
    xgb["XGB_absolute_error"].mean()
)


baseline_rmse = np.sqrt(
    np.mean(
        (
            rf["IV_t_plus_5"]
            - rf["baseline_future_IV"]
        ) ** 2
    )
)

rf_rmse = np.sqrt(
    np.mean(
        (
            rf["IV_t_plus_5"]
            - rf["predicted_future_IV"]
        ) ** 2
    )
)

xgb_rmse = np.sqrt(
    np.mean(
        (
            xgb["IV_t_plus_5"]
            - xgb["predicted_future_IV"]
        ) ** 2
    )
)


performance = pd.DataFrame(
    {
        "Model": [
            "Persistence",
            "Random Forest ΔIV",
            "XGBoost ΔIV"
        ],

        "MAE": [
            baseline_mae,
            rf_mae,
            xgb_mae
        ],

        "RMSE": [
            baseline_rmse,
            rf_rmse,
            xgb_rmse
        ]
    }
)


print("\n" + "=" * 70)
print("MODEL PERFORMANCE")
print("=" * 70)

print(
    performance.to_string(
        index=False
    )
)


# ============================================================
# 7. GRAPH 3
# MAE COMPARISON
# ============================================================

plt.figure(figsize=(8, 6))

plt.bar(
    performance["Model"],
    performance["MAE"]
)

plt.ylabel("MAE")

plt.title(
    "5-Minute IV Forecast: MAE Comparison"
)

plt.xticks(rotation=15)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.show()


# ============================================================
# 8. GRAPH 4
# RMSE COMPARISON
# ============================================================

plt.figure(figsize=(8, 6))

plt.bar(
    performance["Model"],
    performance["RMSE"]
)

plt.ylabel("RMSE")

plt.title(
    "5-Minute IV Forecast: RMSE Comparison"
)

plt.xticks(rotation=15)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.show()


# ============================================================
# 9. CHECK HOW OFTEN RF BEATS PERSISTENCE
# ============================================================

rf["RF_beats_baseline"] = (
    rf["RF_absolute_error"]
    <
    rf["baseline_absolute_error"]
)


beat_percentage = (
    rf["RF_beats_baseline"].mean()
    * 100
)


print("\n" + "=" * 70)
print("RANDOM FOREST VS PERSISTENCE")
print("=" * 70)

print(
    f"RF has lower absolute error on "
    f"{beat_percentage:.2f}% of test observations."
)


# ============================================================
# 10. ERROR QUANTILES
# ============================================================

print("\nRandom Forest absolute error quantiles:")

print(
    rf["RF_absolute_error"].quantile(
        [
            0.50,
            0.75,
            0.90,
            0.95,
            0.99,
            1.00
        ]
    )
)


# ============================================================
# 11. SAVE PERFORMANCE SUMMARY
# ============================================================

OUTPUT_FILE = (
    PROCESSED_FOLDER /
    "model_performance_summary.csv"
)

performance.to_csv(
    OUTPUT_FILE,
    index=False
)

print("\nPerformance summary saved:")
print(OUTPUT_FILE)
```

```python
# ============================================================
# STEP 11: CREATE LAG AND MOMENTUM FEATURES
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path


# ============================================================
# 1. LOAD NEAR-ATM 5-MINUTE DATASET
# ============================================================

BASE_DIR = Path.home() / "Downloads"
PROCESSED_FOLDER = BASE_DIR / "data" / "processed"

INPUT_FILE = (
    PROCESSED_FOLDER /
    "nifty_options_ml_5min_near_atm.csv"
)

df = pd.read_csv(INPUT_FILE)

df["time_5m"] = pd.to_datetime(
    df["time_5m"],
    utc=True
)

print("=" * 70)
print("ORIGINAL DATA")
print("=" * 70)

print("Rows:", len(df))
print("Unique timestamps:", df["time_5m"].nunique())


# ============================================================
# 2. SORT EACH OPTION CONTRACT CHRONOLOGICALLY
# ============================================================

df = df.sort_values(
    [
        "expiry",
        "option_type",
        "strike",
        "time_5m"
    ]
).reset_index(drop=True)


# ============================================================
# 3. CREATE EXACT PREVIOUS TIMES
# ============================================================

df["time_minus_5m"] = (
    df["time_5m"]
    - pd.Timedelta(minutes=5)
)

df["time_minus_10m"] = (
    df["time_5m"]
    - pd.Timedelta(minutes=10)
)


# ============================================================
# 4. VARIABLES WE WANT HISTORICAL INFORMATION FOR
# ============================================================

lag_variables = [
    "IV",
    "spot",
    "LTP",
    "Volume",
    "OI",
    "bid_ask_spread"
]


# ============================================================
# 5. CREATE 5-MINUTE LAG TABLE
# ============================================================

lag5 = df[
    [
        "time_5m",
        "expiry",
        "option_type",
        "strike"
    ]
    + lag_variables
].copy()

lag5 = lag5.rename(
    columns={
        "time_5m": "time_minus_5m",

        **{
            col: f"{col}_lag_5m"
            for col in lag_variables
        }
    }
)


# Merge only same expiry + option type + strike
# exactly 5 minutes earlier

df = df.merge(
    lag5,
    on=[
        "time_minus_5m",
        "expiry",
        "option_type",
        "strike"
    ],
    how="left"
)


# ============================================================
# 6. CREATE 10-MINUTE LAG TABLE
# ============================================================

lag10 = df[
    [
        "time_5m",
        "expiry",
        "option_type",
        "strike"
    ]
    + lag_variables
].copy()

lag10 = lag10.rename(
    columns={
        "time_5m": "time_minus_10m",

        **{
            col: f"{col}_lag_10m"
            for col in lag_variables
        }
    }
)


df = df.merge(
    lag10,
    on=[
        "time_minus_10m",
        "expiry",
        "option_type",
        "strike"
    ],
    how="left"
)


# ============================================================
# 7. CREATE MOMENTUM / CHANGE FEATURES
# ============================================================

for col in lag_variables:

    # Change compared with exactly 5 minutes ago
    df[f"{col}_change_prev_5m"] = (
        df[col]
        - df[f"{col}_lag_5m"]
    )

    # Change compared with exactly 10 minutes ago
    df[f"{col}_change_prev_10m"] = (
        df[col]
        - df[f"{col}_lag_10m"]
    )


# ============================================================
# 8. CHECK IV LAG FEATURES
# ============================================================

print("\n" + "=" * 70)
print("LAG AVAILABILITY")
print("=" * 70)

print(
    "Rows with exact 5-minute IV history:",
    df["IV_lag_5m"].notna().sum()
)

print(
    "Rows with exact 10-minute IV history:",
    df["IV_lag_10m"].notna().sum()
)

print(
    "Rows with BOTH 5m and 10m IV history:",
    (
        df["IV_lag_5m"].notna()
        &
        df["IV_lag_10m"].notna()
    ).sum()
)


# ============================================================
# 9. VERIFY THAT WE DID NOT USE FUTURE INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("SAMPLE IV HISTORY")
print("=" * 70)

sample_columns = [
    "time_5m",
    "option_type",
    "strike",
    "IV_lag_10m",
    "IV_lag_5m",
    "IV",
    "IV_change_prev_5m",
    "IV_change_prev_10m",
    "IV_t_plus_5"
]

print(
    df[
        sample_columns
    ]
    .dropna()
    .head(15)
    .to_string(index=False)
)


# ============================================================
# 10. CREATE FINAL LAG DATASET
# ============================================================

# For the next model, require BOTH exact
# 5-minute and exact 10-minute history.

lagged_df = df[
    df["IV_lag_5m"].notna()
    &
    df["IV_lag_10m"].notna()
].copy()


print("\n" + "=" * 70)
print("FINAL LAGGED DATASET")
print("=" * 70)

print(
    "Rows before lag requirement:",
    len(df)
)

print(
    "Rows after lag requirement:",
    len(lagged_df)
)

print(
    "Rows retained:",
    f"{len(lagged_df) / len(df) * 100:.2f}%"
)

print(
    "Unique timestamps:",
    lagged_df["time_5m"].nunique()
)

print(
    "Unique strikes:",
    lagged_df["strike"].nunique()
)

print("\nOption counts:")

print(
    lagged_df[
        "option_type"
    ].value_counts()
)


# ============================================================
# 11. CHECK NEW IV MOMENTUM
# ============================================================

print("\n" + "=" * 70)
print("IV MOMENTUM SUMMARY")
print("=" * 70)

print(
    lagged_df[
        [
            "IV_change_prev_5m",
            "IV_change_prev_10m",
            "IV_change_5m"
        ]
    ].describe()
)


# ============================================================
# 12. SAVE
# ============================================================

OUTPUT_FILE = (
    PROCESSED_FOLDER /
    "nifty_options_ml_5min_lagged.csv"
)

lagged_df.to_csv(
    OUTPUT_FILE,
    index=False
)

print("\n" + "=" * 70)
print("SAVED")
print("=" * 70)

print(OUTPUT_FILE)
```

```python
# ============================================================
# STEP 12
# RANDOM FOREST ΔIV MODEL WITH LAG / MOMENTUM FEATURES
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ============================================================
# 1. LOAD LAGGED DATA
# ============================================================

BASE_DIR = Path.home() / "Downloads"
PROCESSED_FOLDER = BASE_DIR / "data" / "processed"

FILE = (
    PROCESSED_FOLDER /
    "nifty_options_ml_5min_lagged.csv"
)

df = pd.read_csv(FILE)

df["time_5m"] = pd.to_datetime(
    df["time_5m"],
    utc=True
)


print("=" * 70)
print("LAGGED DATASET")
print("=" * 70)

print("Rows:", len(df))
print("Unique timestamps:", df["time_5m"].nunique())


# ============================================================
# 2. ORIGINAL FEATURES
# ============================================================

original_features = [
    "IV",
    "spot",
    "strike",
    "moneyness",
    "distance_from_ATM_pct",
    "LTP",
    "Volume",
    "OI",
    "Bid",
    "Ask",
    "bid_ask_spread",
    "relative_spread_pct",
    "pcr",
    "days_to_expiry",
    "option_type_code"
]


# ============================================================
# 3. NEW HISTORICAL FEATURES
# ============================================================

lag_features = [
    # IV history
    "IV_lag_5m",
    "IV_lag_10m",
    "IV_change_prev_5m",
    "IV_change_prev_10m",

    # Spot history
    "spot_lag_5m",
    "spot_lag_10m",
    "spot_change_prev_5m",
    "spot_change_prev_10m",

    # Option price history
    "LTP_lag_5m",
    "LTP_lag_10m",
    "LTP_change_prev_5m",
    "LTP_change_prev_10m",

    # Volume history
    "Volume_lag_5m",
    "Volume_lag_10m",
    "Volume_change_prev_5m",
    "Volume_change_prev_10m",

    # Open interest history
    "OI_lag_5m",
    "OI_lag_10m",
    "OI_change_prev_5m",
    "OI_change_prev_10m",

    # Bid-ask spread history
    "bid_ask_spread_lag_5m",
    "bid_ask_spread_lag_10m",
    "bid_ask_spread_change_prev_5m",
    "bid_ask_spread_change_prev_10m"
]


features = (
    original_features
    + lag_features
)

target = "IV_change_5m"


print("\nOriginal features:", len(original_features))
print("New lag features:", len(lag_features))
print("Total features:", len(features))


# ============================================================
# 4. CHRONOLOGICAL 80/20 SPLIT
# ============================================================

unique_times = np.array(
    sorted(df["time_5m"].unique())
)

split_index = int(
    len(unique_times) * 0.80
)

train_times = unique_times[:split_index]
test_times = unique_times[split_index:]


train = df[
    df["time_5m"].isin(train_times)
].copy()

test = df[
    df["time_5m"].isin(test_times)
].copy()


print("\n" + "=" * 70)
print("CHRONOLOGICAL SPLIT")
print("=" * 70)

print("Train rows:", len(train))
print("Test rows:", len(test))

print(
    "Train timestamps:",
    train["time_5m"].nunique()
)

print(
    "Test timestamps:",
    test["time_5m"].nunique()
)

print(
    "\nTrain period:",
    train["time_5m"].min(),
    "to",
    train["time_5m"].max()
)

print(
    "Test period:",
    test["time_5m"].min(),
    "to",
    test["time_5m"].max()
)


# ============================================================
# 5. PREPARE TRAIN / TEST DATA
# ============================================================

X_train = train[features].copy()
X_test = test[features].copy()

y_train = train[target].copy()
y_test = test[target].copy()


# ============================================================
# 6. HANDLE MISSING VALUES
# ============================================================

# IMPORTANT:
# Medians calculated only from training data.

train_medians = X_train.median()

X_train = X_train.fillna(
    train_medians
)

X_test = X_test.fillna(
    train_medians
)


print(
    "\nMissing X_train:",
    X_train.isna().sum().sum()
)

print(
    "Missing X_test:",
    X_test.isna().sum().sum()
)


# ============================================================
# 7. CREATE RANDOM FOREST
# ============================================================

rf_lag = RandomForestRegressor(
    n_estimators=300,
    max_depth=12,
    min_samples_leaf=3,
    max_features="sqrt",
    random_state=42,
    n_jobs=-1
)


# ============================================================
# 8. TRAIN
# ============================================================

print("\nTraining Random Forest with lag features...")

rf_lag.fit(
    X_train,
    y_train
)

print("Training completed.")


# ============================================================
# 9. PREDICT ΔIV
# ============================================================

predicted_delta = rf_lag.predict(
    X_test
)


# ============================================================
# 10. RECONSTRUCT FUTURE IV
# ============================================================

current_iv = test["IV"].values

actual_future_iv = (
    test["IV_t_plus_5"].values
)

predicted_future_iv = (
    current_iv
    + predicted_delta
)


# Persistence:
# future IV = current IV

baseline_future_iv = current_iv


# ============================================================
# 11. PERSISTENCE METRICS
# ============================================================

baseline_mae = mean_absolute_error(
    actual_future_iv,
    baseline_future_iv
)

baseline_rmse = np.sqrt(
    mean_squared_error(
        actual_future_iv,
        baseline_future_iv
    )
)

baseline_r2 = r2_score(
    actual_future_iv,
    baseline_future_iv
)


# ============================================================
# 12. RANDOM FOREST METRICS
# ============================================================

rf_mae = mean_absolute_error(
    actual_future_iv,
    predicted_future_iv
)

rf_rmse = np.sqrt(
    mean_squared_error(
        actual_future_iv,
        predicted_future_iv
    )
)

rf_r2 = r2_score(
    actual_future_iv,
    predicted_future_iv
)


# ============================================================
# 13. IMPROVEMENT OVER PERSISTENCE
# ============================================================

mae_improvement = (
    (baseline_mae - rf_mae)
    / baseline_mae
) * 100

rmse_improvement = (
    (baseline_rmse - rf_rmse)
    / baseline_rmse
) * 100


# ============================================================
# 14. DIRECTION ACCURACY
# ============================================================

actual_direction = np.sign(
    y_test.values
)

predicted_direction = np.sign(
    predicted_delta
)

direction_accuracy = np.mean(
    actual_direction
    ==
    predicted_direction
) * 100


# ============================================================
# 15. PRINT RESULTS
# ============================================================

print("\n" + "=" * 70)
print("PERSISTENCE BASELINE")
print("=" * 70)

print(
    f"MAE  : {baseline_mae:.6f}"
)

print(
    f"RMSE : {baseline_rmse:.6f}"
)

print(
    f"R²   : {baseline_r2:.6f}"
)


print("\n" + "=" * 70)
print("RANDOM FOREST + LAG FEATURES")
print("=" * 70)

print(
    f"MAE  : {rf_mae:.6f}"
)

print(
    f"RMSE : {rf_rmse:.6f}"
)

print(
    f"R²   : {rf_r2:.6f}"
)

print(
    f"MAE improvement vs persistence: "
    f"{mae_improvement:.2f}%"
)

print(
    f"RMSE improvement vs persistence: "
    f"{rmse_improvement:.2f}%"
)

print(
    f"Direction accuracy: "
    f"{direction_accuracy:.2f}%"
)


# ============================================================
# 16. FEATURE IMPORTANCE
# ============================================================

importance = pd.DataFrame(
    {
        "Feature": features,
        "Importance":
            rf_lag.feature_importances_
    }
)

importance = importance.sort_values(
    "Importance",
    ascending=False
)


print("\n" + "=" * 70)
print("TOP 20 FEATURES")
print("=" * 70)

print(
    importance.head(20).to_string(
        index=False
    )
)


# ============================================================
# 17. PREDICTED ΔIV SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("PREDICTED ΔIV SUMMARY")
print("=" * 70)

print(
    pd.Series(
        predicted_delta
    ).describe()
)


# ============================================================
# 18. SAVE PREDICTIONS
# ============================================================

results = test[
    [
        "time_5m",
        "option_type",
        "strike",
        "spot",
        "moneyness",
        "IV",
        "IV_t_plus_5",
        "IV_change_5m"
    ]
].copy()


results[
    "predicted_IV_change"
] = predicted_delta

results[
    "predicted_future_IV"
] = predicted_future_iv

results[
    "baseline_future_IV"
] = baseline_future_iv


OUTPUT_FILE = (
    PROCESSED_FOLDER /
    "random_forest_lagged_predictions.csv"
)

results.to_csv(
    OUTPUT_FILE,
    index=False
)


print("\n" + "=" * 70)
print("SAVED")
print("=" * 70)

print(OUTPUT_FILE)
```

```python
# ============================================================
# STEP 13
# XGBOOST ΔIV MODEL WITH LAG / MOMENTUM FEATURES
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

from xgboost import XGBRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ============================================================
# 1. LOAD LAGGED DATASET
# ============================================================

BASE_DIR = Path.home() / "Downloads"

PROCESSED_FOLDER = (
    BASE_DIR /
    "data" /
    "processed"
)

FILE = (
    PROCESSED_FOLDER /
    "nifty_options_ml_5min_lagged.csv"
)

df = pd.read_csv(FILE)

df["time_5m"] = pd.to_datetime(
    df["time_5m"],
    utc=True
)


print("=" * 70)
print("LAGGED DATASET")
print("=" * 70)

print(
    "Rows:",
    len(df)
)

print(
    "Unique timestamps:",
    df["time_5m"].nunique()
)


# ============================================================
# 2. ORIGINAL FEATURES
# ============================================================

original_features = [

    "IV",

    "spot",

    "strike",

    "moneyness",

    "distance_from_ATM_pct",

    "LTP",

    "Volume",

    "OI",

    "Bid",

    "Ask",

    "bid_ask_spread",

    "relative_spread_pct",

    "pcr",

    "days_to_expiry",

    "option_type_code"
]


# ============================================================
# 3. LAG / MOMENTUM FEATURES
# ============================================================

lag_features = [

    # IV
    "IV_lag_5m",
    "IV_lag_10m",
    "IV_change_prev_5m",
    "IV_change_prev_10m",

    # Spot
    "spot_lag_5m",
    "spot_lag_10m",
    "spot_change_prev_5m",
    "spot_change_prev_10m",

    # LTP
    "LTP_lag_5m",
    "LTP_lag_10m",
    "LTP_change_prev_5m",
    "LTP_change_prev_10m",

    # Volume
    "Volume_lag_5m",
    "Volume_lag_10m",
    "Volume_change_prev_5m",
    "Volume_change_prev_10m",

    # Open Interest
    "OI_lag_5m",
    "OI_lag_10m",
    "OI_change_prev_5m",
    "OI_change_prev_10m",

    # Bid-Ask Spread
    "bid_ask_spread_lag_5m",
    "bid_ask_spread_lag_10m",
    "bid_ask_spread_change_prev_5m",
    "bid_ask_spread_change_prev_10m"
]


features = (
    original_features
    + lag_features
)

target = "IV_change_5m"


print(
    "\nOriginal features:",
    len(original_features)
)

print(
    "Lag features:",
    len(lag_features)
)

print(
    "Total features:",
    len(features)
)


# ============================================================
# 4. CHRONOLOGICAL 80/20 SPLIT
# ============================================================

unique_times = np.array(
    sorted(
        df["time_5m"].unique()
    )
)

split_index = int(
    len(unique_times) * 0.80
)


train_times = (
    unique_times[:split_index]
)

test_times = (
    unique_times[split_index:]
)


train = df[
    df["time_5m"].isin(
        train_times
    )
].copy()


test = df[
    df["time_5m"].isin(
        test_times
    )
].copy()


print("\n" + "=" * 70)
print("CHRONOLOGICAL SPLIT")
print("=" * 70)

print(
    "Train rows:",
    len(train)
)

print(
    "Test rows:",
    len(test)
)

print(
    "Train timestamps:",
    train["time_5m"].nunique()
)

print(
    "Test timestamps:",
    test["time_5m"].nunique()
)

print(
    "\nTrain period:",
    train["time_5m"].min(),
    "to",
    train["time_5m"].max()
)

print(
    "Test period:",
    test["time_5m"].min(),
    "to",
    test["time_5m"].max()
)


# ============================================================
# 5. PREPARE X AND y
# ============================================================

X_train = train[
    features
].copy()

X_test = test[
    features
].copy()


y_train = train[
    target
].copy()

y_test = test[
    target
].copy()


# ============================================================
# 6. HANDLE MISSING VALUES
# ============================================================

# Medians calculated ONLY from training data.

train_medians = (
    X_train.median()
)


X_train = (
    X_train.fillna(
        train_medians
    )
)

X_test = (
    X_test.fillna(
        train_medians
    )
)


print(
    "\nMissing X_train:",
    X_train.isna().sum().sum()
)

print(
    "Missing X_test:",
    X_test.isna().sum().sum()
)


# ============================================================
# 7. CREATE XGBOOST MODEL
# ============================================================

xgb_lag = XGBRegressor(

    n_estimators=300,

    learning_rate=0.03,

    max_depth=4,

    min_child_weight=5,

    subsample=0.8,

    colsample_bytree=0.8,

    reg_alpha=0.1,

    reg_lambda=1.0,

    objective="reg:squarederror",

    random_state=42,

    n_jobs=-1
)


# ============================================================
# 8. TRAIN XGBOOST
# ============================================================

print(
    "\nTraining XGBoost with lag features..."
)

xgb_lag.fit(
    X_train,
    y_train
)

print(
    "Training completed."
)


# ============================================================
# 9. PREDICT ΔIV
# ============================================================

predicted_delta = (
    xgb_lag.predict(
        X_test
    )
)


# ============================================================
# 10. RECONSTRUCT FUTURE IV
# ============================================================

current_iv = (
    test["IV"].values
)

actual_future_iv = (
    test["IV_t_plus_5"].values
)


predicted_future_iv = (

    current_iv
    + predicted_delta

)


# Persistence forecast:
# Future IV = Current IV

baseline_future_iv = (
    current_iv
)


# ============================================================
# 11. PERSISTENCE METRICS
# ============================================================

baseline_mae = mean_absolute_error(
    actual_future_iv,
    baseline_future_iv
)


baseline_rmse = np.sqrt(
    mean_squared_error(
        actual_future_iv,
        baseline_future_iv
    )
)


baseline_r2 = r2_score(
    actual_future_iv,
    baseline_future_iv
)


# ============================================================
# 12. XGBOOST METRICS
# ============================================================

xgb_mae = mean_absolute_error(
    actual_future_iv,
    predicted_future_iv
)


xgb_rmse = np.sqrt(
    mean_squared_error(
        actual_future_iv,
        predicted_future_iv
    )
)


xgb_r2 = r2_score(
    actual_future_iv,
    predicted_future_iv
)


# ============================================================
# 13. IMPROVEMENT OVER PERSISTENCE
# ============================================================

mae_improvement = (

    (
        baseline_mae
        - xgb_mae
    )

    / baseline_mae

    * 100
)


rmse_improvement = (

    (
        baseline_rmse
        - xgb_rmse
    )

    / baseline_rmse

    * 100
)


# ============================================================
# 14. DIRECTION ACCURACY
# ============================================================

actual_direction = np.sign(
    y_test.values
)

predicted_direction = np.sign(
    predicted_delta
)


direction_accuracy = np.mean(

    actual_direction
    ==
    predicted_direction

) * 100


# ============================================================
# 15. PRINT RESULTS
# ============================================================

print("\n" + "=" * 70)
print("PERSISTENCE BASELINE")
print("=" * 70)

print(
    f"MAE  : {baseline_mae:.6f}"
)

print(
    f"RMSE : {baseline_rmse:.6f}"
)

print(
    f"R²   : {baseline_r2:.6f}"
)


print("\n" + "=" * 70)
print("XGBOOST + LAG FEATURES")
print("=" * 70)

print(
    f"MAE  : {xgb_mae:.6f}"
)

print(
    f"RMSE : {xgb_rmse:.6f}"
)

print(
    f"R²   : {xgb_r2:.6f}"
)

print(
    f"MAE improvement vs persistence: "
    f"{mae_improvement:.2f}%"
)

print(
    f"RMSE improvement vs persistence: "
    f"{rmse_improvement:.2f}%"
)

print(
    f"Direction accuracy: "
    f"{direction_accuracy:.2f}%"
)


# ============================================================
# 16. ΔIV METRICS
# ============================================================

delta_mae = mean_absolute_error(
    y_test,
    predicted_delta
)


delta_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        predicted_delta
    )
)


print("\n" + "=" * 70)
print("ΔIV PREDICTION")
print("=" * 70)

print(
    f"ΔIV MAE  : {delta_mae:.6f}"
)

print(
    f"ΔIV RMSE : {delta_rmse:.6f}"
)


# ============================================================
# 17. FEATURE IMPORTANCE
# ============================================================

importance = pd.DataFrame(
    {
        "Feature":
            features,

        "Importance":
            xgb_lag.feature_importances_
    }
)


importance = (
    importance
    .sort_values(
        "Importance",
        ascending=False
    )
)


print("\n" + "=" * 70)
print("TOP 20 FEATURES")
print("=" * 70)

print(
    importance
    .head(20)
    .to_string(
        index=False
    )
)


# ============================================================
# 18. PREDICTED ΔIV SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("PREDICTED ΔIV SUMMARY")
print("=" * 70)

print(
    pd.Series(
        predicted_delta
    ).describe()
)


# ============================================================
# 19. DIRECT COMPARISON WITH LAGGED RANDOM FOREST
# ============================================================

# Results obtained in Step 12
rf_mae = 0.521398
rf_rmse = 1.218771
rf_r2 = 0.986619
rf_direction = 53.20


comparison = pd.DataFrame(
    {
        "Model": [
            "Persistence",
            "Random Forest + Lag",
            "XGBoost + Lag"
        ],

        "MAE": [
            baseline_mae,
            rf_mae,
            xgb_mae
        ],

        "RMSE": [
            baseline_rmse,
            rf_rmse,
            xgb_rmse
        ],

        "R2": [
            baseline_r2,
            rf_r2,
            xgb_r2
        ],

        "Direction_Accuracy_%": [
            np.nan,
            rf_direction,
            direction_accuracy
        ]
    }
)


print("\n" + "=" * 70)
print("FAIR MODEL COMPARISON")
print("=" * 70)

print(
    comparison.to_string(
        index=False
    )
)


# ============================================================
# 20. SAVE PREDICTIONS
# ============================================================

results = test[
    [
        "time_5m",
        "option_type",
        "strike",
        "spot",
        "moneyness",
        "IV",
        "IV_t_plus_5",
        "IV_change_5m"
    ]
].copy()


results[
    "predicted_IV_change"
] = predicted_delta


results[
    "predicted_future_IV"
] = predicted_future_iv


results[
    "baseline_future_IV"
] = baseline_future_iv


OUTPUT_FILE = (
    PROCESSED_FOLDER /
    "xgboost_lagged_predictions.csv"
)


results.to_csv(
    OUTPUT_FILE,
    index=False
)


print("\n" + "=" * 70)
print("SAVED")
print("=" * 70)

print(OUTPUT_FILE)
```

```python
# ============================================================
# STEP 14
# ENSEMBLE + THRESHOLD / CONVICTION ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ============================================================
# 1. FILE PATHS
# ============================================================

BASE_DIR = Path.home() / "Downloads"
PROCESSED_FOLDER = BASE_DIR / "data" / "processed"

RF_FILE = (
    PROCESSED_FOLDER /
    "random_forest_lagged_predictions.csv"
)

XGB_FILE = (
    PROCESSED_FOLDER /
    "xgboost_lagged_predictions.csv"
)


# ============================================================
# 2. LOAD BOTH MODEL PREDICTIONS
# ============================================================

rf = pd.read_csv(RF_FILE)
xgb = pd.read_csv(XGB_FILE)

rf["time_5m"] = pd.to_datetime(
    rf["time_5m"],
    utc=True
)

xgb["time_5m"] = pd.to_datetime(
    xgb["time_5m"],
    utc=True
)


print("=" * 75)
print("FILES LOADED")
print("=" * 75)

print("RF rows :", len(rf))
print("XGB rows:", len(xgb))


# ============================================================
# 3. RENAME PREDICTIONS
# ============================================================

rf = rf.rename(
    columns={
        "predicted_IV_change":
        "RF_predicted_delta"
    }
)

xgb = xgb.rename(
    columns={
        "predicted_IV_change":
        "XGB_predicted_delta"
    }
)


# ============================================================
# 4. KEEP REQUIRED COLUMNS
# ============================================================

keys = [
    "time_5m",
    "option_type",
    "strike"
]


rf_keep = rf[
    keys +
    [
        "spot",
        "moneyness",
        "IV",
        "IV_t_plus_5",
        "IV_change_5m",
        "RF_predicted_delta"
    ]
].copy()


xgb_keep = xgb[
    keys +
    [
        "XGB_predicted_delta"
    ]
].copy()


# ============================================================
# 5. MERGE RF AND XGBOOST
# ============================================================

df = pd.merge(
    rf_keep,
    xgb_keep,
    on=keys,
    how="inner"
)


print("\n" + "=" * 75)
print("MATCHED TEST DATA")
print("=" * 75)

print("Matched rows:", len(df))
print(
    "Unique timestamps:",
    df["time_5m"].nunique()
)

print(
    "Period:",
    df["time_5m"].min(),
    "to",
    df["time_5m"].max()
)


# ============================================================
# 6. CREATE 50/50 ENSEMBLE
# ============================================================

df["Ensemble_predicted_delta"] = (
    0.50 * df["RF_predicted_delta"]
    +
    0.50 * df["XGB_predicted_delta"]
)


# ============================================================
# 7. RECONSTRUCT FUTURE IV
# ============================================================

df["Persistence_future_IV"] = df["IV"]

df["RF_future_IV"] = (
    df["IV"]
    + df["RF_predicted_delta"]
)

df["XGB_future_IV"] = (
    df["IV"]
    + df["XGB_predicted_delta"]
)

df["Ensemble_future_IV"] = (
    df["IV"]
    + df["Ensemble_predicted_delta"]
)


# ============================================================
# 8. FUNCTION FOR FORECAST METRICS
# ============================================================

def forecast_metrics(
    name,
    actual,
    predicted
):

    mae = mean_absolute_error(
        actual,
        predicted
    )

    rmse = np.sqrt(
        mean_squared_error(
            actual,
            predicted
        )
    )

    r2 = r2_score(
        actual,
        predicted
    )

    return {
        "Model": name,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    }


# ============================================================
# 9. COMPARE FUTURE-IV FORECASTS
# ============================================================

actual_future = df["IV_t_plus_5"]


performance = pd.DataFrame(
    [
        forecast_metrics(
            "Persistence",
            actual_future,
            df["Persistence_future_IV"]
        ),

        forecast_metrics(
            "Random Forest + Lag",
            actual_future,
            df["RF_future_IV"]
        ),

        forecast_metrics(
            "XGBoost + Lag",
            actual_future,
            df["XGB_future_IV"]
        ),

        forecast_metrics(
            "50/50 Ensemble",
            actual_future,
            df["Ensemble_future_IV"]
        )
    ]
)


print("\n" + "=" * 75)
print("FUTURE-IV FORECAST COMPARISON")
print("=" * 75)

print(
    performance
    .sort_values("MAE")
    .to_string(index=False)
)


# ============================================================
# 10. DIRECTION ACCURACY FUNCTION
# ============================================================

def direction_accuracy(
    actual_delta,
    predicted_delta
):

    actual_sign = np.sign(
        actual_delta
    )

    predicted_sign = np.sign(
        predicted_delta
    )

    return (
        np.mean(
            actual_sign
            ==
            predicted_sign
        )
        * 100
    )


# ============================================================
# 11. OVERALL DIRECTION ACCURACY
# ============================================================

models = {
    "Random Forest + Lag":
        "RF_predicted_delta",

    "XGBoost + Lag":
        "XGB_predicted_delta",

    "50/50 Ensemble":
        "Ensemble_predicted_delta"
}


print("\n" + "=" * 75)
print("OVERALL DIRECTION ACCURACY")
print("=" * 75)


for model_name, column in models.items():

    accuracy = direction_accuracy(
        df["IV_change_5m"],
        df[column]
    )

    print(
        f"{model_name:<22}: "
        f"{accuracy:.2f}%"
    )


# ============================================================
# 12. THRESHOLDS TO TEST
# ============================================================

thresholds = [
    0.00,
    0.05,
    0.10,
    0.15,
    0.20,
    0.25,
    0.30,
    0.50,
    0.75,
    1.00,
    1.50,
    2.00
]


# ============================================================
# 13. THRESHOLD ANALYSIS
# ============================================================

threshold_results = []


for model_name, column in models.items():

    for threshold in thresholds:

        # Select only predictions whose predicted
        # magnitude is greater than the threshold.

        selected = df[
            df[column].abs()
            > threshold
        ].copy()


        signals = len(selected)


        if signals == 0:
            continue


        coverage = (
            signals
            / len(df)
            * 100
        )


        accuracy = direction_accuracy(
            selected["IV_change_5m"],
            selected[column]
        )


        mean_predicted_move = (
            selected[column]
            .abs()
            .mean()
        )


        mean_actual_move = (
            selected["IV_change_5m"]
            .abs()
            .mean()
        )


        median_actual_move = (
            selected["IV_change_5m"]
            .abs()
            .median()
        )


        threshold_results.append(
            {
                "Model": model_name,
                "Threshold": threshold,
                "Signals": signals,
                "Coverage_%": coverage,
                "Direction_Accuracy_%": accuracy,
                "Mean_Abs_Predicted_Move":
                    mean_predicted_move,
                "Mean_Abs_Actual_Move":
                    mean_actual_move,
                "Median_Abs_Actual_Move":
                    median_actual_move
            }
        )


threshold_table = pd.DataFrame(
    threshold_results
)


# ============================================================
# 14. PRINT ALL THRESHOLD RESULTS
# ============================================================

print("\n" + "=" * 75)
print("THRESHOLD / CONVICTION ANALYSIS")
print("=" * 75)


print(
    threshold_table.to_string(
        index=False,
        formatters={
            "Coverage_%":
                "{:.2f}".format,

            "Direction_Accuracy_%":
                "{:.2f}".format,

            "Mean_Abs_Predicted_Move":
                "{:.4f}".format,

            "Mean_Abs_Actual_Move":
                "{:.4f}".format,

            "Median_Abs_Actual_Move":
                "{:.4f}".format
        }
    )
)


# ============================================================
# 15. SPECIFIC 0.10 AND 0.25 RESULTS
# ============================================================

print("\n" + "=" * 75)
print("FOCUS: THRESHOLDS 0.10 AND 0.25")
print("=" * 75)


focus = threshold_table[
    threshold_table["Threshold"].isin(
        [0.10, 0.25]
    )
].copy()


print(
    focus.to_string(
        index=False,
        formatters={
            "Coverage_%":
                "{:.2f}".format,

            "Direction_Accuracy_%":
                "{:.2f}".format,

            "Mean_Abs_Predicted_Move":
                "{:.4f}".format,

            "Mean_Abs_Actual_Move":
                "{:.4f}".format,

            "Median_Abs_Actual_Move":
                "{:.4f}".format
        }
    )
)


# ============================================================
# 16. BEST THRESHOLDS WITH >= 10% COVERAGE
# ============================================================

reasonable = threshold_table[
    threshold_table["Coverage_%"]
    >= 10
].copy()


reasonable = reasonable.sort_values(
    [
        "Direction_Accuracy_%",
        "Coverage_%"
    ],
    ascending=[
        False,
        False
    ]
)


print("\n" + "=" * 75)
print("BEST RESULTS WITH AT LEAST 10% COVERAGE")
print("=" * 75)


print(
    reasonable
    .head(15)
    .to_string(
        index=False,
        formatters={
            "Coverage_%":
                "{:.2f}".format,

            "Direction_Accuracy_%":
                "{:.2f}".format,

            "Mean_Abs_Predicted_Move":
                "{:.4f}".format,

            "Mean_Abs_Actual_Move":
                "{:.4f}".format,

            "Median_Abs_Actual_Move":
                "{:.4f}".format
        }
    )
)


# ============================================================
# 17. UP VS DOWN ANALYSIS AT THRESHOLD 0.25
# ============================================================

diagnostic_threshold = 0.25


print("\n" + "=" * 75)
print("UP / DOWN ANALYSIS AT THRESHOLD 0.25")
print("=" * 75)


for model_name, column in models.items():

    selected = df[
        df[column].abs()
        > diagnostic_threshold
    ].copy()


    selected["Predicted_Direction"] = np.where(
        selected[column] > 0,
        "UP",
        "DOWN"
    )


    selected["Actual_Direction"] = np.where(
        selected["IV_change_5m"] > 0,
        "UP",
        np.where(
            selected["IV_change_5m"] < 0,
            "DOWN",
            "FLAT"
        )
    )


    print(
        f"\n{model_name}"
    )

    print(
        "Signals:",
        len(selected)
    )

    print(
        pd.crosstab(
            selected["Predicted_Direction"],
            selected["Actual_Direction"]
        )
    )


# ============================================================
# 18. SAVE OUTPUTS
# ============================================================

THRESHOLD_FILE = (
    PROCESSED_FOLDER /
    "lagged_threshold_analysis.csv"
)

ENSEMBLE_FILE = (
    PROCESSED_FOLDER /
    "lagged_ensemble_predictions.csv"
)


threshold_table.to_csv(
    THRESHOLD_FILE,
    index=False
)

df.to_csv(
    ENSEMBLE_FILE,
    index=False
)


print("\n" + "=" * 75)
print("FILES SAVED")
print("=" * 75)

print(THRESHOLD_FILE)
print(ENSEMBLE_FILE)
```

```python
# ============================================================
# STEP 15
# VALIDATE 19-OCT-2026 RAW DATA COLLECTION
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path


# ============================================================
# 1. DEFINE FOLDER
# ============================================================

BASE_DIR = Path.home() / "Downloads"

RAW_19_FOLDER = (
    BASE_DIR /
    "data" /
    "raw" /
    "2026-10-19"
)

LOG_FILE = (
    RAW_19_FOLDER /
    "collection_log_2026-10-19.csv"
)


print("=" * 75)
print("19-OCT DATA VALIDATION")
print("=" * 75)

print("Folder:")
print(RAW_19_FOLDER)

print(
    "\nFolder exists:",
    RAW_19_FOLDER.exists()
)


# ============================================================
# 2. FIND SNAPSHOT FILES
# ============================================================

snapshot_files = sorted(
    RAW_19_FOLDER.glob(
        "nifty_options_*.csv"
    )
)


print("\n" + "=" * 75)
print("SNAPSHOT FILES")
print("=" * 75)

print(
    "Number of snapshot CSV files:",
    len(snapshot_files)
)


if len(snapshot_files) > 0:

    print(
        "First file:",
        snapshot_files[0].name
    )

    print(
        "Last file :",
        snapshot_files[-1].name
    )


# ============================================================
# 3. READ COLLECTION LOG
# ============================================================

print("\n" + "=" * 75)
print("COLLECTION LOG")
print("=" * 75)


if LOG_FILE.exists():

    log = pd.read_csv(LOG_FILE)

    print(
        "Total log entries:",
        len(log)
    )

    print(
        "\nStatus counts:"
    )

    print(
        log["status"].value_counts(
            dropna=False
        )
    )


    success_log = log[
        log["status"] == "SUCCESS"
    ].copy()


    error_log = log[
        log["status"] != "SUCCESS"
    ].copy()


    print(
        "\nSuccessful snapshots:",
        len(success_log)
    )

    print(
        "Error entries:",
        len(error_log)
    )


    if len(error_log) > 0:

        print(
            "\nError rows:"
        )

        print(
            error_log.to_string(
                index=False
            )
        )

else:

    print(
        "Collection log was not found."
    )

    log = None


# ============================================================
# 4. LOAD ALL SNAPSHOT FILES
# ============================================================

frames = []

bad_files = []


for file in snapshot_files:

    try:

        temp = pd.read_csv(file)

        temp["source_file"] = (
            file.name
        )

        frames.append(temp)

    except Exception as e:

        bad_files.append(
            {
                "file": file.name,
                "error": str(e)
            }
        )


print("\n" + "=" * 75)
print("FILE LOADING CHECK")
print("=" * 75)

print(
    "Files loaded successfully:",
    len(frames)
)

print(
    "Files that failed:",
    len(bad_files)
)


if len(bad_files) > 0:

    print(
        pd.DataFrame(
            bad_files
        ).to_string(
            index=False
        )
    )


# ============================================================
# 5. STOP IF THERE IS NO DATA
# ============================================================

if len(frames) == 0:

    print(
        "\nNo usable 19-Oct snapshot data found."
    )

else:

    # ========================================================
    # 6. COMBINE ALL SNAPSHOTS
    # ========================================================

    raw_19 = pd.concat(
        frames,
        ignore_index=True
    )


    print("\n" + "=" * 75)
    print("COMBINED RAW DATA")
    print("=" * 75)

    print(
        "Total rows:",
        len(raw_19)
    )

    print(
        "Columns:",
        len(raw_19.columns)
    )


    # ========================================================
    # 7. TIMESTAMP CHECK
    # ========================================================

    raw_19["timestamp"] = pd.to_datetime(
        raw_19["timestamp"],
        utc=True,
        errors="coerce"
    )


    print("\n" + "=" * 75)
    print("TIMESTAMP CHECK")
    print("=" * 75)


    print(
        "Missing timestamps:",
        raw_19["timestamp"]
        .isna()
        .sum()
    )


    unique_times = (
        raw_19["timestamp"]
        .dropna()
        .drop_duplicates()
        .sort_values()
    )


    print(
        "Unique snapshot timestamps:",
        len(unique_times)
    )


    if len(unique_times) > 0:

        print(
            "First timestamp:",
            unique_times.iloc[0]
        )

        print(
            "Last timestamp :",
            unique_times.iloc[-1]
        )


    # ========================================================
    # 8. TIME GAP CHECK
    # ========================================================

    if len(unique_times) > 1:

        time_gaps = (
            unique_times
            .diff()
            .dt.total_seconds()
            .div(60)
            .dropna()
        )


        print("\n" + "=" * 75)
        print("TIME GAPS BETWEEN SNAPSHOTS")
        print("=" * 75)


        print(
            time_gaps
            .round(2)
            .value_counts()
            .sort_index()
            .to_string()
        )


        print(
            "\nLargest gap:",
            round(
                time_gaps.max(),
                2
            ),
            "minutes"
        )


    # ========================================================
    # 9. ROWS PER SNAPSHOT
    # ========================================================

    rows_per_snapshot = (
        raw_19
        .groupby("timestamp")
        .size()
    )


    print("\n" + "=" * 75)
    print("ROWS PER SNAPSHOT")
    print("=" * 75)


    print(
        rows_per_snapshot
        .describe()
    )


    print(
        "\nSnapshots not containing 104 rows:",
        (
            rows_per_snapshot != 104
        ).sum()
    )


    if (
        rows_per_snapshot != 104
    ).any():

        print(
            "\nUnusual snapshots:"
        )

        print(
            rows_per_snapshot[
                rows_per_snapshot != 104
            ]
        )


    # ========================================================
    # 10. STRIKE CHECK
    # ========================================================

    raw_19["strike"] = pd.to_numeric(
        raw_19["strike"],
        errors="coerce"
    )


    print("\n" + "=" * 75)
    print("STRIKE CHECK")
    print("=" * 75)


    print(
        "Unique strikes:",
        raw_19["strike"].nunique()
    )

    print(
        "Minimum strike:",
        raw_19["strike"].min()
    )

    print(
        "Maximum strike:",
        raw_19["strike"].max()
    )


    # ========================================================
    # 11. SPOT CHECK
    # ========================================================

    raw_19["spot"] = pd.to_numeric(
        raw_19["spot"],
        errors="coerce"
    )


    print("\n" + "=" * 75)
    print("SPOT CHECK")
    print("=" * 75)


    print(
        raw_19["spot"]
        .describe()
    )


    # ========================================================
    # 12. IV AVAILABILITY
    # ========================================================

    for col in [
        "CE_IV",
        "PE_IV"
    ]:

        raw_19[col] = pd.to_numeric(
            raw_19[col],
            errors="coerce"
        )


    print("\n" + "=" * 75)
    print("IV AVAILABILITY")
    print("=" * 75)


    print(
        "Total possible CE rows:",
        len(raw_19)
    )

    print(
        "Valid positive CE IV:",
        (
            raw_19["CE_IV"] > 0
        ).sum()
    )

    print(
        "Valid positive PE IV:",
        (
            raw_19["PE_IV"] > 0
        ).sum()
    )


    print(
        "\nCE IV missing/non-positive:",
        (
            ~(
                raw_19["CE_IV"] > 0
            )
        ).sum()
    )

    print(
        "PE IV missing/non-positive:",
        (
            ~(
                raw_19["PE_IV"] > 0
            )
        ).sum()
    )


    # ========================================================
    # 13. DUPLICATE CONTRACT CHECK
    # ========================================================

    duplicate_contracts = (
        raw_19.duplicated(
            subset=[
                "timestamp",
                "expiry",
                "strike"
            ]
        ).sum()
    )


    print("\n" + "=" * 75)
    print("DUPLICATE CHECK")
    print("=" * 75)


    print(
        "Duplicate timestamp-expiry-strike rows:",
        duplicate_contracts
    )


    # ========================================================
    # 14. EXPECTED ROW COUNT
    # ========================================================

    expected_rows = (
        len(snapshot_files)
        * 104
    )


    print("\n" + "=" * 75)
    print("ROW COUNT CONSISTENCY")
    print("=" * 75)


    print(
        "Expected rows if every snapshot has 104:",
        expected_rows
    )

    print(
        "Actual combined rows:",
        len(raw_19)
    )

    print(
        "Difference:",
        len(raw_19)
        - expected_rows
    )


    # ========================================================
    # 15. SAVE COMBINED RAW 19-OCT FILE
    # ========================================================

    PROCESSED_FOLDER = (
        BASE_DIR /
        "data" /
        "processed"
    )

    PROCESSED_FOLDER.mkdir(
        parents=True,
        exist_ok=True
    )


    OUTPUT_FILE = (
        PROCESSED_FOLDER /
        "nifty_options_19oct_raw_combined.csv"
    )


    raw_19.to_csv(
        OUTPUT_FILE,
        index=False
    )


    print("\n" + "=" * 75)
    print("SAVED")
    print("=" * 75)

    print(OUTPUT_FILE)
```

```python
# ============================================================
# STEP 16
# PREPARE 19-OCT UNSEEN ML DATASET
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path


# ============================================================
# 1. LOAD COMBINED 19-OCT RAW DATA
# ============================================================

BASE_DIR = Path.home() / "Downloads"
PROCESSED_FOLDER = BASE_DIR / "data" / "processed"

FILE = (
    PROCESSED_FOLDER /
    "nifty_options_19oct_raw_combined.csv"
)

raw = pd.read_csv(FILE)

raw["timestamp"] = pd.to_datetime(
    raw["timestamp"],
    utc=True,
    errors="coerce"
)

print("=" * 75)
print("19-OCT RAW DATA")
print("=" * 75)

print("Rows:", len(raw))
print(
    "Unique timestamps:",
    raw["timestamp"].nunique()
)


# ============================================================
# 2. CREATE CALL DATA
# ============================================================

calls = raw[
    [
        "timestamp",
        "expiry",
        "spot",
        "strike",
        "pcr",
        "CE_LTP",
        "CE_Volume",
        "CE_OI",
        "CE_Bid",
        "CE_Ask",
        "CE_IV"
    ]
].copy()


calls = calls.rename(
    columns={
        "CE_LTP": "LTP",
        "CE_Volume": "Volume",
        "CE_OI": "OI",
        "CE_Bid": "Bid",
        "CE_Ask": "Ask",
        "CE_IV": "IV"
    }
)

calls["option_type"] = "CE"


# ============================================================
# 3. CREATE PUT DATA
# ============================================================

puts = raw[
    [
        "timestamp",
        "expiry",
        "spot",
        "strike",
        "pcr",
        "PE_LTP",
        "PE_Volume",
        "PE_OI",
        "PE_Bid",
        "PE_Ask",
        "PE_IV"
    ]
].copy()


puts = puts.rename(
    columns={
        "PE_LTP": "LTP",
        "PE_Volume": "Volume",
        "PE_OI": "OI",
        "PE_Bid": "Bid",
        "PE_Ask": "Ask",
        "PE_IV": "IV"
    }
)

puts["option_type"] = "PE"


# ============================================================
# 4. COMBINE CE + PE
# ============================================================

df = pd.concat(
    [calls, puts],
    ignore_index=True
)


print("\n" + "=" * 75)
print("CE + PE RESTRUCTURING")
print("=" * 75)

print(
    "Rows before IV cleaning:",
    len(df)
)


# ============================================================
# 5. NUMERIC CONVERSION
# ============================================================

numeric_columns = [
    "spot",
    "strike",
    "pcr",
    "LTP",
    "Volume",
    "OI",
    "Bid",
    "Ask",
    "IV"
]


for col in numeric_columns:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )


# ============================================================
# 6. CLEAN ESSENTIAL VALUES
# ============================================================

df = df.dropna(
    subset=[
        "timestamp",
        "spot",
        "strike",
        "IV"
    ]
).copy()


# Keep only positive IV

df = df[
    df["IV"] > 0
].copy()


print(
    "Rows after valid IV filter:",
    len(df)
)

print(
    "\nOption counts:"
)

print(
    df["option_type"].value_counts()
)


# ============================================================
# 7. CREATE CURRENT FEATURES
# ============================================================

df["moneyness"] = (
    df["strike"]
    / df["spot"]
)


df["distance_from_ATM"] = (
    df["strike"]
    - df["spot"]
).abs()


df["distance_from_ATM_pct"] = (

    df["distance_from_ATM"]
    / df["spot"]

    * 100
)


df["bid_ask_spread"] = (
    df["Ask"]
    - df["Bid"]
)


df["mid_price"] = (
    df["Bid"]
    + df["Ask"]
) / 2


df["relative_spread_pct"] = np.where(

    df["mid_price"] > 0,

    (
        df["bid_ask_spread"]
        / df["mid_price"]
    ) * 100,

    np.nan
)


df["option_type_code"] = (
    df["option_type"]
    .map(
        {
            "CE": 0,
            "PE": 1
        }
    )
)


# ============================================================
# 8. DAYS TO EXPIRY
# ============================================================

df["expiry"] = pd.to_datetime(
    df["expiry"],
    errors="coerce"
)


# 19-Oct is expiry day
# so this should equal 0.

df["days_to_expiry"] = (
    df["expiry"].dt.date
    -
    df["timestamp"].dt.date
).apply(
    lambda x:
    x.days if pd.notna(x)
    else np.nan
)


# ============================================================
# 9. ROUND COLLECTION TIMESTAMP TO MINUTE
# ============================================================

df["time_5m"] = (
    df["timestamp"]
    .dt.round("min")
)


print("\n" + "=" * 75)
print("CURRENT FEATURE DATA")
print("=" * 75)

print(
    "Rows:",
    len(df)
)

print(
    "Unique timestamps:",
    df["time_5m"].nunique()
)

print(
    "Days to expiry values:",
    sorted(
        df["days_to_expiry"]
        .dropna()
        .unique()
    )
)


# ============================================================
# 10. NEAR-ATM FILTER
# ============================================================

# EXACT same economic filter used during development.

near_atm = df[
    (df["moneyness"] >= 0.95)
    &
    (df["moneyness"] <= 1.05)
].copy()


print("\n" + "=" * 75)
print("NEAR-ATM FILTER")
print("=" * 75)

print(
    "Rows before:",
    len(df)
)

print(
    "Rows after:",
    len(near_atm)
)

print(
    "Retained:",
    round(
        len(near_atm)
        / len(df)
        * 100,
        2
    ),
    "%"
)

print(
    "Unique timestamps:",
    near_atm["time_5m"].nunique()
)

print(
    "Unique strikes:",
    near_atm["strike"].nunique()
)

print(
    "\nOption counts:"
)

print(
    near_atm["option_type"]
    .value_counts()
)


# ============================================================
# 11. CHECK DUPLICATES BEFORE TARGET CREATION
# ============================================================

duplicates = near_atm.duplicated(
    subset=[
        "time_5m",
        "expiry",
        "strike",
        "option_type"
    ]
).sum()


print("\nDuplicate contract/time rows:", duplicates)


# ============================================================
# 12. CREATE EXACT +5 MINUTE TARGET
# ============================================================

future = near_atm[
    [
        "time_5m",
        "expiry",
        "strike",
        "option_type",
        "IV"
    ]
].copy()


future["target_time_5m"] = (
    future["time_5m"]
)


future = future.rename(
    columns={
        "IV":
        "IV_t_plus_5"
    }
)


# For current row at time t,
# search for IV at t + 5 minutes.

near_atm["target_time_5m"] = (
    near_atm["time_5m"]
    + pd.Timedelta(minutes=5)
)


ml = pd.merge(
    near_atm,
    future[
        [
            "target_time_5m",
            "expiry",
            "strike",
            "option_type",
            "IV_t_plus_5"
        ]
    ],
    on=[
        "target_time_5m",
        "expiry",
        "strike",
        "option_type"
    ],
    how="left"
)


# ============================================================
# 13. CREATE ACTUAL FUTURE ΔIV
# ============================================================

ml["IV_change_5m"] = (
    ml["IV_t_plus_5"]
    - ml["IV"]
)


ml["IV_change_pct_5m"] = np.where(

    ml["IV"] != 0,

    (
        ml["IV_change_5m"]
        / ml["IV"]
    ) * 100,

    np.nan
)


target_available = (
    ml["IV_t_plus_5"]
    .notna()
    .sum()
)


print("\n" + "=" * 75)
print("EXACT +5 MINUTE TARGET")
print("=" * 75)

print(
    "Rows before target requirement:",
    len(ml)
)

print(
    "Rows with exact +5m target:",
    target_available
)


# Keep rows where future IV exists.

ml = ml[
    ml["IV_t_plus_5"].notna()
].copy()


print(
    "Rows retained:",
    len(ml)
)

print(
    "Unique input timestamps:",
    ml["time_5m"].nunique()
)


# ============================================================
# 14. CREATE EXACT LAG FEATURES
# ============================================================

lag_variables = [
    "IV",
    "spot",
    "LTP",
    "Volume",
    "OI",
    "bid_ask_spread"
]


# IMPORTANT:
# Build lag source from near_atm rather than ml.
# This preserves valid historical rows even if those rows
# themselves do not have a future target.

history_source = near_atm.copy()


# ============================================================
# 15. EXACT 5-MINUTE LAG
# ============================================================

lag5 = history_source[
    [
        "time_5m",
        "expiry",
        "strike",
        "option_type"
    ]
    + lag_variables
].copy()


lag5["time_5m"] = (
    lag5["time_5m"]
    + pd.Timedelta(minutes=5)
)


lag5 = lag5.rename(
    columns={
        col:
        f"{col}_lag_5m"

        for col in lag_variables
    }
)


ml = pd.merge(
    ml,
    lag5,
    on=[
        "time_5m",
        "expiry",
        "strike",
        "option_type"
    ],
    how="left"
)


# ============================================================
# 16. EXACT 10-MINUTE LAG
# ============================================================

lag10 = history_source[
    [
        "time_5m",
        "expiry",
        "strike",
        "option_type"
    ]
    + lag_variables
].copy()


lag10["time_5m"] = (
    lag10["time_5m"]
    + pd.Timedelta(minutes=10)
)


lag10 = lag10.rename(
    columns={
        col:
        f"{col}_lag_10m"

        for col in lag_variables
    }
)


ml = pd.merge(
    ml,
    lag10,
    on=[
        "time_5m",
        "expiry",
        "strike",
        "option_type"
    ],
    how="left"
)


# ============================================================
# 17. CREATE MOMENTUM FEATURES
# ============================================================

for col in lag_variables:

    ml[
        f"{col}_change_prev_5m"
    ] = (
        ml[col]
        - ml[f"{col}_lag_5m"]
    )


    ml[
        f"{col}_change_prev_10m"
    ] = (
        ml[col]
        - ml[f"{col}_lag_10m"]
    )


# ============================================================
# 18. CHECK LAG AVAILABILITY
# ============================================================

has_5m = (
    ml["IV_lag_5m"]
    .notna()
)

has_10m = (
    ml["IV_lag_10m"]
    .notna()
)


print("\n" + "=" * 75)
print("LAG AVAILABILITY")
print("=" * 75)

print(
    "Rows before lag requirement:",
    len(ml)
)

print(
    "Rows with exact 5m history:",
    has_5m.sum()
)

print(
    "Rows with exact 10m history:",
    has_10m.sum()
)

print(
    "Rows with BOTH:",
    (
        has_5m
        & has_10m
    ).sum()
)


# ============================================================
# 19. REQUIRE BOTH IV LAGS
# ============================================================

final_19 = ml[
    has_5m
    &
    has_10m
].copy()


print("\n" + "=" * 75)
print("FINAL 19-OCT UNSEEN DATASET")
print("=" * 75)

print(
    "Rows:",
    len(final_19)
)

print(
    "Unique timestamps:",
    final_19["time_5m"].nunique()
)

print(
    "Unique strikes:",
    final_19["strike"].nunique()
)

print(
    "\nOption counts:"
)

print(
    final_19["option_type"]
    .value_counts()
)


# ============================================================
# 20. VERIFY ALL 39 MODEL FEATURES
# ============================================================

original_features = [
    "IV",
    "spot",
    "strike",
    "moneyness",
    "distance_from_ATM_pct",
    "LTP",
    "Volume",
    "OI",
    "Bid",
    "Ask",
    "bid_ask_spread",
    "relative_spread_pct",
    "pcr",
    "days_to_expiry",
    "option_type_code"
]


lag_features = [
    "IV_lag_5m",
    "IV_lag_10m",
    "IV_change_prev_5m",
    "IV_change_prev_10m",

    "spot_lag_5m",
    "spot_lag_10m",
    "spot_change_prev_5m",
    "spot_change_prev_10m",

    "LTP_lag_5m",
    "LTP_lag_10m",
    "LTP_change_prev_5m",
    "LTP_change_prev_10m",

    "Volume_lag_5m",
    "Volume_lag_10m",
    "Volume_change_prev_5m",
    "Volume_change_prev_10m",

    "OI_lag_5m",
    "OI_lag_10m",
    "OI_change_prev_5m",
    "OI_change_prev_10m",

    "bid_ask_spread_lag_5m",
    "bid_ask_spread_lag_10m",
    "bid_ask_spread_change_prev_5m",
    "bid_ask_spread_change_prev_10m"
]


model_features = (
    original_features
    + lag_features
)


missing_columns = [
    col
    for col in model_features
    if col not in final_19.columns
]


print("\n" + "=" * 75)
print("MODEL FEATURE CHECK")
print("=" * 75)

print(
    "Required model features:",
    len(model_features)
)

print(
    "Missing feature columns:",
    missing_columns
)


if len(missing_columns) == 0:

    missing_values = (
        final_19[
            model_features
        ]
        .isna()
        .sum()
    )

    missing_values = (
        missing_values[
            missing_values > 0
        ]
    )

    print(
        "\nFeatures containing missing values:"
    )

    if len(missing_values) == 0:
        print("None")
    else:
        print(missing_values)


# ============================================================
# 21. IV CHANGE SUMMARY
# ============================================================

print("\n" + "=" * 75)
print("ACTUAL 5-MINUTE IV CHANGE SUMMARY")
print("=" * 75)

print(
    final_19["IV_change_5m"]
    .describe()
)


# ============================================================
# 22. SAVE FINAL UNSEEN DATASET
# ============================================================

OUTPUT_FILE = (
    PROCESSED_FOLDER /
    "nifty_options_19oct_ml_lagged.csv"
)


final_19.to_csv(
    OUTPUT_FILE,
    index=False
)


print("\n" + "=" * 75)
print("SAVED")
print("=" * 75)

print(OUTPUT_FILE)
```

```python
# ============================================================
# STEP 17
# DIAGNOSE 19-OCT EXTREME IV MOVEMENTS
# CORRECTED FULL VERSION
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path


# ============================================================
# 1. LOAD 19-OCT ML DATASET
# ============================================================

BASE_DIR = Path.home() / "Downloads"

FILE = (
    BASE_DIR
    / "data"
    / "processed"
    / "nifty_options_19oct_ml_lagged.csv"
)


df = pd.read_csv(FILE)


# Convert timestamp

df["time_5m"] = pd.to_datetime(
    df["time_5m"],
    utc=True,
    errors="coerce"
)


# Create absolute IV change

df["abs_IV_change_5m"] = (
    df["IV_change_5m"].abs()
)


print("=" * 80)
print("FILE LOADED")
print("=" * 80)

print("Rows:", len(df))

print(
    "Unique timestamps:",
    df["time_5m"].nunique()
)

print(
    "Option counts:"
)

print(
    df["option_type"].value_counts()
)


# ============================================================
# 2. BASIC IV CHANGE DISTRIBUTION
# ============================================================

print("\n" + "=" * 80)
print("19-OCT IV CHANGE DISTRIBUTION")
print("=" * 80)


print(
    df["IV_change_5m"].describe(
        percentiles=[
            0.01,
            0.05,
            0.10,
            0.25,
            0.50,
            0.75,
            0.90,
            0.95,
            0.99
        ]
    )
)


# ============================================================
# 3. ABSOLUTE IV CHANGE QUANTILES
# ============================================================

print("\n" + "=" * 80)
print("ABSOLUTE IV CHANGE QUANTILES")
print("=" * 80)


quantiles = [
    0.50,
    0.75,
    0.90,
    0.95,
    0.99,
    0.995,
    0.999,
    1.00
]


for q in quantiles:

    value = (
        df["abs_IV_change_5m"]
        .quantile(q)
    )

    print(
        f"{q * 100:6.1f}% : "
        f"{value:.4f}"
    )


# ============================================================
# 4. EXTREME MOVEMENT COUNTS
# ============================================================

print("\n" + "=" * 80)
print("EXTREME MOVEMENT COUNTS")
print("=" * 80)


thresholds = [
    1,
    2,
    5,
    10,
    20,
    30,
    50,
    100,
    200
]


for threshold in thresholds:

    count = (
        df["abs_IV_change_5m"]
        > threshold
    ).sum()

    percentage = (
        count
        / len(df)
        * 100
    )

    print(
        f"|ΔIV| > {threshold:3} : "
        f"{count:4} rows "
        f"({percentage:.2f}%)"
    )


# ============================================================
# 5. IV CHANGE BY OPTION TYPE
# ============================================================

print("\n" + "=" * 80)
print("IV CHANGE BY OPTION TYPE")
print("=" * 80)


print(
    df.groupby(
        "option_type"
    )["IV_change_5m"]
    .describe()
)


# ============================================================
# 6. COLUMNS WE WANT TO DISPLAY
# ============================================================

columns_to_show = [
    "time_5m",
    "option_type",
    "strike",
    "spot",
    "moneyness",
    "IV",
    "IV_t_plus_5",
    "IV_change_5m",
    "LTP",
    "Volume",
    "OI",
    "Bid",
    "Ask",
    "bid_ask_spread",
    "relative_spread_pct"
]


# ============================================================
# 7. TOP 30 EXTREME OBSERVATIONS
# ============================================================

extreme = (
    df
    .sort_values(
        "abs_IV_change_5m",
        ascending=False
    )
    .head(30)
    .copy()
)


print("\n" + "=" * 80)
print("30 LARGEST ABSOLUTE IV CHANGES")
print("=" * 80)


print(
    extreme[
        columns_to_show
    ].to_string(
        index=False
    )
)


# ============================================================
# 8. CURRENT IV DISTRIBUTION
# ============================================================

print("\n" + "=" * 80)
print("CURRENT IV DISTRIBUTION")
print("=" * 80)


print(
    df["IV"].describe(
        percentiles=[
            0.50,
            0.75,
            0.90,
            0.95,
            0.99
        ]
    )
)


# ============================================================
# 9. FUTURE IV DISTRIBUTION
# ============================================================

print("\n" + "=" * 80)
print("FUTURE IV DISTRIBUTION")
print("=" * 80)


print(
    df["IV_t_plus_5"].describe(
        percentiles=[
            0.50,
            0.75,
            0.90,
            0.95,
            0.99
        ]
    )
)


# ============================================================
# 10. HIGH IV OBSERVATIONS
# ============================================================

high_iv = df[
    (df["IV"] > 100)
    |
    (df["IV_t_plus_5"] > 100)
].copy()


print("\n" + "=" * 80)
print("ROWS WHERE CURRENT OR FUTURE IV > 100")
print("=" * 80)


print(
    "Number of rows:",
    len(high_iv)
)


if len(high_iv) > 0:

    # IMPORTANT FIX:
    # Sort BEFORE selecting columns_to_show.

    high_iv_sorted = (
        high_iv
        .sort_values(
            "abs_IV_change_5m",
            ascending=False
        )
        .head(40)
    )

    print(
        high_iv_sorted[
            columns_to_show
        ].to_string(
            index=False
        )
    )


# ============================================================
# 11. CHECK LOW-PRICE OPTIONS
# ============================================================

# Very low option prices can produce unstable IV,
# especially close to expiry.

low_price = df[
    df["LTP"] <= 1
].copy()


print("\n" + "=" * 80)
print("VERY LOW-PRICE OPTIONS (LTP <= ₹1)")
print("=" * 80)


print(
    "Number of rows:",
    len(low_price)
)

print(
    "Percentage:",
    round(
        len(low_price)
        / len(df)
        * 100,
        2
    ),
    "%"
)


if len(low_price) > 0:

    print(
        "\nIV change for low-price options:"
    )

    print(
        low_price[
            "IV_change_5m"
        ].describe()
    )


# ============================================================
# 12. CHECK HIGH RELATIVE SPREAD
# ============================================================

print("\n" + "=" * 80)
print("RELATIVE SPREAD DISTRIBUTION")
print("=" * 80)


print(
    df["relative_spread_pct"]
    .describe(
        percentiles=[
            0.50,
            0.75,
            0.90,
            0.95,
            0.99
        ]
    )
)


# ============================================================
# 13. IV MOVEMENT BY TIMESTAMP
# ============================================================

time_summary = (
    df
    .groupby(
        "time_5m"
    )
    .agg(

        observations=(
            "IV_change_5m",
            "size"
        ),

        mean_delta=(
            "IV_change_5m",
            "mean"
        ),

        median_delta=(
            "IV_change_5m",
            "median"
        ),

        mean_abs_delta=(
            "abs_IV_change_5m",
            "mean"
        ),

        max_abs_delta=(
            "abs_IV_change_5m",
            "max"
        ),

        mean_IV=(
            "IV",
            "mean"
        ),

        median_IV=(
            "IV",
            "median"
        )
    )
    .reset_index()
)


print("\n" + "=" * 80)
print("IV MOVEMENT BY TIMESTAMP")
print("=" * 80)


print(
    time_summary.to_string(
        index=False
    )
)


# ============================================================
# 14. EXTREME MOVEMENTS BY TIMESTAMP
# ============================================================

extreme_by_time = (
    df.assign(
        extreme_20=(
            df["abs_IV_change_5m"] > 20
        ),
        extreme_50=(
            df["abs_IV_change_5m"] > 50
        ),
        extreme_100=(
            df["abs_IV_change_5m"] > 100
        )
    )
    .groupby(
        "time_5m"
    )
    .agg(

        total_rows=(
            "IV_change_5m",
            "size"
        ),

        above_20=(
            "extreme_20",
            "sum"
        ),

        above_50=(
            "extreme_50",
            "sum"
        ),

        above_100=(
            "extreme_100",
            "sum"
        )
    )
    .reset_index()
)


print("\n" + "=" * 80)
print("EXTREME MOVEMENTS BY TIMESTAMP")
print("=" * 80)


print(
    extreme_by_time.to_string(
        index=False
    )
)


# ============================================================
# 15. CHECK LAST PART OF TRADING SESSION
# ============================================================

print("\n" + "=" * 80)
print("LAST 5 TIMESTAMPS")
print("=" * 80)


last_times = (
    time_summary
    .sort_values(
        "time_5m"
    )
    .tail(5)
)


print(
    last_times.to_string(
        index=False
    )
)


# ============================================================
# 16. SAVE TOP EXTREME OBSERVATIONS
# ============================================================

EXTREME_FILE = (
    BASE_DIR
    / "data"
    / "processed"
    / "19oct_iv_extreme_diagnostics.csv"
)


extreme.to_csv(
    EXTREME_FILE,
    index=False
)


# ============================================================
# 17. SAVE TIMESTAMP SUMMARY
# ============================================================

TIME_FILE = (
    BASE_DIR
    / "data"
    / "processed"
    / "19oct_iv_timestamp_diagnostics.csv"
)


time_summary.to_csv(
    TIME_FILE,
    index=False
)


# ============================================================
# 18. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("FINAL DIAGNOSTIC SUMMARY")
print("=" * 80)


print(
    "Total observations:",
    len(df)
)

print(
    "Unique timestamps:",
    df["time_5m"].nunique()
)

print(
    "Median current IV:",
    round(
        df["IV"].median(),
        4
    )
)

print(
    "Maximum current IV:",
    round(
        df["IV"].max(),
        4
    )
)

print(
    "Median absolute 5-min IV change:",
    round(
        df["abs_IV_change_5m"].median(),
        4
    )
)

print(
    "Maximum absolute 5-min IV change:",
    round(
        df["abs_IV_change_5m"].max(),
        4
    )
)

print(
    "Rows with |ΔIV| > 20:",
    (
        df["abs_IV_change_5m"] > 20
    ).sum()
)

print(
    "Rows with |ΔIV| > 50:",
    (
        df["abs_IV_change_5m"] > 50
    ).sum()
)

print(
    "Rows with |ΔIV| > 100:",
    (
        df["abs_IV_change_5m"] > 100
    ).sum()
)

print(
    "Rows where current/future IV > 100:",
    len(high_iv)
)

print(
    "Rows with LTP <= ₹1:",
    len(low_price)
)


print("\n" + "=" * 80)
print("FILES SAVED")
print("=" * 80)

print(EXTREME_FILE)
print(TIME_FILE)
```

```python
# ============================================================
# STEP 18
# FINAL RANDOM FOREST
# TRAIN: 15–16 OCT DEVELOPMENT DATA
# TEST : 19-OCT EXPIRY-DAY DATA
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ============================================================
# 1. FILE PATHS
# ============================================================

BASE_DIR = Path.home() / "Downloads"
PROCESSED_FOLDER = BASE_DIR / "data" / "processed"

TRAIN_FILE = (
    PROCESSED_FOLDER /
    "nifty_options_ml_5min_lagged.csv"
)

TEST_FILE = (
    PROCESSED_FOLDER /
    "nifty_options_19oct_ml_lagged.csv"
)


# ============================================================
# 2. LOAD DATA
# ============================================================

train_df = pd.read_csv(TRAIN_FILE)
test_df = pd.read_csv(TEST_FILE)


train_df["time_5m"] = pd.to_datetime(
    train_df["time_5m"],
    utc=True,
    errors="coerce"
)

test_df["time_5m"] = pd.to_datetime(
    test_df["time_5m"],
    utc=True,
    errors="coerce"
)


print("=" * 80)
print("DATA LOADED")
print("=" * 80)

print("Development rows :", len(train_df))
print("19-Oct test rows :", len(test_df))

print(
    "Development timestamps:",
    train_df["time_5m"].nunique()
)

print(
    "19-Oct timestamps:",
    test_df["time_5m"].nunique()
)

print(
    "\nDevelopment period:",
    train_df["time_5m"].min(),
    "to",
    train_df["time_5m"].max()
)

print(
    "19-Oct period:",
    test_df["time_5m"].min(),
    "to",
    test_df["time_5m"].max()
)


# ============================================================
# 3. SAME 39 FEATURES USED IN STEP 12
# ============================================================

original_features = [
    "IV",
    "spot",
    "strike",
    "moneyness",
    "distance_from_ATM_pct",
    "LTP",
    "Volume",
    "OI",
    "Bid",
    "Ask",
    "bid_ask_spread",
    "relative_spread_pct",
    "pcr",
    "days_to_expiry",
    "option_type_code"
]


lag_features = [
    "IV_lag_5m",
    "IV_lag_10m",
    "IV_change_prev_5m",
    "IV_change_prev_10m",

    "spot_lag_5m",
    "spot_lag_10m",
    "spot_change_prev_5m",
    "spot_change_prev_10m",

    "LTP_lag_5m",
    "LTP_lag_10m",
    "LTP_change_prev_5m",
    "LTP_change_prev_10m",

    "Volume_lag_5m",
    "Volume_lag_10m",
    "Volume_change_prev_5m",
    "Volume_change_prev_10m",

    "OI_lag_5m",
    "OI_lag_10m",
    "OI_change_prev_5m",
    "OI_change_prev_10m",

    "bid_ask_spread_lag_5m",
    "bid_ask_spread_lag_10m",
    "bid_ask_spread_change_prev_5m",
    "bid_ask_spread_change_prev_10m"
]


features = (
    original_features
    + lag_features
)


TARGET = "IV_change_5m"


print("\n" + "=" * 80)
print("FEATURE CHECK")
print("=" * 80)

print("Number of features:", len(features))


missing_train_columns = [
    col for col in features
    if col not in train_df.columns
]

missing_test_columns = [
    col for col in features
    if col not in test_df.columns
]


print(
    "Missing development columns:",
    missing_train_columns
)

print(
    "Missing 19-Oct columns:",
    missing_test_columns
)


if len(missing_train_columns) > 0:
    raise ValueError(
        "Development dataset is missing required features."
    )

if len(missing_test_columns) > 0:
    raise ValueError(
        "19-Oct dataset is missing required features."
    )


# ============================================================
# 4. PREPARE X AND y
# ============================================================

X_train = train_df[features].copy()
y_train = train_df[TARGET].copy()

X_test = test_df[features].copy()
y_test = test_df[TARGET].copy()


# ============================================================
# 5. HANDLE MISSING VALUES
# ============================================================

# IMPORTANT:
# Medians are calculated ONLY from development data.
# We do NOT calculate medians using 19-Oct.

train_medians = X_train.median(
    numeric_only=True
)

X_train = X_train.fillna(
    train_medians
)

X_test = X_test.fillna(
    train_medians
)


print("\nMissing values after filling:")

print(
    "Development:",
    X_train.isna().sum().sum()
)

print(
    "19-Oct:",
    X_test.isna().sum().sum()
)


# ============================================================
# 6. TRAIN FROZEN RANDOM FOREST
# ============================================================

# SAME architecture used previously.
# We are NOT tuning this using 19-Oct.

rf_final = RandomForestRegressor(
    n_estimators=300,
    max_depth=12,
    min_samples_leaf=3,
    max_features="sqrt",
    random_state=42,
    n_jobs=-1
)


rf_final.fit(
    X_train,
    y_train
)


print("\n" + "=" * 80)
print("FINAL RF TRAINING")
print("=" * 80)

print(
    "Random Forest trained on:",
    len(X_train),
    "development observations"
)

print(
    "19-Oct observations used in training: 0"
)


# ============================================================
# 7. PREDICT 19-OCT ΔIV
# ============================================================

test_df["RF_predicted_delta"] = (
    rf_final.predict(
        X_test
    )
)


# Predicted future IV

test_df["RF_predicted_future_IV"] = (
    test_df["IV"]
    +
    test_df["RF_predicted_delta"]
)


# Persistence forecast:
# Future IV = Current IV

test_df["Persistence_future_IV"] = (
    test_df["IV"]
)


# ============================================================
# 8. PERSISTENCE METRICS
# ============================================================

persistence_mae = mean_absolute_error(
    test_df["IV_t_plus_5"],
    test_df["Persistence_future_IV"]
)

persistence_rmse = np.sqrt(
    mean_squared_error(
        test_df["IV_t_plus_5"],
        test_df["Persistence_future_IV"]
    )
)

persistence_r2 = r2_score(
    test_df["IV_t_plus_5"],
    test_df["Persistence_future_IV"]
)


# ============================================================
# 9. RANDOM FOREST METRICS
# ============================================================

rf_mae = mean_absolute_error(
    test_df["IV_t_plus_5"],
    test_df["RF_predicted_future_IV"]
)

rf_rmse = np.sqrt(
    mean_squared_error(
        test_df["IV_t_plus_5"],
        test_df["RF_predicted_future_IV"]
    )
)

rf_r2 = r2_score(
    test_df["IV_t_plus_5"],
    test_df["RF_predicted_future_IV"]
)


# ============================================================
# 10. IMPROVEMENT VS PERSISTENCE
# ============================================================

mae_improvement = (
    (
        persistence_mae
        - rf_mae
    )
    / persistence_mae
    * 100
)

rmse_improvement = (
    (
        persistence_rmse
        - rf_rmse
    )
    / persistence_rmse
    * 100
)


print("\n" + "=" * 80)
print("19-OCT EXPIRY-DAY FORECAST RESULTS")
print("=" * 80)


comparison = pd.DataFrame(
    {
        "Model": [
            "Persistence",
            "Random Forest + Lag"
        ],

        "MAE": [
            persistence_mae,
            rf_mae
        ],

        "RMSE": [
            persistence_rmse,
            rf_rmse
        ],

        "R2": [
            persistence_r2,
            rf_r2
        ]
    }
)


print(
    comparison.to_string(
        index=False
    )
)


print(
    "\nRF MAE improvement vs persistence:",
    round(mae_improvement, 2),
    "%"
)

print(
    "RF RMSE improvement vs persistence:",
    round(rmse_improvement, 2),
    "%"
)


# ============================================================
# 11. OVERALL DIRECTION ACCURACY
# ============================================================

actual_direction = np.sign(
    test_df["IV_change_5m"]
)

predicted_direction = np.sign(
    test_df["RF_predicted_delta"]
)


test_df["direction_correct"] = (
    actual_direction
    ==
    predicted_direction
)


overall_direction_accuracy = (
    test_df["direction_correct"]
    .mean()
    * 100
)


print("\n" + "=" * 80)
print("OVERALL DIRECTION ACCURACY")
print("=" * 80)

print(
    "Random Forest + Lag:",
    round(
        overall_direction_accuracy,
        2
    ),
    "%"
)


# ============================================================
# 12. THRESHOLD ANALYSIS
# ============================================================

# These are the candidate thresholds we already identified
# BEFORE looking at 19-Oct model performance.

thresholds = [
    0.00,
    0.05,
    0.10
]


threshold_results = []


for threshold in thresholds:

    selected = test_df[
        test_df[
            "RF_predicted_delta"
        ].abs() > threshold
    ].copy()


    signals = len(selected)

    coverage = (
        signals
        / len(test_df)
        * 100
    )


    if signals > 0:

        direction_accuracy = (
            np.sign(
                selected[
                    "IV_change_5m"
                ]
            )
            ==
            np.sign(
                selected[
                    "RF_predicted_delta"
                ]
            )
        ).mean() * 100


        mean_abs_predicted = (
            selected[
                "RF_predicted_delta"
            ]
            .abs()
            .mean()
        )


        mean_abs_actual = (
            selected[
                "IV_change_5m"
            ]
            .abs()
            .mean()
        )


        median_abs_actual = (
            selected[
                "IV_change_5m"
            ]
            .abs()
            .median()
        )


    else:

        direction_accuracy = np.nan
        mean_abs_predicted = np.nan
        mean_abs_actual = np.nan
        median_abs_actual = np.nan


    threshold_results.append(
        {
            "Threshold":
                threshold,

            "Signals":
                signals,

            "Coverage_%":
                coverage,

            "Direction_Accuracy_%":
                direction_accuracy,

            "Mean_Abs_Predicted_Move":
                mean_abs_predicted,

            "Mean_Abs_Actual_Move":
                mean_abs_actual,

            "Median_Abs_Actual_Move":
                median_abs_actual
        }
    )


threshold_df = pd.DataFrame(
    threshold_results
)


print("\n" + "=" * 80)
print("19-OCT THRESHOLD RESULTS")
print("=" * 80)


print(
    threshold_df
    .round(4)
    .to_string(
        index=False
    )
)


# ============================================================
# 13. UP / DOWN SIGNAL CONFUSION TABLE
# THRESHOLD = 0.05
# ============================================================

for threshold in [
    0.05,
    0.10
]:

    selected = test_df[
        test_df[
            "RF_predicted_delta"
        ].abs() > threshold
    ].copy()


    selected[
        "Predicted_Direction"
    ] = np.where(
        selected[
            "RF_predicted_delta"
        ] > 0,
        "UP",
        "DOWN"
    )


    selected[
        "Actual_Direction"
    ] = np.where(
        selected[
            "IV_change_5m"
        ] > 0,
        "UP",

        np.where(
            selected[
                "IV_change_5m"
            ] < 0,
            "DOWN",
            "FLAT"
        )
    )


    print("\n" + "=" * 80)

    print(
        f"UP / DOWN ANALYSIS — "
        f"THRESHOLD {threshold}"
    )

    print("=" * 80)

    print(
        "Signals:",
        len(selected)
    )

    print(
        pd.crosstab(
            selected[
                "Predicted_Direction"
            ],
            selected[
                "Actual_Direction"
            ]
        )
    )


# ============================================================
# 14. PERFORMANCE BY OPTION TYPE
# ============================================================

option_results = []


for option_type in [
    "CE",
    "PE"
]:

    subset = test_df[
        test_df["option_type"]
        == option_type
    ].copy()


    if len(subset) == 0:
        continue


    mae = mean_absolute_error(
        subset["IV_t_plus_5"],
        subset[
            "RF_predicted_future_IV"
        ]
    )


    rmse = np.sqrt(
        mean_squared_error(
            subset["IV_t_plus_5"],
            subset[
                "RF_predicted_future_IV"
            ]
        )
    )


    direction = (
        np.sign(
            subset["IV_change_5m"]
        )
        ==
        np.sign(
            subset["RF_predicted_delta"]
        )
    ).mean() * 100


    option_results.append(
        {
            "Option_Type":
                option_type,

            "Observations":
                len(subset),

            "MAE":
                mae,

            "RMSE":
                rmse,

            "Direction_Accuracy_%":
                direction
        }
    )


option_df = pd.DataFrame(
    option_results
)


print("\n" + "=" * 80)
print("PERFORMANCE BY OPTION TYPE")
print("=" * 80)


print(
    option_df
    .round(4)
    .to_string(
        index=False
    )
)


# ============================================================
# 15. PERFORMANCE BY TIMESTAMP
# ============================================================

timestamp_results = []


for timestamp, subset in test_df.groupby(
    "time_5m"
):

    mae = mean_absolute_error(
        subset["IV_t_plus_5"],
        subset[
            "RF_predicted_future_IV"
        ]
    )


    rmse = np.sqrt(
        mean_squared_error(
            subset["IV_t_plus_5"],
            subset[
                "RF_predicted_future_IV"
            ]
        )
    )


    direction = (
        np.sign(
            subset["IV_change_5m"]
        )
        ==
        np.sign(
            subset["RF_predicted_delta"]
        )
    ).mean() * 100


    timestamp_results.append(
        {
            "time_5m":
                timestamp,

            "Observations":
                len(subset),

            "Actual_Mean_Abs_Delta":
                subset[
                    "IV_change_5m"
                ].abs().mean(),

            "RF_MAE":
                mae,

            "RF_RMSE":
                rmse,

            "Direction_Accuracy_%":
                direction
        }
    )


timestamp_df = pd.DataFrame(
    timestamp_results
)


print("\n" + "=" * 80)
print("PERFORMANCE BY TIMESTAMP")
print("=" * 80)


print(
    timestamp_df
    .round(4)
    .to_string(
        index=False
    )
)


# ============================================================
# 16. PREDICTION DISTRIBUTION
# ============================================================

print("\n" + "=" * 80)
print("RF PREDICTED ΔIV DISTRIBUTION")
print("=" * 80)


print(
    test_df[
        "RF_predicted_delta"
    ].describe()
)


print("\n" + "=" * 80)
print("ACTUAL ΔIV DISTRIBUTION")
print("=" * 80)


print(
    test_df[
        "IV_change_5m"
    ].describe()
)


# ============================================================
# 17. SAVE PREDICTIONS
# ============================================================

PREDICTION_FILE = (
    PROCESSED_FOLDER /
    "rf_19oct_expiry_day_predictions.csv"
)


test_df.to_csv(
    PREDICTION_FILE,
    index=False
)


# ============================================================
# 18. SAVE THRESHOLD RESULTS
# ============================================================

THRESHOLD_FILE = (
    PROCESSED_FOLDER /
    "rf_19oct_threshold_results.csv"
)


threshold_df.to_csv(
    THRESHOLD_FILE,
    index=False
)


# ============================================================
# 19. SAVE TIMESTAMP RESULTS
# ============================================================

TIMESTAMP_FILE = (
    PROCESSED_FOLDER /
    "rf_19oct_timestamp_performance.csv"
)


timestamp_df.to_csv(
    TIMESTAMP_FILE,
    index=False
)


# ============================================================
# 20. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("STEP 18 FINAL SUMMARY")
print("=" * 80)

print(
    "Development observations used for training:",
    len(train_df)
)

print(
    "19-Oct observations used for testing:",
    len(test_df)
)

print(
    "\nPersistence MAE:",
    round(
        persistence_mae,
        6
    )
)

print(
    "RF MAE:",
    round(
        rf_mae,
        6
    )
)

print(
    "Persistence RMSE:",
    round(
        persistence_rmse,
        6
    )
)

print(
    "RF RMSE:",
    round(
        rf_rmse,
        6
    )
)

print(
    "RF overall direction accuracy:",
    round(
        overall_direction_accuracy,
        2
    ),
    "%"
)


print("\n" + "=" * 80)
print("FILES SAVED")
print("=" * 80)

print(PREDICTION_FILE)
print(THRESHOLD_FILE)
print(TIMESTAMP_FILE)
```

```python
# Exact time to expiry feature engineeringg

# ============================================================
# STEP 19
# EXACT TIME-TO-EXPIRY FEATURE ENGINEERING
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path


# ============================================================
# 1. FILE PATHS
# ============================================================

BASE_DIR = Path.home() / "Downloads"
PROCESSED_FOLDER = BASE_DIR / "data" / "processed"

DEV_FILE = (
    PROCESSED_FOLDER /
    "nifty_options_ml_5min_lagged.csv"
)

EXPIRY_FILE = (
    PROCESSED_FOLDER /
    "nifty_options_19oct_ml_lagged.csv"
)


# ============================================================
# 2. LOAD DATA
# ============================================================

dev = pd.read_csv(DEV_FILE)
expiry_day = pd.read_csv(EXPIRY_FILE)


print("=" * 80)
print("DATA LOADED")
print("=" * 80)

print("Development rows:", len(dev))
print("19-Oct rows:", len(expiry_day))


# ============================================================
# 3. CONVERT TIMESTAMPS
# ============================================================

dev["time_5m"] = pd.to_datetime(
    dev["time_5m"],
    utc=True,
    errors="coerce"
)

expiry_day["time_5m"] = pd.to_datetime(
    expiry_day["time_5m"],
    utc=True,
    errors="coerce"
)


# ============================================================
# 4. CONVERT EXPIRY DATE
# ============================================================

dev["expiry"] = pd.to_datetime(
    dev["expiry"],
    errors="coerce"
)

expiry_day["expiry"] = pd.to_datetime(
    expiry_day["expiry"],
    errors="coerce"
)


print("\nExpiry values:")

print(
    "Development:",
    dev["expiry"].unique()
)

print(
    "19-Oct:",
    expiry_day["expiry"].unique()
)


# ============================================================
# 5. CONVERT COLLECTION TIME TO INDIA TIME
# ============================================================

dev["timestamp_ist_exact"] = (
    dev["time_5m"]
    .dt.tz_convert("Asia/Kolkata")
)

expiry_day["timestamp_ist_exact"] = (
    expiry_day["time_5m"]
    .dt.tz_convert("Asia/Kolkata")
)


# ============================================================
# 6. CREATE EXPIRY TIMESTAMP
# ============================================================
#
# For this project we use 15:30 IST as the end-of-market
# reference point on expiry day.
#
# This is a practical modelling reference for remaining
# market-session time.
# ============================================================

def create_expiry_timestamp(df):

    expiry_date_string = (
        df["expiry"]
        .dt.strftime("%Y-%m-%d")
    )

    expiry_timestamp = pd.to_datetime(
        expiry_date_string + " 15:30:00"
    )

    expiry_timestamp = (
        expiry_timestamp
        .dt.tz_localize("Asia/Kolkata")
    )

    return expiry_timestamp


dev["expiry_timestamp_ist"] = (
    create_expiry_timestamp(dev)
)

expiry_day["expiry_timestamp_ist"] = (
    create_expiry_timestamp(expiry_day)
)


# ============================================================
# 7. CALCULATE EXACT TIME REMAINING
# ============================================================

dev["time_to_expiry_minutes"] = (
    (
        dev["expiry_timestamp_ist"]
        -
        dev["timestamp_ist_exact"]
    )
    .dt.total_seconds()
    / 60
)


expiry_day["time_to_expiry_minutes"] = (
    (
        expiry_day["expiry_timestamp_ist"]
        -
        expiry_day["timestamp_ist_exact"]
    )
    .dt.total_seconds()
    / 60
)


# ============================================================
# 8. HOURS TO EXPIRY
# ============================================================

dev["time_to_expiry_hours"] = (
    dev["time_to_expiry_minutes"]
    / 60
)

expiry_day["time_to_expiry_hours"] = (
    expiry_day["time_to_expiry_minutes"]
    / 60
)


# ============================================================
# 9. FRACTIONAL DAYS TO EXPIRY
# ============================================================

dev["time_to_expiry_days_exact"] = (
    dev["time_to_expiry_hours"]
    / 24
)

expiry_day["time_to_expiry_days_exact"] = (
    expiry_day["time_to_expiry_hours"]
    / 24
)


# ============================================================
# 10. FRACTIONAL YEARS TO EXPIRY
# ============================================================
#
# Useful later for option-pricing calculations.
#
# T = time remaining / 365 days
# ============================================================

dev["time_to_expiry_years"] = (
    dev["time_to_expiry_days_exact"]
    / 365
)

expiry_day["time_to_expiry_years"] = (
    expiry_day["time_to_expiry_days_exact"]
    / 365
)


# ============================================================
# 11. SAFETY CHECK
# ============================================================

print("\n" + "=" * 80)
print("NEGATIVE TIME CHECK")
print("=" * 80)

print(
    "Development negative values:",
    (
        dev["time_to_expiry_minutes"] < 0
    ).sum()
)

print(
    "19-Oct negative values:",
    (
        expiry_day["time_to_expiry_minutes"] < 0
    ).sum()
)


# ============================================================
# 12. DEVELOPMENT DATA SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("DEVELOPMENT TIME-TO-EXPIRY SUMMARY")
print("=" * 80)

print(
    dev[
        [
            "time_to_expiry_minutes",
            "time_to_expiry_hours",
            "time_to_expiry_days_exact",
            "time_to_expiry_years"
        ]
    ]
    .describe()
)


# ============================================================
# 13. 19-OCT SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("19-OCT TIME-TO-EXPIRY SUMMARY")
print("=" * 80)

print(
    expiry_day[
        [
            "time_to_expiry_minutes",
            "time_to_expiry_hours",
            "time_to_expiry_days_exact",
            "time_to_expiry_years"
        ]
    ]
    .describe()
)


# ============================================================
# 14. CHECK EACH 19-OCT TIMESTAMP
# ============================================================

time_check = (
    expiry_day[
        [
            "time_5m",
            "timestamp_ist_exact",
            "time_to_expiry_minutes",
            "time_to_expiry_hours"
        ]
    ]
    .drop_duplicates()
    .sort_values("time_5m")
)


print("\n" + "=" * 80)
print("19-OCT EXACT TIME REMAINING")
print("=" * 80)

print(
    time_check.to_string(
        index=False
    )
)


# ============================================================
# 15. COMPARE OLD VS NEW FEATURE
# ============================================================

comparison = (
    expiry_day[
        [
            "time_5m",
            "days_to_expiry",
            "time_to_expiry_minutes",
            "time_to_expiry_hours",
            "time_to_expiry_years"
        ]
    ]
    .drop_duplicates()
    .sort_values("time_5m")
)


print("\n" + "=" * 80)
print("OLD vs NEW TIME-TO-EXPIRY FEATURE")
print("=" * 80)

print(
    comparison.to_string(
        index=False
    )
)


# ============================================================
# 16. CHECK NUMBER OF UNIQUE VALUES
# ============================================================

print("\n" + "=" * 80)
print("FEATURE VARIATION CHECK")
print("=" * 80)

print(
    "19-Oct unique days_to_expiry:",
    expiry_day["days_to_expiry"].nunique()
)

print(
    "19-Oct unique exact time-to-expiry values:",
    expiry_day[
        "time_to_expiry_minutes"
    ].nunique()
)


# ============================================================
# 17. SAVE UPDATED DEVELOPMENT DATA
# ============================================================

DEV_OUTPUT = (
    PROCESSED_FOLDER /
    "nifty_options_ml_5min_lagged_exact_tte.csv"
)

dev.to_csv(
    DEV_OUTPUT,
    index=False
)


# ============================================================
# 18. SAVE UPDATED 19-OCT DATA
# ============================================================

EXPIRY_OUTPUT = (
    PROCESSED_FOLDER /
    "nifty_options_19oct_ml_lagged_exact_tte.csv"
)

expiry_day.to_csv(
    EXPIRY_OUTPUT,
    index=False
)


# ============================================================
# 19. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("STEP 19 COMPLETE")
print("=" * 80)

print(
    "Development rows:",
    len(dev)
)

print(
    "19-Oct rows:",
    len(expiry_day)
)

print(
    "\n19-Oct earliest observation:"
)

print(
    time_check.head(1).to_string(
        index=False
    )
)

print(
    "\n19-Oct latest observation:"
)

print(
    time_check.tail(1).to_string(
        index=False
    )
)

print("\nFiles saved:")

print(DEV_OUTPUT)
print(EXPIRY_OUTPUT)
```

```python
# Relative IV valuation

# ============================================================
# STEP 20
# RELATIVE IV VALUATION
#
# Goal:
# Convert our 5-minute IV forecast into:
#
# 1. Forecast direction
# 2. Relative IV valuation
# 3. Signal strength
#
# IMPORTANT:
# We keep the SAME Random Forest model and SAME 39 features.
# We do NOT use 19-Oct to tune the model.
# ============================================================


import pandas as pd
import numpy as np
from pathlib import Path

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ============================================================
# 1. FILE PATH
# ============================================================

BASE_DIR = Path.home() / "Downloads"
PROCESSED_FOLDER = BASE_DIR / "data" / "processed"

DEV_FILE = (
    PROCESSED_FOLDER /
    "nifty_options_ml_5min_lagged_exact_tte.csv"
)


# ============================================================
# 2. LOAD DEVELOPMENT DATA
# ============================================================

df = pd.read_csv(DEV_FILE)


df["time_5m"] = pd.to_datetime(
    df["time_5m"],
    utc=True,
    errors="coerce"
)


print("=" * 80)
print("STEP 20 — RELATIVE IV VALUATION")
print("=" * 80)

print("Rows:", len(df))
print("Unique timestamps:", df["time_5m"].nunique())

print(
    "Period:",
    df["time_5m"].min(),
    "to",
    df["time_5m"].max()
)


# ============================================================
# 3. SAME 39 FEATURES USED BY OUR BEST RF MODEL
# ============================================================

original_features = [

    "IV",
    "spot",
    "strike",
    "moneyness",
    "distance_from_ATM_pct",

    "LTP",
    "Volume",
    "OI",

    "Bid",
    "Ask",

    "bid_ask_spread",
    "relative_spread_pct",

    "pcr",
    "days_to_expiry",
    "option_type_code"
]


lag_features = [

    "IV_lag_5m",
    "IV_lag_10m",
    "IV_change_prev_5m",
    "IV_change_prev_10m",

    "spot_lag_5m",
    "spot_lag_10m",
    "spot_change_prev_5m",
    "spot_change_prev_10m",

    "LTP_lag_5m",
    "LTP_lag_10m",
    "LTP_change_prev_5m",
    "LTP_change_prev_10m",

    "Volume_lag_5m",
    "Volume_lag_10m",
    "Volume_change_prev_5m",
    "Volume_change_prev_10m",

    "OI_lag_5m",
    "OI_lag_10m",
    "OI_change_prev_5m",
    "OI_change_prev_10m",

    "bid_ask_spread_lag_5m",
    "bid_ask_spread_lag_10m",
    "bid_ask_spread_change_prev_5m",
    "bid_ask_spread_change_prev_10m"
]


features = original_features + lag_features

TARGET = "IV_change_5m"


print("\nNumber of model features:", len(features))


# ============================================================
# 4. FEATURE CHECK
# ============================================================

missing_columns = [

    column
    for column in features
    if column not in df.columns
]


print(
    "Missing columns:",
    missing_columns
)


if len(missing_columns) > 0:

    raise ValueError(
        "Some required model features are missing."
    )


# ============================================================
# 5. CHRONOLOGICAL 80/20 SPLIT
#
# IMPORTANT:
# We split by timestamp.
#
# We do NOT randomly split individual option rows.
# ============================================================

unique_times = (

    df["time_5m"]
    .dropna()
    .sort_values()
    .unique()
)


split_position = int(
    len(unique_times) * 0.80
)


train_times = unique_times[
    :split_position
]

test_times = unique_times[
    split_position:
]


train_df = df[
    df["time_5m"].isin(
        train_times
    )
].copy()


test_df = df[
    df["time_5m"].isin(
        test_times
    )
].copy()


print("\n" + "=" * 80)
print("CHRONOLOGICAL SPLIT")
print("=" * 80)

print(
    "Training timestamps:",
    len(train_times)
)

print(
    "Testing timestamps:",
    len(test_times)
)

print(
    "Training rows:",
    len(train_df)
)

print(
    "Testing rows:",
    len(test_df)
)


# ============================================================
# 6. CREATE X AND y
# ============================================================

X_train = train_df[
    features
].copy()

y_train = train_df[
    TARGET
].copy()


X_test = test_df[
    features
].copy()

y_test = test_df[
    TARGET
].copy()


# ============================================================
# 7. MISSING VALUE CHECK
# ============================================================

train_medians = X_train.median(
    numeric_only=True
)


X_train = X_train.fillna(
    train_medians
)

X_test = X_test.fillna(
    train_medians
)


print("\nMissing X_train values:", X_train.isna().sum().sum())
print("Missing X_test values :", X_test.isna().sum().sum())


# ============================================================
# 8. TRAIN SAME RANDOM FOREST
#
# We are keeping the previously selected architecture.
# ============================================================

rf_model = RandomForestRegressor(

    n_estimators=300,

    max_depth=12,

    min_samples_leaf=3,

    max_features="sqrt",

    random_state=42,

    n_jobs=-1
)


rf_model.fit(
    X_train,
    y_train
)


print("\nRandom Forest trained.")


# ============================================================
# 9. PREDICT 5-MINUTE IV CHANGE
# ============================================================

test_df["predicted_IV_change_5m"] = (
    rf_model.predict(
        X_test
    )
)


# ============================================================
# 10. PREDICT FUTURE IV
#
# Future IV =
# Current IV + Predicted ΔIV
# ============================================================

test_df["predicted_future_IV"] = (

    test_df["IV"]

    +

    test_df["predicted_IV_change_5m"]
)


# ============================================================
# 11. FORECAST GAP
#
# This is the difference between:
#
# predicted future IV
# and
# current market IV
#
# Because:
#
# predicted_future_IV
#     = current IV + predicted ΔIV
#
# Forecast gap is numerically equal to predicted ΔIV.
#
# We keep a separate column because it is easier to explain
# in the final dashboard/report.
# ============================================================

test_df["IV_forecast_gap"] = (

    test_df["predicted_future_IV"]

    -

    test_df["IV"]
)


# ============================================================
# 12. VALUATION THRESHOLD
# ============================================================
#
# We use 0.10 IV points as the working threshold.
#
# +0.10 or more:
# current IV is below model's expected future IV.
#
# -0.10 or less:
# current IV is above model's expected future IV.
#
# Between -0.10 and +0.10:
# neutral / fairly valued in IV terms.
# ============================================================

VALUATION_THRESHOLD = 0.10


# ============================================================
# 13. FORECAST DIRECTION
# ============================================================

test_df["IV_forecast_direction"] = np.where(

    test_df["IV_forecast_gap"]
    > VALUATION_THRESHOLD,

    "IV UP",

    np.where(

        test_df["IV_forecast_gap"]
        < -VALUATION_THRESHOLD,

        "IV DOWN",

        "FLAT"
    )
)


# ============================================================
# 14. RELATIVE IV VALUATION
# ============================================================
#
# IMPORTANT INTERPRETATION:
#
# Predicted IV rises:
# Current IV is relatively LOW compared with
# model's expected near-term IV.
#
# Predicted IV falls:
# Current IV is relatively HIGH compared with
# model's expected near-term IV.
# ============================================================

test_df["relative_IV_valuation"] = np.where(

    test_df["IV_forecast_gap"]
    > VALUATION_THRESHOLD,

    "RELATIVELY UNDERVALUED IV",

    np.where(

        test_df["IV_forecast_gap"]
        < -VALUATION_THRESHOLD,

        "RELATIVELY OVERVALUED IV",

        "FAIR / NEUTRAL"
    )
)


# ============================================================
# 15. SIGNAL STRENGTH
#
# This is NOT probability.
#
# It is simply the absolute size of the predicted IV move.
# ============================================================

test_df["signal_strength"] = (

    test_df["IV_forecast_gap"]
    .abs()
)


# ============================================================
# 16. SIMPLE SIGNAL-STRENGTH CATEGORY
# ============================================================
#
# These categories are descriptive.
#
# They are not probabilities and not trading guarantees.
# ============================================================

test_df["signal_strength_category"] = pd.cut(

    test_df["signal_strength"],

    bins=[
        -np.inf,
        0.10,
        0.25,
        0.50,
        1.00,
        np.inf
    ],

    labels=[
        "NEUTRAL",
        "LOW",
        "MODERATE",
        "STRONG",
        "VERY STRONG"
    ]
)


# ============================================================
# 17. MODEL PERFORMANCE CHECK
# ============================================================

predicted_future = (
    test_df["predicted_future_IV"]
)


actual_future = (
    test_df["IV_t_plus_5"]
)


rf_mae = mean_absolute_error(

    actual_future,

    predicted_future
)


rf_rmse = np.sqrt(

    mean_squared_error(

        actual_future,

        predicted_future
    )
)


rf_r2 = r2_score(

    actual_future,

    predicted_future
)


# Persistence benchmark

persistence_mae = mean_absolute_error(

    actual_future,

    test_df["IV"]
)


persistence_rmse = np.sqrt(

    mean_squared_error(

        actual_future,

        test_df["IV"]
    )
)


print("\n" + "=" * 80)
print("MODEL CHECK")
print("=" * 80)

print(
    "Persistence MAE:",
    round(
        persistence_mae,
        6
    )
)

print(
    "RF MAE:",
    round(
        rf_mae,
        6
    )
)

print(
    "Persistence RMSE:",
    round(
        persistence_rmse,
        6
    )
)

print(
    "RF RMSE:",
    round(
        rf_rmse,
        6
    )
)

print(
    "RF R²:",
    round(
        rf_r2,
        6
    )
)


# ============================================================
# 18. VALUATION COUNTS
# ============================================================

valuation_counts = (

    test_df[
        "relative_IV_valuation"
    ]

    .value_counts()

    .rename_axis(
        "Relative_IV_Valuation"
    )

    .reset_index(
        name="Count"
    )
)


valuation_counts[
    "Percentage"
] = (

    valuation_counts["Count"]

    /

    len(test_df)

    * 100
)


print("\n" + "=" * 80)
print("RELATIVE IV VALUATION DISTRIBUTION")
print("=" * 80)

print(

    valuation_counts
    .round(2)
    .to_string(
        index=False
    )
)


# ============================================================
# 19. FORECAST-DIRECTION DISTRIBUTION
# ============================================================

direction_counts = (

    test_df[
        "IV_forecast_direction"
    ]

    .value_counts()

    .rename_axis(
        "Forecast_Direction"
    )

    .reset_index(
        name="Count"
    )
)


direction_counts["Percentage"] = (

    direction_counts["Count"]

    /

    len(test_df)

    * 100
)


print("\n" + "=" * 80)
print("FORECAST DIRECTION DISTRIBUTION")
print("=" * 80)

print(

    direction_counts
    .round(2)
    .to_string(
        index=False
    )
)


# ============================================================
# 20. SIGNAL STRENGTH DISTRIBUTION
# ============================================================

print("\n" + "=" * 80)
print("SIGNAL STRENGTH DISTRIBUTION")
print("=" * 80)

print(

    test_df[
        "signal_strength_category"
    ]

    .value_counts(
        sort=False
    )
)


# ============================================================
# 21. SHOW STRONGEST UNDERVALUED-IV OPTIONS
# ============================================================

undervalued = (

    test_df[

        test_df[
            "relative_IV_valuation"
        ]

        ==

        "RELATIVELY UNDERVALUED IV"

    ]

    .sort_values(

        "IV_forecast_gap",

        ascending=False
    )
)


columns_to_show = [

    "time_5m",

    "option_type",

    "strike",

    "spot",

    "moneyness",

    "LTP",

    "IV",

    "predicted_future_IV",

    "IV_forecast_gap",

    "signal_strength_category",

    "relative_IV_valuation"
]


print("\n" + "=" * 80)
print("TOP RELATIVELY UNDERVALUED-IV OPTIONS")
print("=" * 80)

print(

    undervalued[
        columns_to_show
    ]

    .head(15)

    .round(4)

    .to_string(
        index=False
    )
)


# ============================================================
# 22. SHOW STRONGEST OVERVALUED-IV OPTIONS
# ============================================================

overvalued = (

    test_df[

        test_df[
            "relative_IV_valuation"
        ]

        ==

        "RELATIVELY OVERVALUED IV"

    ]

    .sort_values(

        "IV_forecast_gap",

        ascending=True
    )
)


print("\n" + "=" * 80)
print("TOP RELATIVELY OVERVALUED-IV OPTIONS")
print("=" * 80)

print(

    overvalued[
        columns_to_show
    ]

    .head(15)

    .round(4)

    .to_string(
        index=False
    )
)


# ============================================================
# 23. FAIR / NEUTRAL OPTIONS
# ============================================================

neutral = test_df[

    test_df[
        "relative_IV_valuation"
    ]

    ==

    "FAIR / NEUTRAL"
]


print("\n" + "=" * 80)
print("FAIR / NEUTRAL OPTIONS")
print("=" * 80)

print(
    "Number of neutral options:",
    len(neutral)
)


if len(neutral) > 0:

    print(

        neutral[
            columns_to_show
        ]

        .head(10)

        .round(4)

        .to_string(
            index=False
        )
    )


# ============================================================
# 24. CREATE SIMPLE DASHBOARD-FRIENDLY SIGNAL
# ============================================================
#
# IMPORTANT:
# This is NOT BUY / SELL.
#
# It describes the IV opportunity only.
# ============================================================

test_df["dashboard_signal"] = np.where(

    test_df[
        "relative_IV_valuation"
    ]

    ==

    "RELATIVELY UNDERVALUED IV",

    "LOW IV vs FORECAST",

    np.where(

        test_df[
            "relative_IV_valuation"
        ]

        ==

        "RELATIVELY OVERVALUED IV",

        "HIGH IV vs FORECAST",

        "NEUTRAL"
    )
)


# ============================================================
# 25. FINAL OUTPUT TABLE
# ============================================================

final_columns = [

    "time_5m",

    "option_type",

    "strike",

    "spot",

    "moneyness",

    "LTP",

    "Bid",

    "Ask",

    "Volume",

    "OI",

    "IV",

    "predicted_IV_change_5m",

    "predicted_future_IV",

    "IV_forecast_gap",

    "IV_forecast_direction",

    "relative_IV_valuation",

    "signal_strength",

    "signal_strength_category",

    "dashboard_signal",

    "IV_t_plus_5"
]


valuation_output = (

    test_df[
        final_columns
    ]

    .copy()
)


# ============================================================
# 26. SAVE OUTPUT
# ============================================================

OUTPUT_FILE = (

    PROCESSED_FOLDER /

    "relative_iv_valuation_results.csv"
)


valuation_output.to_csv(

    OUTPUT_FILE,

    index=False
)


print("\n" + "=" * 80)
print("STEP 20 COMPLETE")
print("=" * 80)

print(
    "Valuation threshold:",
    VALUATION_THRESHOLD
)

print(
    "Output rows:",
    len(
        valuation_output
    )
)

print(
    "\nSaved file:"
)

print(
    OUTPUT_FILE
)
```

```python
# ============================================================
# STEP 21
# IV DIRECTION CLASSIFICATION + PROBABILITIES
#
# Goal:
#
# Predict whether IV over the next 5 minutes will:
#
# DOWN
# FLAT
# UP
#
# Then obtain probability-like model scores:
#
# P(DOWN)
# P(FLAT)
# P(UP)
#
# IMPORTANT:
# Random Forest predict_proba values are model probabilities,
# but they are NOT automatically perfectly calibrated.
# We will evaluate them before using them in the dashboard.
# ============================================================


import pandas as pd
import numpy as np
from pathlib import Path

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    confusion_matrix,
    classification_report
)


# ============================================================
# 1. FILE PATH
# ============================================================

BASE_DIR = Path.home() / "Downloads"

PROCESSED_FOLDER = (
    BASE_DIR /
    "data" /
    "processed"
)


FILE = (
    PROCESSED_FOLDER /
    "nifty_options_ml_5min_lagged_exact_tte.csv"
)


# ============================================================
# 2. LOAD DATA
# ============================================================

df = pd.read_csv(FILE)


df["time_5m"] = pd.to_datetime(
    df["time_5m"],
    utc=True,
    errors="coerce"
)


print("=" * 80)
print("STEP 21 — IV DIRECTION CLASSIFICATION")
print("=" * 80)

print("Rows:", len(df))

print(
    "Unique timestamps:",
    df["time_5m"].nunique()
)


# ============================================================
# 3. DEFINE DIRECTION THRESHOLD
# ============================================================
#
# Same 0.10 IV-point threshold used in our
# relative-IV interpretation.
#
# Actual ΔIV > +0.10  -> UP
#
# Actual ΔIV < -0.10  -> DOWN
#
# Otherwise           -> FLAT
# ============================================================

DIRECTION_THRESHOLD = 0.10


df["actual_direction_class"] = np.where(

    df["IV_change_5m"]
    > DIRECTION_THRESHOLD,

    "UP",

    np.where(

        df["IV_change_5m"]
        < -DIRECTION_THRESHOLD,

        "DOWN",

        "FLAT"
    )
)


# ============================================================
# 4. CHECK CLASS DISTRIBUTION
# ============================================================

class_distribution = (

    df["actual_direction_class"]
    .value_counts()
)


class_percentage = (

    df["actual_direction_class"]
    .value_counts(
        normalize=True
    )
    * 100
)


print("\n" + "=" * 80)
print("ACTUAL DIRECTION CLASS DISTRIBUTION")
print("=" * 80)


distribution_table = pd.DataFrame({

    "Count":
        class_distribution,

    "Percentage":
        class_percentage

})


print(
    distribution_table
    .round(2)
    .to_string()
)


# ============================================================
# 5. SAME 39 FEATURES
# ============================================================

original_features = [

    "IV",

    "spot",

    "strike",

    "moneyness",

    "distance_from_ATM_pct",

    "LTP",

    "Volume",

    "OI",

    "Bid",

    "Ask",

    "bid_ask_spread",

    "relative_spread_pct",

    "pcr",

    "days_to_expiry",

    "option_type_code"
]


lag_features = [

    "IV_lag_5m",
    "IV_lag_10m",

    "IV_change_prev_5m",
    "IV_change_prev_10m",

    "spot_lag_5m",
    "spot_lag_10m",

    "spot_change_prev_5m",
    "spot_change_prev_10m",

    "LTP_lag_5m",
    "LTP_lag_10m",

    "LTP_change_prev_5m",
    "LTP_change_prev_10m",

    "Volume_lag_5m",
    "Volume_lag_10m",

    "Volume_change_prev_5m",
    "Volume_change_prev_10m",

    "OI_lag_5m",
    "OI_lag_10m",

    "OI_change_prev_5m",
    "OI_change_prev_10m",

    "bid_ask_spread_lag_5m",
    "bid_ask_spread_lag_10m",

    "bid_ask_spread_change_prev_5m",
    "bid_ask_spread_change_prev_10m"
]


features = (
    original_features
    +
    lag_features
)


print(
    "\nNumber of features:",
    len(features)
)


# ============================================================
# 6. CHRONOLOGICAL SPLIT
# ============================================================
#
# Same principle as our regression model.
#
# Earlier timestamps -> training
#
# Later timestamps -> testing
#
# Never random row split.
# ============================================================

unique_times = (

    df["time_5m"]
    .dropna()
    .sort_values()
    .unique()
)


split_position = int(

    len(unique_times)
    * 0.80
)


train_times = (
    unique_times[
        :split_position
    ]
)


test_times = (
    unique_times[
        split_position:
    ]
)


train_df = df[

    df["time_5m"]
    .isin(
        train_times
    )

].copy()


test_df = df[

    df["time_5m"]
    .isin(
        test_times
    )

].copy()


print("\n" + "=" * 80)
print("CHRONOLOGICAL SPLIT")
print("=" * 80)


print(
    "Training timestamps:",
    len(train_times)
)

print(
    "Testing timestamps:",
    len(test_times)
)

print(
    "Training rows:",
    len(train_df)
)

print(
    "Testing rows:",
    len(test_df)
)


# ============================================================
# 7. PREPARE TRAIN / TEST
# ============================================================

X_train = (
    train_df[
        features
    ].copy()
)


X_test = (
    test_df[
        features
    ].copy()
)


y_train = (
    train_df[
        "actual_direction_class"
    ].copy()
)


y_test = (
    test_df[
        "actual_direction_class"
    ].copy()
)


# ============================================================
# 8. HANDLE MISSING VALUES
# ============================================================

train_medians = (

    X_train
    .median(
        numeric_only=True
    )
)


X_train = (
    X_train
    .fillna(
        train_medians
    )
)


X_test = (
    X_test
    .fillna(
        train_medians
    )
)


print(
    "\nMissing X_train:",
    X_train.isna().sum().sum()
)

print(
    "Missing X_test:",
    X_test.isna().sum().sum()
)


# ============================================================
# 9. TRAIN RANDOM FOREST CLASSIFIER
# ============================================================
#
# class_weight="balanced_subsample"
#
# helps because UP / DOWN / FLAT may not contain
# equal numbers of observations.
# ============================================================

classifier = RandomForestClassifier(

    n_estimators=300,

    max_depth=12,

    min_samples_leaf=3,

    max_features="sqrt",

    class_weight="balanced_subsample",

    random_state=42,

    n_jobs=-1
)


classifier.fit(

    X_train,

    y_train
)


print(
    "\nRandom Forest classifier trained."
)


# ============================================================
# 10. CLASS PREDICTIONS
# ============================================================

predicted_class = (

    classifier.predict(
        X_test
    )
)


test_df[
    "predicted_direction_class"
] = predicted_class


# ============================================================
# 11. CLASS PROBABILITIES
# ============================================================

probabilities = (

    classifier.predict_proba(
        X_test
    )
)


print(
    "\nClassifier classes:",
    classifier.classes_
)


# ============================================================
# 12. CREATE PROBABILITY COLUMNS
# ============================================================

for index, class_name in enumerate(
    classifier.classes_
):

    test_df[
        f"prob_{class_name}"
    ] = (
        probabilities[:, index]
    )


# ============================================================
# 13. MAKE SURE ALL 3 COLUMNS EXIST
# ============================================================

for class_name in [

    "DOWN",
    "FLAT",
    "UP"

]:

    column_name = (
        f"prob_{class_name}"
    )

    if column_name not in test_df.columns:

        test_df[
            column_name
        ] = 0.0


# ============================================================
# 14. CONVERT TO PERCENTAGES
# ============================================================

test_df[
    "prob_DOWN_pct"
] = (

    test_df["prob_DOWN"]
    * 100
)


test_df[
    "prob_FLAT_pct"
] = (

    test_df["prob_FLAT"]
    * 100
)


test_df[
    "prob_UP_pct"
] = (

    test_df["prob_UP"]
    * 100
)


# ============================================================
# 15. MODEL CONFIDENCE
#
# Highest predicted class probability.
#
# Example:
#
# DOWN = 15%
# FLAT = 20%
# UP   = 65%
#
# Confidence = 65%
#
# IMPORTANT:
# This does not yet mean calibrated real-world probability.
# ============================================================

test_df[
    "model_confidence_pct"
] = (

    test_df[
        [
            "prob_DOWN",
            "prob_FLAT",
            "prob_UP"
        ]
    ]

    .max(axis=1)

    * 100
)


# ============================================================
# 16. OVERALL ACCURACY
# ============================================================

accuracy = accuracy_score(

    y_test,

    predicted_class
)


balanced_accuracy = (
    balanced_accuracy_score(

        y_test,

        predicted_class
    )
)


print("\n" + "=" * 80)
print("CLASSIFICATION PERFORMANCE")
print("=" * 80)


print(
    "Overall accuracy:",
    round(
        accuracy * 100,
        2
    ),
    "%"
)


print(
    "Balanced accuracy:",
    round(
        balanced_accuracy * 100,
        2
    ),
    "%"
)


# ============================================================
# 17. CONFUSION MATRIX
# ============================================================

labels = [

    "DOWN",
    "FLAT",
    "UP"
]


cm = confusion_matrix(

    y_test,

    predicted_class,

    labels=labels
)


cm_df = pd.DataFrame(

    cm,

    index=[
        "Actual DOWN",
        "Actual FLAT",
        "Actual UP"
    ],

    columns=[
        "Predicted DOWN",
        "Predicted FLAT",
        "Predicted UP"
    ]
)


print("\n" + "=" * 80)
print("CONFUSION MATRIX")
print("=" * 80)


print(
    cm_df.to_string()
)


# ============================================================
# 18. PRECISION / RECALL / F1
# ============================================================

print("\n" + "=" * 80)
print("CLASSIFICATION REPORT")
print("=" * 80)


print(

    classification_report(

        y_test,

        predicted_class,

        labels=labels,

        zero_division=0
    )
)


# ============================================================
# 19. PROBABILITY SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("PROBABILITY SUMMARY")
print("=" * 80)


print(

    test_df[
        [
            "prob_DOWN_pct",
            "prob_FLAT_pct",
            "prob_UP_pct",
            "model_confidence_pct"
        ]
    ]

    .describe()

    .round(2)
)


# ============================================================
# 20. SHOW HIGHEST UP-PROBABILITY OPTIONS
# ============================================================

display_columns = [

    "time_5m",

    "option_type",

    "strike",

    "spot",

    "IV",

    "IV_change_5m",

    "actual_direction_class",

    "predicted_direction_class",

    "prob_DOWN_pct",

    "prob_FLAT_pct",

    "prob_UP_pct",

    "model_confidence_pct"
]


highest_up = (

    test_df
    .sort_values(

        "prob_UP_pct",

        ascending=False
    )
    [
        display_columns
    ]
    .head(15)
)


print("\n" + "=" * 80)
print("HIGHEST MODEL UP-PROBABILITY OPTIONS")
print("=" * 80)


print(

    highest_up
    .round(2)
    .to_string(
        index=False
    )
)


# ============================================================
# 21. SHOW HIGHEST DOWN-PROBABILITY OPTIONS
# ============================================================

highest_down = (

    test_df
    .sort_values(

        "prob_DOWN_pct",

        ascending=False
    )
    [
        display_columns
    ]
    .head(15)
)


print("\n" + "=" * 80)
print("HIGHEST MODEL DOWN-PROBABILITY OPTIONS")
print("=" * 80)


print(

    highest_down
    .round(2)
    .to_string(
        index=False
    )
)


# ============================================================
# 22. SAVE RESULTS
# ============================================================

OUTPUT_FILE = (

    PROCESSED_FOLDER /

    "iv_direction_probability_results.csv"
)


test_df.to_csv(

    OUTPUT_FILE,

    index=False
)


print("\n" + "=" * 80)
print("STEP 21 COMPLETE")
print("=" * 80)


print(
    "Direction threshold:",
    DIRECTION_THRESHOLD
)


print(
    "Test rows:",
    len(test_df)
)


print(
    "Overall accuracy:",
    round(
        accuracy * 100,
        2
    ),
    "%"
)


print(
    "Balanced accuracy:",
    round(
        balanced_accuracy * 100,
        2
    ),
    "%"
)


print(
    "\nSaved:"
)

print(
    OUTPUT_FILE
)
```

```python
# ============================================================
# STEP 22
# BLACK-SCHOLES STRESS TESTING
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path
from scipy.stats import norm


# ============================================================
# 1. FILE PATH
# ============================================================

BASE_DIR = Path.home() / "Downloads"
PROCESSED_FOLDER = BASE_DIR / "data" / "processed"

FILE = (
    PROCESSED_FOLDER /
    "nifty_options_ml_5min_lagged_exact_tte.csv"
)


# ============================================================
# 2. LOAD DATA
# ============================================================

df = pd.read_csv(FILE)

df["time_5m"] = pd.to_datetime(
    df["time_5m"],
    utc=True,
    errors="coerce"
)


print("=" * 80)
print("STEP 22 — OPTION STRESS TESTING")
print("=" * 80)

print("Rows:", len(df))
print("Timestamps:", df["time_5m"].nunique())


# ============================================================
# 3. BLACK-SCHOLES FUNCTIONS
# ============================================================

def black_scholes_call(S, K, T, r, sigma):

    if T <= 0 or sigma <= 0:
        return max(S - K, 0)

    d1 = (
        np.log(S / K)
        +
        (r + 0.5 * sigma**2) * T
    ) / (
        sigma * np.sqrt(T)
    )

    d2 = (
        d1
        -
        sigma * np.sqrt(T)
    )

    call_price = (
        S * norm.cdf(d1)
        -
        K * np.exp(-r * T) * norm.cdf(d2)
    )

    return call_price


def black_scholes_put(S, K, T, r, sigma):

    if T <= 0 or sigma <= 0:
        return max(K - S, 0)

    d1 = (
        np.log(S / K)
        +
        (r + 0.5 * sigma**2) * T
    ) / (
        sigma * np.sqrt(T)
    )

    d2 = (
        d1
        -
        sigma * np.sqrt(T)
    )

    put_price = (
        K * np.exp(-r * T) * norm.cdf(-d2)
        -
        S * norm.cdf(-d1)
    )

    return put_price


def option_price(S, K, T, r, sigma, option_type):

    if option_type == "CE":

        return black_scholes_call(
            S,
            K,
            T,
            r,
            sigma
        )

    elif option_type == "PE":

        return black_scholes_put(
            S,
            K,
            T,
            r,
            sigma
        )

    else:

        return np.nan


# ============================================================
# 4. CHOOSE ONE REPRESENTATIVE OPTION
#
# We choose the option closest to ATM
# from the latest development timestamp.
# ============================================================

latest_time = df["time_5m"].max()

latest_data = df[
    df["time_5m"] == latest_time
].copy()


latest_data["atm_distance"] = (

    latest_data["moneyness"]
    - 1
).abs()


selected_option = (

    latest_data
    .sort_values("atm_distance")
    .iloc[0]
)


print("\n" + "=" * 80)
print("SELECTED OPTION")
print("=" * 80)

print(
    "Timestamp:",
    selected_option["time_5m"]
)

print(
    "Option Type:",
    selected_option["option_type"]
)

print(
    "Strike:",
    selected_option["strike"]
)

print(
    "Spot:",
    selected_option["spot"]
)

print(
    "Market LTP:",
    selected_option["LTP"]
)

print(
    "IV:",
    selected_option["IV"]
)

print(
    "Moneyness:",
    selected_option["moneyness"]
)

print(
    "Time to expiry hours:",
    selected_option["time_to_expiry_hours"]
)


# ============================================================
# 5. BASE INPUTS
# ============================================================

S = float(
    selected_option["spot"]
)

K = float(
    selected_option["strike"]
)

IV = float(
    selected_option["IV"]
)


# IV in dataset is percentage
# Example:
# 15 means 15%
#
# Black-Scholes needs decimal:
# 15% = 0.15

sigma = (
    IV / 100
)


T = float(
    selected_option["time_to_expiry_years"]
)


option_type = (
    selected_option["option_type"]
)


# ============================================================
# 6. RISK-FREE RATE
#
# Working assumption:
# 6% annual risk-free rate
#
# We will clearly report this as an assumption.
# ============================================================

risk_free_rate = 0.06


# ============================================================
# 7. BASE BLACK-SCHOLES VALUE
# ============================================================

base_bs_price = option_price(

    S,
    K,
    T,
    risk_free_rate,
    sigma,
    option_type
)


market_price = float(
    selected_option["LTP"]
)


print("\n" + "=" * 80)
print("BASE CASE")
print("=" * 80)

print(
    "Spot:",
    round(S, 2)
)

print(
    "Strike:",
    round(K, 2)
)

print(
    "IV:",
    round(IV, 2),
    "%"
)

print(
    "Time to expiry:",
    round(
        selected_option[
            "time_to_expiry_hours"
        ],
        2
    ),
    "hours"
)

print(
    "Risk-free rate:",
    risk_free_rate * 100,
    "%"
)

print(
    "Market LTP:",
    round(
        market_price,
        4
    )
)

print(
    "Black-Scholes theoretical value:",
    round(
        base_bs_price,
        4
    )
)


# ============================================================
# 8. STRESS TEST 1 — SPOT PRICE SHOCK
# ============================================================

spot_shocks = [

    -0.02,
    -0.01,
    0.00,
    0.01,
    0.02
]


spot_results = []


for shock in spot_shocks:

    stressed_spot = (
        S * (1 + shock)
    )


    stressed_price = option_price(

        stressed_spot,

        K,

        T,

        risk_free_rate,

        sigma,

        option_type
    )


    price_change = (
        stressed_price
        -
        base_bs_price
    )


    price_change_pct = (

        price_change
        /
        base_bs_price
        * 100

        if base_bs_price != 0

        else np.nan
    )


    spot_results.append({

        "Spot_Shock_%":
            shock * 100,

        "Stressed_Spot":
            stressed_spot,

        "Option_Value":
            stressed_price,

        "Option_Value_Change":
            price_change,

        "Option_Value_Change_%":
            price_change_pct
    })


spot_df = pd.DataFrame(
    spot_results
)


print("\n" + "=" * 80)
print("STRESS TEST 1 — SPOT SHOCK")
print("=" * 80)

print(
    spot_df
    .round(4)
    .to_string(
        index=False
    )
)


# ============================================================
# 9. STRESS TEST 2 — IV SHOCK
# ============================================================
#
# Example:
#
# IV = 15%
#
# +20% IV shock means:
#
# 15% × 1.20 = 18%
#
# It does NOT mean 15% + 20 percentage points.
# ============================================================

iv_shocks = [

    -0.20,
    -0.10,
    0.00,
    0.10,
    0.20
]


iv_results = []


for shock in iv_shocks:

    stressed_sigma = (

        sigma
        *
        (1 + shock)
    )


    stressed_price = option_price(

        S,

        K,

        T,

        risk_free_rate,

        stressed_sigma,

        option_type
    )


    price_change = (
        stressed_price
        -
        base_bs_price
    )


    price_change_pct = (

        price_change
        /
        base_bs_price
        * 100

        if base_bs_price != 0

        else np.nan
    )


    iv_results.append({

        "IV_Shock_%":
            shock * 100,

        "Original_IV_%":
            IV,

        "Stressed_IV_%":
            stressed_sigma * 100,

        "Option_Value":
            stressed_price,

        "Option_Value_Change":
            price_change,

        "Option_Value_Change_%":
            price_change_pct
    })


iv_df = pd.DataFrame(
    iv_results
)


print("\n" + "=" * 80)
print("STRESS TEST 2 — IV SHOCK")
print("=" * 80)

print(
    iv_df
    .round(4)
    .to_string(
        index=False
    )
)


# ============================================================
# 10. STRESS TEST 3 — TIME DECAY
# ============================================================
#
# Reduce remaining time by:
#
# 5 minutes
# 30 minutes
# 60 minutes
#
# We never allow negative time.
# ============================================================

time_decay_minutes = [

    0,
    5,
    30,
    60
]


time_results = []


current_minutes = float(

    selected_option[
        "time_to_expiry_minutes"
    ]
)


for decay in time_decay_minutes:

    remaining_minutes = max(

        current_minutes
        -
        decay,

        0
    )


    stressed_T = (

        remaining_minutes
        /
        60
        /
        24
        /
        365
    )


    stressed_price = option_price(

        S,

        K,

        stressed_T,

        risk_free_rate,

        sigma,

        option_type
    )


    price_change = (

        stressed_price
        -
        base_bs_price
    )


    time_results.append({

        "Time_Decay_Minutes":
            decay,

        "Remaining_Minutes":
            remaining_minutes,

        "Option_Value":
            stressed_price,

        "Option_Value_Change":
            price_change
    })


time_df = pd.DataFrame(
    time_results
)


print("\n" + "=" * 80)
print("STRESS TEST 3 — TIME DECAY")
print("=" * 80)

print(
    time_df
    .round(4)
    .to_string(
        index=False
    )
)


# ============================================================
# 11. STRESS TEST 4 — COMBINED SCENARIOS
# ============================================================

scenarios = [

    {
        "Scenario":
            "Base",

        "Spot_Shock":
            0.00,

        "IV_Shock":
            0.00,

        "Time_Decay_Min":
            0
    },


    {
        "Scenario":
            "Moderate Bullish",

        "Spot_Shock":
            0.01,

        "IV_Shock":
            0.10,

        "Time_Decay_Min":
            5
    },


    {
        "Scenario":
            "Moderate Bearish",

        "Spot_Shock":
            -0.01,

        "IV_Shock":
            0.10,

        "Time_Decay_Min":
            5
    },


    {
        "Scenario":
            "Severe Spot Fall + IV Spike",

        "Spot_Shock":
            -0.02,

        "IV_Shock":
            0.20,

        "Time_Decay_Min":
            5
    },


    {
        "Scenario":
            "Spot Rise + IV Crush",

        "Spot_Shock":
            0.02,

        "IV_Shock":
            -0.20,

        "Time_Decay_Min":
            5
    },


    {
        "Scenario":
            "IV Crush Only",

        "Spot_Shock":
            0.00,

        "IV_Shock":
            -0.20,

        "Time_Decay_Min":
            5
    },


    {
        "Scenario":
            "IV Spike Only",

        "Spot_Shock":
            0.00,

        "IV_Shock":
            0.20,

        "Time_Decay_Min":
            5
    }
]


scenario_results = []


for scenario in scenarios:

    stressed_spot = (

        S
        *
        (
            1
            +
            scenario[
                "Spot_Shock"
            ]
        )
    )


    stressed_sigma = (

        sigma
        *
        (
            1
            +
            scenario[
                "IV_Shock"
            ]
        )
    )


    remaining_minutes = max(

        current_minutes
        -
        scenario[
            "Time_Decay_Min"
        ],

        0
    )


    stressed_T = (

        remaining_minutes
        /
        60
        /
        24
        /
        365
    )


    stressed_price = option_price(

        stressed_spot,

        K,

        stressed_T,

        risk_free_rate,

        stressed_sigma,

        option_type
    )


    price_change = (

        stressed_price
        -
        base_bs_price
    )


    price_change_pct = (

        price_change
        /
        base_bs_price
        * 100

        if base_bs_price != 0

        else np.nan
    )


    scenario_results.append({

        "Scenario":
            scenario[
                "Scenario"
            ],

        "Spot":
            stressed_spot,

        "IV_%":
            stressed_sigma * 100,

        "Remaining_Minutes":
            remaining_minutes,

        "Option_Value":
            stressed_price,

        "Value_Change":
            price_change,

        "Value_Change_%":
            price_change_pct
    })


scenario_df = pd.DataFrame(
    scenario_results
)


print("\n" + "=" * 80)
print("STRESS TEST 4 — COMBINED SCENARIOS")
print("=" * 80)

print(
    scenario_df
    .round(4)
    .to_string(
        index=False
    )
)


# ============================================================
# 12. FIND WORST SCENARIO
# ============================================================

worst_scenario = (

    scenario_df
    .sort_values(
        "Value_Change"
    )
    .iloc[0]
)


best_scenario = (

    scenario_df
    .sort_values(
        "Value_Change",
        ascending=False
    )
    .iloc[0]
)


print("\n" + "=" * 80)
print("WORST / BEST STRESS RESULT")
print("=" * 80)


print(
    "Worst scenario:",
    worst_scenario[
        "Scenario"
    ]
)

print(
    "Option value change:",
    round(
        worst_scenario[
            "Value_Change"
        ],
        4
    )
)

print(
    "Percentage change:",
    round(
        worst_scenario[
            "Value_Change_%"
        ],
        2
    ),
    "%"
)


print(
    "\nBest scenario:",
    best_scenario[
        "Scenario"
    ]
)

print(
    "Option value change:",
    round(
        best_scenario[
            "Value_Change"
        ],
        4
    )
)

print(
    "Percentage change:",
    round(
        best_scenario[
            "Value_Change_%"
        ],
        2
    ),
    "%"
)


# ============================================================
# 13. SAVE RESULTS
# ============================================================

SPOT_FILE = (
    PROCESSED_FOLDER /
    "stress_test_spot.csv"
)

IV_FILE = (
    PROCESSED_FOLDER /
    "stress_test_iv.csv"
)

TIME_FILE = (
    PROCESSED_FOLDER /
    "stress_test_time_decay.csv"
)

SCENARIO_FILE = (
    PROCESSED_FOLDER /
    "stress_test_combined_scenarios.csv"
)


spot_df.to_csv(
    SPOT_FILE,
    index=False
)

iv_df.to_csv(
    IV_FILE,
    index=False
)

time_df.to_csv(
    TIME_FILE,
    index=False
)

scenario_df.to_csv(
    SCENARIO_FILE,
    index=False
)


print("\n" + "=" * 80)
print("STEP 22 COMPLETE")
print("=" * 80)


print(
    "Selected option:",
    option_type,
    "Strike",
    K
)

print(
    "Market LTP:",
    round(
        market_price,
        4
    )
)

print(
    "Base Black-Scholes value:",
    round(
        base_bs_price,
        4
    )
)


print("\nSaved files:")

print(SPOT_FILE)
print(IV_FILE)
print(TIME_FILE)
print(SCENARIO_FILE)
```

```python
# ============================================================
# STEP 23
# FINAL PROJECT VISUALIZATIONS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error


# ============================================================
# 1. PATHS
# ============================================================

BASE_DIR = Path.home() / "Downloads"

PROCESSED_FOLDER = (
    BASE_DIR /
    "data" /
    "processed"
)

CHART_FOLDER = (
    PROCESSED_FOLDER /
    "final_charts"
)

CHART_FOLDER.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 2. LOAD MAIN DEVELOPMENT DATA
# ============================================================

DEV_FILE = (
    PROCESSED_FOLDER /
    "nifty_options_ml_5min_lagged_exact_tte.csv"
)

df = pd.read_csv(DEV_FILE)

df["time_5m"] = pd.to_datetime(
    df["time_5m"],
    utc=True,
    errors="coerce"
)


print("=" * 80)
print("STEP 23 — FINAL VISUALIZATIONS")
print("=" * 80)

print("Development rows:", len(df))
print("Timestamps:", df["time_5m"].nunique())


# ============================================================
# 3. SAME 39 FEATURES
# ============================================================

original_features = [

    "IV",
    "spot",
    "strike",
    "moneyness",
    "distance_from_ATM_pct",

    "LTP",
    "Volume",
    "OI",

    "Bid",
    "Ask",

    "bid_ask_spread",
    "relative_spread_pct",

    "pcr",
    "days_to_expiry",
    "option_type_code"
]


lag_features = [

    "IV_lag_5m",
    "IV_lag_10m",

    "IV_change_prev_5m",
    "IV_change_prev_10m",

    "spot_lag_5m",
    "spot_lag_10m",

    "spot_change_prev_5m",
    "spot_change_prev_10m",

    "LTP_lag_5m",
    "LTP_lag_10m",

    "LTP_change_prev_5m",
    "LTP_change_prev_10m",

    "Volume_lag_5m",
    "Volume_lag_10m",

    "Volume_change_prev_5m",
    "Volume_change_prev_10m",

    "OI_lag_5m",
    "OI_lag_10m",

    "OI_change_prev_5m",
    "OI_change_prev_10m",

    "bid_ask_spread_lag_5m",
    "bid_ask_spread_lag_10m",

    "bid_ask_spread_change_prev_5m",
    "bid_ask_spread_change_prev_10m"
]


features = (
    original_features
    +
    lag_features
)


# ============================================================
# 4. SAME CHRONOLOGICAL 80/20 SPLIT
# ============================================================

unique_times = (

    df["time_5m"]
    .dropna()
    .sort_values()
    .unique()
)


split_position = int(
    len(unique_times) * 0.80
)


train_times = (
    unique_times[
        :split_position
    ]
)


test_times = (
    unique_times[
        split_position:
    ]
)


train_df = df[
    df["time_5m"].isin(
        train_times
    )
].copy()


test_df = df[
    df["time_5m"].isin(
        test_times
    )
].copy()


print("\nTraining rows:", len(train_df))
print("Testing rows :", len(test_df))


# ============================================================
# 5. PREPARE DATA
# ============================================================

X_train = (
    train_df[
        features
    ].copy()
)

X_test = (
    test_df[
        features
    ].copy()
)

y_train = (
    train_df[
        "IV_change_5m"
    ].copy()
)

y_test = (
    test_df[
        "IV_change_5m"
    ].copy()
)


train_medians = (
    X_train
    .median(
        numeric_only=True
    )
)


X_train = X_train.fillna(
    train_medians
)

X_test = X_test.fillna(
    train_medians
)


# ============================================================
# 6. TRAIN FINAL RF USED FOR DEVELOPMENT EVALUATION
# ============================================================

rf = RandomForestRegressor(

    n_estimators=300,

    max_depth=12,

    min_samples_leaf=3,

    max_features="sqrt",

    random_state=42,

    n_jobs=-1
)


rf.fit(
    X_train,
    y_train
)


# ============================================================
# 7. CREATE PREDICTIONS
# ============================================================

test_df["predicted_delta_IV"] = (
    rf.predict(
        X_test
    )
)


test_df["predicted_future_IV"] = (

    test_df["IV"]

    +

    test_df["predicted_delta_IV"]
)


# Persistence prediction

test_df["persistence_future_IV"] = (
    test_df["IV"]
)


# ============================================================
# CHART 1
# ACTUAL vs PREDICTED FUTURE IV
# ============================================================

plt.figure(
    figsize=(8, 6)
)


plt.scatter(

    test_df["IV_t_plus_5"],

    test_df["predicted_future_IV"],

    alpha=0.5
)


minimum = min(

    test_df["IV_t_plus_5"].min(),

    test_df["predicted_future_IV"].min()
)


maximum = max(

    test_df["IV_t_plus_5"].max(),

    test_df["predicted_future_IV"].max()
)


plt.plot(

    [minimum, maximum],

    [minimum, maximum],

    linestyle="--"
)


plt.xlabel(
    "Actual Future IV"
)

plt.ylabel(
    "Predicted Future IV"
)

plt.title(
    "Actual vs Predicted 5-Minute Future IV"
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()


chart1 = (
    CHART_FOLDER /
    "01_actual_vs_predicted_future_iv.png"
)

plt.savefig(
    chart1,
    dpi=300
)

plt.show()


# ============================================================
# CHART 2
# ACTUAL vs PREDICTED DELTA IV
# ============================================================

plt.figure(
    figsize=(8, 6)
)


plt.scatter(

    test_df["IV_change_5m"],

    test_df["predicted_delta_IV"],

    alpha=0.5
)


minimum = min(

    test_df["IV_change_5m"].min(),

    test_df["predicted_delta_IV"].min()
)


maximum = max(

    test_df["IV_change_5m"].max(),

    test_df["predicted_delta_IV"].max()
)


plt.plot(

    [minimum, maximum],

    [minimum, maximum],

    linestyle="--"
)


plt.xlabel(
    "Actual ΔIV"
)

plt.ylabel(
    "Predicted ΔIV"
)

plt.title(
    "Actual vs Predicted 5-Minute ΔIV"
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()


chart2 = (
    CHART_FOLDER /
    "02_actual_vs_predicted_delta_iv.png"
)

plt.savefig(
    chart2,
    dpi=300
)

plt.show()


# ============================================================
# 8. ERROR METRICS
# ============================================================

rf_mae = mean_absolute_error(

    test_df["IV_t_plus_5"],

    test_df["predicted_future_IV"]
)


rf_rmse = np.sqrt(

    mean_squared_error(

        test_df["IV_t_plus_5"],

        test_df["predicted_future_IV"]
    )
)


persistence_mae = mean_absolute_error(

    test_df["IV_t_plus_5"],

    test_df["persistence_future_IV"]
)


persistence_rmse = np.sqrt(

    mean_squared_error(

        test_df["IV_t_plus_5"],

        test_df["persistence_future_IV"]
    )
)


# ============================================================
# CHART 3
# RF vs PERSISTENCE
# ============================================================

metrics_df = pd.DataFrame({

    "Model": [
        "Persistence",
        "Random Forest"
    ],

    "MAE": [
        persistence_mae,
        rf_mae
    ],

    "RMSE": [
        persistence_rmse,
        rf_rmse
    ]
})


x = np.arange(
    len(metrics_df["Model"])
)


width = 0.35


plt.figure(
    figsize=(8, 6)
)


plt.bar(

    x - width / 2,

    metrics_df["MAE"],

    width,

    label="MAE"
)


plt.bar(

    x + width / 2,

    metrics_df["RMSE"],

    width,

    label="RMSE"
)


plt.xticks(

    x,

    metrics_df["Model"]
)


plt.ylabel(
    "Error"
)

plt.title(
    "Persistence vs Random Forest Forecast Error"
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()


chart3 = (
    CHART_FOLDER /
    "03_rf_vs_persistence.png"
)

plt.savefig(
    chart3,
    dpi=300
)

plt.show()


# ============================================================
# 9. FEATURE IMPORTANCE
# ============================================================

importance_df = pd.DataFrame({

    "Feature":
        features,

    "Importance":
        rf.feature_importances_
})


importance_df = (

    importance_df

    .sort_values(

        "Importance",

        ascending=False
    )

    .reset_index(
        drop=True
    )
)


top_features = (
    importance_df
    .head(15)
)


# ============================================================
# CHART 4
# FEATURE IMPORTANCE
# ============================================================

plt.figure(
    figsize=(9, 7)
)


plt.barh(

    top_features["Feature"][::-1],

    top_features["Importance"][::-1]
)


plt.xlabel(
    "Feature Importance"
)

plt.ylabel(
    "Feature"
)

plt.title(
    "Top 15 Random Forest Features"
)

plt.grid(
    axis="x",
    alpha=0.3
)

plt.tight_layout()


chart4 = (
    CHART_FOLDER /
    "04_feature_importance.png"
)

plt.savefig(
    chart4,
    dpi=300
)

plt.show()


print("\nTop 15 features:")

print(

    top_features
    .round(6)
    .to_string(
        index=False
    )
)


# ============================================================
# 10. LOAD RELATIVE VALUATION RESULTS
# ============================================================

VALUATION_FILE = (

    PROCESSED_FOLDER /

    "relative_iv_valuation_results.csv"
)


valuation_df = pd.read_csv(
    VALUATION_FILE
)


valuation_counts = (

    valuation_df[
        "relative_IV_valuation"
    ]

    .value_counts()
)


# ============================================================
# CHART 5
# RELATIVE IV VALUATION DISTRIBUTION
# ============================================================

plt.figure(
    figsize=(9, 6)
)


plt.bar(

    valuation_counts.index,

    valuation_counts.values
)


plt.ylabel(
    "Number of Observations"
)

plt.xlabel(
    "Relative IV Status"
)

plt.title(
    "Relative IV Valuation Distribution"
)

plt.xticks(
    rotation=20
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()


chart5 = (
    CHART_FOLDER /
    "05_relative_iv_valuation_distribution.png"
)

plt.savefig(
    chart5,
    dpi=300
)

plt.show()


# ============================================================
# 11. LOAD SPOT STRESS RESULTS
# ============================================================

SPOT_STRESS_FILE = (

    PROCESSED_FOLDER /

    "stress_test_spot.csv"
)


spot_df = pd.read_csv(
    SPOT_STRESS_FILE
)


# ============================================================
# CHART 6
# SPOT STRESS SENSITIVITY
# ============================================================

plt.figure(
    figsize=(8, 6)
)


plt.plot(

    spot_df[
        "Spot_Shock_%"
    ],

    spot_df[
        "Option_Value"
    ],

    marker="o"
)


plt.xlabel(
    "Spot Price Shock (%)"
)

plt.ylabel(
    "Theoretical Option Value"
)

plt.title(
    "Option Value Sensitivity to NIFTY Spot Shock"
)

plt.axvline(
    x=0,
    linestyle="--"
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()


chart6 = (
    CHART_FOLDER /
    "06_spot_stress_sensitivity.png"
)

plt.savefig(
    chart6,
    dpi=300
)

plt.show()


# ============================================================
# 12. LOAD IV STRESS RESULTS
# ============================================================

IV_STRESS_FILE = (

    PROCESSED_FOLDER /

    "stress_test_iv.csv"
)


iv_df = pd.read_csv(
    IV_STRESS_FILE
)


# ============================================================
# CHART 7
# IV STRESS SENSITIVITY
# ============================================================

plt.figure(
    figsize=(8, 6)
)


plt.plot(

    iv_df[
        "Stressed_IV_%"
    ],

    iv_df[
        "Option_Value"
    ],

    marker="o"
)


plt.xlabel(
    "Stressed Implied Volatility (%)"
)

plt.ylabel(
    "Theoretical Option Value"
)

plt.title(
    "Option Value Sensitivity to Implied Volatility"
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()


chart7 = (
    CHART_FOLDER /
    "07_iv_stress_sensitivity.png"
)

plt.savefig(
    chart7,
    dpi=300
)

plt.show()


# ============================================================
# 13. LOAD 19-OCT ROBUSTNESS RESULTS
# ============================================================

EXPIRY_FILE = (

    PROCESSED_FOLDER /

    "rf_19oct_timestamp_performance.csv"
)


expiry_df = pd.read_csv(
    EXPIRY_FILE
)


expiry_df["time_5m"] = pd.to_datetime(

    expiry_df["time_5m"],

    utc=True,

    errors="coerce"
)


# Convert UTC to IST

expiry_df["time_IST"] = (

    expiry_df["time_5m"]

    .dt.tz_convert(
        "Asia/Kolkata"
    )
)


# ============================================================
# CHART 8
# EXPIRY-DAY DETERIORATION
# ============================================================

plt.figure(
    figsize=(10, 6)
)


plt.plot(

    expiry_df[
        "time_IST"
    ],

    expiry_df[
        "Actual_Mean_Abs_Delta"
    ],

    marker="o",

    label="Actual Mean |ΔIV|"
)


plt.plot(

    expiry_df[
        "time_IST"
    ],

    expiry_df[
        "RF_MAE"
    ],

    marker="o",

    label="RF MAE"
)


plt.xlabel(
    "Time on 19-Oct (IST)"
)

plt.ylabel(
    "IV Points"
)

plt.title(
    "Expiry-Day IV Instability and Model Error"
)

plt.xticks(
    rotation=45
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()


chart8 = (
    CHART_FOLDER /
    "08_expiry_day_deterioration.png"
)

plt.savefig(
    chart8,
    dpi=300
)

plt.show()


# ============================================================
# 14. SAVE FEATURE IMPORTANCE TABLE
# ============================================================

IMPORTANCE_FILE = (

    PROCESSED_FOLDER /

    "final_feature_importance.csv"
)


importance_df.to_csv(

    IMPORTANCE_FILE,

    index=False
)


# ============================================================
# 15. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 80)

print("STEP 23 COMPLETE")

print("=" * 80)


print("\nRandom Forest Performance:")

print(
    "RF MAE:",
    round(
        rf_mae,
        6
    )
)

print(
    "RF RMSE:",
    round(
        rf_rmse,
        6
    )
)

print(
    "Persistence MAE:",
    round(
        persistence_mae,
        6
    )
)

print(
    "Persistence RMSE:",
    round(
        persistence_rmse,
        6
    )
)


mae_improvement = (

    (
        persistence_mae
        -
        rf_mae
    )

    /

    persistence_mae

    * 100
)


rmse_improvement = (

    (
        persistence_rmse
        -
        rf_rmse
    )

    /

    persistence_rmse

    * 100
)


print(
    "\nMAE improvement:",
    round(
        mae_improvement,
        2
    ),
    "%"
)


print(
    "RMSE improvement:",
    round(
        rmse_improvement,
        2
    ),
    "%"
)


print("\nCharts saved in:")

print(
    CHART_FOLDER
)


print("\nSaved charts:")

print(chart1)
print(chart2)
print(chart3)
print(chart4)
print(chart5)
print(chart6)
print(chart7)
print(chart8)


print("\nFeature importance table:")

print(
    IMPORTANCE_FILE
)
```

```python
# ============================================================
# STEP 24A
# TRAIN AND SAVE FINAL RANDOM FOREST MODEL
# ============================================================

import pandas as pd
import numpy as np
import joblib

from pathlib import Path
from sklearn.ensemble import RandomForestRegressor


# ============================================================
# 1. PATHS
# ============================================================

BASE_DIR = Path.home() / "Downloads"

PROCESSED_FOLDER = (
    BASE_DIR /
    "data" /
    "processed"
)

MODEL_FOLDER = (
    BASE_DIR /
    "models"
)

MODEL_FOLDER.mkdir(
    parents=True,
    exist_ok=True
)


DATA_FILE = (
    PROCESSED_FOLDER /
    "nifty_options_ml_5min_lagged_exact_tte.csv"
)


# ============================================================
# 2. LOAD DEVELOPMENT DATA
# ============================================================

df = pd.read_csv(DATA_FILE)


print("=" * 80)
print("STEP 24A — SAVE FINAL MODEL")
print("=" * 80)

print(
    "Development rows:",
    len(df)
)


# ============================================================
# 3. FINAL 39 FEATURES
# ============================================================

original_features = [

    "IV",
    "spot",
    "strike",
    "moneyness",
    "distance_from_ATM_pct",

    "LTP",
    "Volume",
    "OI",

    "Bid",
    "Ask",

    "bid_ask_spread",
    "relative_spread_pct",

    "pcr",
    "days_to_expiry",
    "option_type_code"
]


lag_features = [

    "IV_lag_5m",
    "IV_lag_10m",

    "IV_change_prev_5m",
    "IV_change_prev_10m",

    "spot_lag_5m",
    "spot_lag_10m",

    "spot_change_prev_5m",
    "spot_change_prev_10m",

    "LTP_lag_5m",
    "LTP_lag_10m",

    "LTP_change_prev_5m",
    "LTP_change_prev_10m",

    "Volume_lag_5m",
    "Volume_lag_10m",

    "Volume_change_prev_5m",
    "Volume_change_prev_10m",

    "OI_lag_5m",
    "OI_lag_10m",

    "OI_change_prev_5m",
    "OI_change_prev_10m",

    "bid_ask_spread_lag_5m",
    "bid_ask_spread_lag_10m",

    "bid_ask_spread_change_prev_5m",
    "bid_ask_spread_change_prev_10m"
]


features = (
    original_features
    +
    lag_features
)


TARGET = "IV_change_5m"


print(
    "Number of features:",
    len(features)
)


# ============================================================
# 4. CHECK FEATURES
# ============================================================

missing_columns = [

    col

    for col in features

    if col not in df.columns
]


print(
    "Missing features:",
    missing_columns
)


if len(missing_columns) > 0:

    raise ValueError(
        "Required model features are missing."
    )


# ============================================================
# 5. PREPARE FINAL TRAINING DATA
# ============================================================
#
# IMPORTANT:
#
# Model selection has already been completed.
#
# Now we train the selected RF architecture using
# ALL available development data.
#
# 19-Oct is NOT included.
# ============================================================

X = (
    df[
        features
    ].copy()
)


y = (
    df[
        TARGET
    ].copy()
)


# ============================================================
# 6. TRAINING MEDIANS
# ============================================================

feature_medians = (

    X
    .median(
        numeric_only=True
    )
)


X = X.fillna(
    feature_medians
)


print(
    "Missing values after filling:",
    X.isna().sum().sum()
)


# ============================================================
# 7. FINAL RANDOM FOREST
# ============================================================

final_rf = RandomForestRegressor(

    n_estimators=300,

    max_depth=12,

    min_samples_leaf=3,

    max_features="sqrt",

    random_state=42,

    n_jobs=-1
)


final_rf.fit(
    X,
    y
)


print(
    "\nFinal Random Forest trained."
)


print(
    "Training rows:",
    len(X)
)


# ============================================================
# 8. SAVE MODEL
# ============================================================

MODEL_FILE = (

    MODEL_FOLDER /

    "nifty_iv_random_forest.joblib"
)


joblib.dump(

    final_rf,

    MODEL_FILE
)


# ============================================================
# 9. SAVE FEATURE LIST
# ============================================================

FEATURE_FILE = (

    MODEL_FOLDER /

    "nifty_iv_model_features.joblib"
)


joblib.dump(

    features,

    FEATURE_FILE
)


# ============================================================
# 10. SAVE TRAINING MEDIANS
# ============================================================

MEDIAN_FILE = (

    MODEL_FOLDER /

    "nifty_iv_feature_medians.joblib"
)


joblib.dump(

    feature_medians,

    MEDIAN_FILE
)


# ============================================================
# 11. VERIFY SAVED MODEL
# ============================================================

loaded_model = joblib.load(
    MODEL_FILE
)


test_prediction = loaded_model.predict(

    X.head(5)
)


print("\n" + "=" * 80)
print("MODEL SAVE CHECK")
print("=" * 80)


print(
    "Test predictions:"
)

print(
    test_prediction
)


print(
    "\nModel type:"
)

print(
    type(
        loaded_model
    )
)


# ============================================================
# 12. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("STEP 24A COMPLETE")
print("=" * 80)


print(
    "Final model training rows:",
    len(X)
)


print(
    "Features:",
    len(features)
)


print(
    "\nSaved model:"
)

print(
    MODEL_FILE
)


print(
    "\nSaved feature list:"
)

print(
    FEATURE_FILE
)


print(
    "\nSaved training medians:"
)

print(
    MEDIAN_FILE
)
```
